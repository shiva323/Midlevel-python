## ۱. تابع range: کارخانه‌ی اعداد

تابع <code dir="ltr">range</code> دنباله‌ای از اعداد صحیح می‌سازد و سه شکل دارد:

| شکل | معنی | نمونه | اعداد تولیدشده |
|:---|:---|:---|:---|
| <code dir="ltr">range(stop)</code> | از ۰ تا **قبل از** <code dir="ltr">stop</code> | <code dir="ltr">range(5)</code> | <code dir="ltr">0, 1, 2, 3, 4</code> |
| <code dir="ltr">range(start, stop)</code> | از <code dir="ltr">start</code> تا **قبل از** <code dir="ltr">stop</code> | <code dir="ltr">range(2, 6)</code> | <code dir="ltr">2, 3, 4, 5</code> |
| <code dir="ltr">range(start, stop, step)</code> | با گام <code dir="ltr">step</code> | <code dir="ltr">range(0, 20, 5)</code> | <code dir="ltr">0, 5, 10, 15</code> |

⚠️ درست مثل برش لیست، عدد <code dir="ltr">stop</code> **هیچ‌وقت** داخل خروجی نیست.

خروجی <code dir="ltr">range</code> خودش لیست نیست، اما با <code dir="ltr">list(...)</code> می‌توانیم ببینیم چه اعدادی می‌سازد:

</div>

````python
print(list(range(5)))
print(list(range(2, 6)))
print(list(range(0, 20, 5)))
print(list(range(10, 0, -1)))   # a negative step counts down
````

### 🧠 علوم اعصاب: زمان هر نمونه‌ی EEG

دستگاه EEG با **نرخ نمونه‌برداری** (sampling rate) مشخصی کار می‌کند. اگر نرخ نمونه‌برداری ۲۵۰ هرتز باشد، یعنی دستگاه **در هر ثانیه ۲۵۰ نمونه** ثبت می‌کند. پس فاصله‌ی زمانی بین دو نمونه ۱ تقسیم بر ۲۵۰ ثانیه است؛ یعنی ۴ میلی‌ثانیه.

زمان نمونه‌ی شماره‌ی i برابر است با:

$$t_i = \frac{i}{f_s}$$

</div>

````python
fs = 250   # sampling rate (Hz)

for i in range(6):
    t_ms = i / fs * 1000
    print("Sample", i, "was recorded at", t_ms, "ms")
````

## ۳. حلقه روی اندیس‌ها

### 🧠 علوم اعصاب: فاصله‌ی بین اسپایک‌ها (ISI)

وقتی یک نورون شلیک می‌کند، زمان هر **اسپایک** ثبت می‌شود. به فاصله‌ی زمانی بین دو اسپایک پشت سر هم، **فاصله‌ی بین‌اسپایکی** (Inter-Spike Interval یا ISI) می‌گویند. این فاصله به دانشمندان کمک می‌کند بفهمند نورون **منظم** شلیک می‌کند یا **رگباری** (bursting).

اگر لیست زمان اسپایک‌ها n عضو داشته باشد، تعداد فاصله‌ها n منهای ۱ است. پس حلقه باید روی <code dir="ltr">range(len(spikes) - 1)</code> بچرخد؛ وگرنه <code dir="ltr">spikes[i + 1]</code> در دور آخر از لیست بیرون می‌زند!

</div>

````python
spike_times = [12, 30, 47, 52, 55, 90, 140]   # milliseconds

isi = []
for i in range(len(spike_times) - 1):
    gap = spike_times[i + 1] - spike_times[i]
    isi.append(gap)

print("Inter-spike intervals:", isi)
print("Shortest interval:", min(isi), "ms")
print("Mean interval:", round(sum(isi) / len(isi), 2), "ms")
````

### ⚠️ خطای یکی کم یا یکی زیاد (Off-by-one)

یکی از رایج‌ترین باگ‌های دنیای برنامه‌نویسی این است که حلقه **یک دور بیشتر یا کمتر** از حد لازم بچرخد. همیشه از خودت بپرس:

- اولین اندیس چیست؟ (معمولاً صفر)
- آخرین اندیس چیست؟ (طول لیست منهای ۱)
- آیا در بدنه‌ی حلقه به <code dir="ltr">i + 1</code> دسترسی دارم؟ اگر بله، حلقه باید یک دور **کمتر** بچرخد.

سلول زیر را اجرا کن تا ببینی اگر حواست نباشد چه اتفاقی می‌افتد:

</div>

````python
spike_times = [12, 30, 47]
for i in range(len(spike_times)):          # wrong: should be len(...) - 1
    print(spike_times[i + 1] - spike_times[i])
````

## ۴. تابع enumerate: اندیس و مقدار با هم

نوشتن <code dir="ltr">range(len(...))</code> و بعد <code dir="ltr">my_list[i]</code> کمی شلوغ است. پایتون راه زیباتری دارد: تابع <code dir="ltr">enumerate</code> در هر دور **دو چیز** به ما می‌دهد؛ **اندیس** و **مقدار**.

</div>

````python
lobes = ["frontal", "parietal", "temporal", "occipital"]

for i, lobe in enumerate(lobes):
    print(i, lobe)
````

### 💻 علوم کامپیوتر: تبدیل دودویی به دهدهی

در دستگاه **دودویی** (باینری)، ارزش هر بیت به **جایگاهش** بستگی دارد. از سمت راست، ارزش جایگاه‌ها به ترتیب ۱، ۲، ۴، ۸، ۱۶ و ... است؛ یعنی توان‌های ۲.

مثلاً عدد دودویی 1011 یعنی:

$$1 \times 8 + 0 \times 4 + 1 \times 2 + 1 \times 1 = 11$$

اگر لیست بیت‌ها را **برعکس** کنیم، اندیس هر بیت دقیقاً همان توان ۲ است. پس <code dir="ltr">enumerate</code> کار را خیلی راحت می‌کند!

</div>

````python
bits = [1, 0, 1, 1]          # the binary number 1011

value = 0
for position, bit in enumerate(bits[::-1]):
    value = value + bit * 2 ** position
    print("position", position, ": bit", bit, "x weight", 2 ** position)

print("Decimal value:", value)
````

````python
# The byte from session 1: [0, 1, 0, 0, 0, 0, 0, 1]
byte = [0, 1, 0, 0, 0, 0, 0, 1]

value = 0
for position, bit in enumerate(byte[::-1]):
    value = value + bit * 2 ** position

print("Value:", value)
print("Matching character:", chr(value))   # chr turns a number into a character
````

### 📜 داستان تاریخ: جان فون نویمان، حافظه‌ی شماره‌دار و مغز

**جان فون نویمان** (John von Neumann، ۱۹۰۳ تا ۱۹۵۷)، ریاضی‌دان مجارستانی‌ـ‌آمریکایی، یکی از باهوش‌ترین دانشمندان قرن بیستم بود. او در سال ۱۹۴۵ گزارشی به نام **«پیش‌نویس نخست گزارشی درباره‌ی EDVAC»** نوشت و در آن ایده‌ی **کامپیوتر با برنامه‌ی ذخیره‌شده** (stored-program computer) را شرح داد:

- **برنامه** و **داده** هر دو در یک **حافظه** نگه‌داری می‌شوند.
- حافظه از خانه‌های **شماره‌دار** ساخته شده است. به شماره‌ی هر خانه **آدرس** می‌گویند؛ چیزی شبیه اندیس لیست!
- پردازنده دستورها را **یکی‌یکی و به ترتیب** از حافظه می‌خواند و اجرا می‌کند.

تقریباً همه‌ی کامپیوترها، گوشی‌ها و حتی ساعت‌های هوشمند امروزی بر اساس همین طرح ساخته شده‌اند که به آن **معماری فون نویمان** می‌گویند.

🧠 فون نویمان به **مغز** هم علاقه‌ی زیادی داشت. آخرین کتابش، **«کامپیوتر و مغز»** (The Computer and the Brain)، که پس از مرگش در سال ۱۹۵۸ منتشر شد، شباهت‌ها و تفاوت‌های مغز و کامپیوتر را بررسی می‌کند. او متوجه شده بود که مغز، برخلاف کامپیوترها، **بسیاری از کارها را هم‌زمان** (به صورت موازی) انجام می‌دهد.

</div>

message = [

### 📝 خلاصه‌ی جلسه

- تابع <code dir="ltr">range</code> با یک، دو یا سه عدد دنباله‌ای از اعداد می‌سازد؛ عدد <code dir="ltr">stop</code> هیچ‌وقت داخل خروجی نیست.
- حلقه‌ی <code dir="ltr">for i in range(n):</code> یعنی «این کار را n بار انجام بده».
- حلقه‌ی <code dir="ltr">for i in range(len(L)):</code> روی **اندیس‌ها** می‌چرخد و به ما اجازه می‌دهد لیست را تغییر دهیم.
- برای دسترسی به <code dir="ltr">L[i + 1]</code> باید از <code dir="ltr">range(len(L) - 1)</code> استفاده کنیم.
- حلقه‌ی <code dir="ltr">for i, x in enumerate(L, start=1):</code> اندیس و مقدار را با هم می‌دهد.
