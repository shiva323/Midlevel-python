## ۲. آماده کردن بوم نقاشی

سلول زیر یک **تابع کمکی** به نام <code dir="ltr">new_canvas</code> می‌سازد. هر بار که آن را صدا بزنی، یک بوم تازه و یک لاک‌پشت آماده تحویل می‌گیری. **این سلول را یک بار اجرا کن.**

</div>

````python
import turtle

def new_canvas(title="Python Turtle", speed=0, bg="white"):
    """یک بوم تازه آماده می‌کند و یک لاک‌پشت برمی‌گرداند."""
    turtle.TurtleScreen._RUNNING = True   # اجازه می‌دهد در Jupyter دوباره نقاشی کنیم
    screen = turtle.Screen()
    screen.clearscreen()
    screen.title(title)
    screen.bgcolor(bg)
    t = turtle.Turtle()
    t.shape("turtle")
    t.speed(speed)                        # 1 = آهسته، 10 = تند، 0 = تندترین
    return t

print("✅ آماده‌ایم! حالا new_canvas() را صدا بزن.")
````

## ۳. دستورهای پایه

صفحه‌ی turtle یک **دستگاه مختصات** است. لاک‌پشت در **مرکز**، یعنی نقطه‌ی <code dir="ltr">(0, 0)</code>، شروع می‌کند و رو به **راست** (شرق) است.

| دستور | کار |
|:---|:---|
| <code dir="ltr">t.forward(100)</code> | ۱۰۰ واحد جلو برو |
| <code dir="ltr">t.backward(50)</code> | ۵۰ واحد عقب برو |
| <code dir="ltr">t.left(90)</code> | ۹۰ درجه به چپ بچرخ |
| <code dir="ltr">t.right(45)</code> | ۴۵ درجه به راست بچرخ |
| <code dir="ltr">t.penup()</code> | قلم را بردار (بدون کشیدن خط حرکت کن) |
| <code dir="ltr">t.pendown()</code> | قلم را بگذار |
| <code dir="ltr">t.goto(x, y)</code> | مستقیم به نقطه‌ی <code dir="ltr">(x, y)</code> برو |
| <code dir="ltr">t.setheading(90)</code> | رو به جهت ۹۰ درجه (بالا) بایست |
| <code dir="ltr">t.color("red")</code> | رنگ قلم را قرمز کن |
| <code dir="ltr">t.pensize(5)</code> | ضخامت قلم را ۵ کن |
| <code dir="ltr">t.circle(50)</code> | دایره‌ای به شعاع ۵۰ بکش |
| <code dir="ltr">t.dot(20, "blue")</code> | یک نقطه‌ی آبی به قطر ۲۰ بگذار |
| <code dir="ltr">t.begin_fill()</code> / <code dir="ltr">t.end_fill()</code> | شکل بین این دو را رنگ کن |
| <code dir="ltr">t.write("Hi")</code> | متن بنویس |
| <code dir="ltr">t.hideturtle()</code> | لاک‌پشت را پنهان کن |
| <code dir="ltr">turtle.done()</code> | نقاشی تمام است؛ پنجره را باز نگه دار |

## ۵. چندضلعی‌های منتظم: یک فرمول برای همه

لاک‌پشت برای اینکه به جای اولش برگردد، باید روی هم رفته **۳۶۰ درجه** بچرخد. پس در یک چندضلعی منتظم با <code dir="ltr">n</code> ضلع، در هر گوشه باید <code dir="ltr">360 / n</code> درجه بچرخد.

| شکل | <code dir="ltr">n</code> | زاویه‌ی چرخش |
|:---|:---:|:---:|
| مثلث | ۳ | ۱۲۰° |
| مربع | ۴ | ۹۰° |
| شش‌ضلعی | ۶ | ۶۰° |
| هشت‌ضلعی | ۸ | ۴۵° |

این بار روی یک **لیست** از تعداد ضلع‌ها و یک **لیست** از رنگ‌ها حلقه می‌زنیم:

</div>

````python
t = new_canvas("Polygons", speed=0)
t.pensize(3)

sides_list = [3, 4, 5, 6, 8]
colors = ["#e53935", "#fb8c00", "#43a047", "#1e88e5", "#8e24aa"]

for i, n in enumerate(sides_list):
    t.color(colors[i])
    angle = 360 / n
    for side in range(n):
        t.forward(80)
        t.left(angle)

turtle.done()
````

## ۶. ستاره ⭐

برای ستاره‌ی پنج‌پر، در هر نوک **۱۴۴ درجه** می‌چرخیم. (پنج بار ۱۴۴ درجه می‌شود ۷۲۰ درجه؛ یعنی لاک‌پشت **دو دور کامل** می‌چرخد.)

</div>

````python
t = new_canvas("Star", speed=4, bg="#0d1b2a")
t.color("gold", "yellow")

t.begin_fill()
for point in range(5):
    t.forward(200)
    t.right(144)
t.end_fill()

t.hideturtle()
turtle.done()
````

## ۷. مارپیچ: به‌روزرسانی متغیر در عمل 🌀

در هر دور، طول حرکت را کمی **بیشتر** می‌کنیم. شرط <code dir="ltr">while</code> تضمین می‌کند که حلقه بالاخره تمام شود، چون <code dir="ltr">length</code> هر بار **بزرگ‌تر** می‌شود و به ۳۰۰ می‌رسد.

</div>

````python
t = new_canvas("Spiral", speed=0, bg="black")
colors = ["#ff1744", "#ffea00", "#00e676", "#2979ff", "#d500f9"]

length = 5
i = 0
while length < 300:
    t.color(colors[i % len(colors)])   # % باعث می‌شود رنگ‌ها چرخشی تکرار شوند
    t.forward(length)
    t.left(91)
    length += 3                        # ⚠️ اگر این خط را پاک کنی، حلقه بی‌نهایت می‌شود!
    i += 1

turtle.done()
````

### 💡 ترفند <code dir="ltr">i % len(colors)</code>

وقتی <code dir="ltr">i</code> از تعداد رنگ‌ها بیشتر می‌شود، <code dir="ltr">i % 5</code> دوباره از <code dir="ltr">0</code> شروع می‌کند: <code dir="ltr">0, 1, 2, 3, 4, 0, 1, 2, ...</code>. این‌طوری هیچ‌وقت خطای <code dir="ltr">IndexError</code> نمی‌گیریم.

## ۸. حلقه‌ی تو در تو: گل 🌸

حلقه‌ی **داخلی** یک مربع می‌کشد. حلقه‌ی **بیرونی** بعد از هر مربع، لاک‌پشت را ۱۰ درجه می‌چرخاند و این کار را ۳۶ بار تکرار می‌کند (۳۶ × ۱۰ = ۳۶۰).

</div>

````python
t = new_canvas("Flower", speed=0)
petal_colors = ["#ec407a", "#ab47bc", "#7e57c2", "#5c6bc0"]

for i in range(36):
    t.color(petal_colors[i % len(petal_colors)])
    for side in range(4):
        t.forward(120)
        t.left(90)
    t.right(10)

t.hideturtle()
turtle.done()
````

## ۹. 🧠 پروژه‌ی کوچک: یک نورون بکش!

یک نورون سه بخش اصلی دارد:

1. **جسم سلولی** (soma): بخش گرد مرکزی که هسته در آن است.
2. **دندریت‌ها** (dendrites): شاخه‌هایی که پیام‌ها را از نورون‌های دیگر **دریافت** می‌کنند.
3. **آکسون** (axon): رشته‌ای بلند که پیام را **می‌فرستد**. آکسون معمولاً با لایه‌ای از **میلین** پوشیده شده که سرعت پیام را بالا می‌برد. در انتهای آن **پایانه‌های آکسون** قرار دارند.

</div>

````python
t = new_canvas("A Neuron", speed=0, bg="#fff8f0")
t.hideturtle()

SOMA_X, SOMA_Y, SOMA_R = -150, 0, 35

# ۱) جسم سلولی: یک دایره‌ی رنگ‌شده
t.penup()
t.goto(SOMA_X, SOMA_Y - SOMA_R)
t.setheading(0)
t.pendown()
t.color("#6a1b9a", "#ce93d8")
t.begin_fill()
t.circle(SOMA_R)
t.end_fill()

# ۲) دندریت‌ها: هر دندریت دو شاخه‌ی کوچک دارد (حلقه‌ی تو در تو!)
dendrite_angles = [100, 140, 180, 220, 260]
t.color("#6a1b9a")
for angle in dendrite_angles:
    t.penup()
    t.goto(SOMA_X, SOMA_Y)
    t.setheading(angle)
    t.forward(SOMA_R)
    t.pendown()
    t.pensize(5)
    t.forward(50)
    t.pensize(2)
    for turn in [-30, 30]:
        t.left(turn)
        t.forward(35)
        t.backward(35)
        t.right(turn)

# ۳) آکسون با غلاف میلین: قطعه‌های یکی در میان زرد و بنفش
t.penup()
t.goto(SOMA_X + SOMA_R, SOMA_Y)
t.setheading(0)
t.pendown()
t.pensize(9)
for segment in range(6):
    if segment % 2 == 0:
        t.color("#fbc02d")      # میلین
    else:
        t.color("#6a1b9a")      # گره رانویه
    t.forward(45)

# ۴) پایانه‌های آکسون
t.pensize(3)
t.color("#6a1b9a")
for turn in [-40, -15, 15, 40]:
    t.left(turn)
    t.forward(40)
    t.dot(14, "#e53935")
    t.backward(40)
    t.right(turn)

# ۵) برچسب‌ها (انگلیسی، چون پنجره‌ی turtle همیشه حروف فارسی را درست نشان نمی‌دهد)
labels = [["Dendrites", -260, 80], ["Soma", -150, -70], ["Axon (myelin)", 0, 20], ["Terminals", 190, -70]]
t.color("#263238")
for text, x, y in labels:
    t.penup()
    t.goto(x, y)
    t.write(text, align="center", font=("Arial", 12, "bold"))

turtle.done()
````

## ۱۰. 🧠 یک سیگنال EEG را از روی لیست بکش!

حالا مهم‌ترین مهارت یک دانشمند داده را تمرین می‌کنیم: **نمایش داده**. لیستی از نمونه‌های یک سیگنال می‌سازیم و لاک‌پشت را از نقطه‌ای به نقطه‌ی دیگر می‌بریم:

- محور **افقی** (<code dir="ltr">x</code>): شماره‌ی نمونه (زمان).
- محور **عمودی** (<code dir="ltr">y</code>): ولتاژ.
- نمونه‌هایی که از **آستانه** بزرگ‌ترند (احتمالاً آرتیفکت) را با **نقطه‌ی قرمز** علامت می‌زنیم.

</div>

````python
import math

# ساختن یک سیگنال: موج آلفای ۱۰ هرتزی با نرخ نمونه‌برداری ۲۵۰ هرتز
fs = 250
signal = []
for i in range(200):
    value = 30 * math.sin(2 * math.pi * 10 * i / fs)
    signal.append(value)

# دو آرتیفکت پلک‌زدن اضافه می‌کنیم
signal[60] = 120
signal[61] = 95
signal[140] = -110

THRESHOLD = 80
X_START = -300
X_STEP = 3

t = new_canvas("EEG signal", speed=0, bg="#fafafa")
t.hideturtle()

# محور زمان
t.color("#b0bec5")
t.penup()
t.goto(X_START, 0)
t.pendown()
t.goto(X_START + X_STEP * len(signal), 0)

# خود سیگنال
t.color("#1565c0")
t.pensize(2)
t.penup()
artifacts = 0
for i, v in enumerate(signal):
    x = X_START + i * X_STEP
    t.goto(x, v)
    t.pendown()                  # از نقطه‌ی دوم به بعد خط کشیده می‌شود
    if abs(v) > THRESHOLD:
        t.dot(10, "#e53935")
        artifacts += 1

t.penup()
t.goto(0, -170)
t.color("#263238")
t.write("Artifacts: " + str(artifacts), align="center", font=("Arial", 14, "bold"))
turtle.done()
````

## ۱۱. 🎲 قدم‌زدن تصادفی با مجموعه

لاک‌پشت در هر گام **به‌طور تصادفی** یکی از چهار جهت را انتخاب می‌کند. می‌خواهیم بدانیم چند گام طول می‌کشد تا **۶۰ خانه‌ی متفاوت** را ببیند. برای نگه‌داشتن خانه‌های دیده‌شده، **مجموعه** بهترین انتخاب است؛ چون تکرار ندارد و <code dir="ltr">in</code> روی آن خیلی سریع است!

این برنامه **همه‌ی** مفهوم‌های دوره را کنار هم دارد: لیست، حلقه‌ی while، به‌روزرسانی متغیر، شرط ترکیبی، اعتبارسنجی، <code dir="ltr">continue</code>، شمارنده‌ی ایمنی و مجموعه.

🧠 **پیوند با علوم اعصاب:** مدل‌های «قدم‌زدن تصادفی» برای توصیف حرکت **مولکول‌ها در مایع بین سلول‌ها** و حتی **ولتاژ غشای نورون** قبل از شلیک به کار می‌روند.

</div>

````python
import random

STEP = 20
LIMIT = 160                     # لاک‌پشت نباید از این مرز بیرون برود
TARGET = 60                     # تعداد خانه‌های متفاوتی که می‌خواهیم ببینیم
MAX_STEPS = 5000                # 🛡️ شمارنده‌ی ایمنی

moves = [[STEP, 0], [-STEP, 0], [0, STEP], [0, -STEP]]

t = new_canvas("Random walk", speed=0)
t.color("#00897b")

x, y = 0, 0
visited = {(x, y)}              # (x, y) یک «تاپل» است: جفتی تغییرناپذیر که می‌تواند عضو مجموعه شود
steps = 0

while len(visited) < TARGET and steps < MAX_STEPS:
    steps += 1
    dx, dy = random.choice(moves)
    nx, ny = x + dx, y + dy

    # اعتبارسنجی: آیا خانه‌ی بعدی داخل مرز است؟
    if not (-LIMIT <= nx <= LIMIT and -LIMIT <= ny <= LIMIT):
        continue

    t.goto(nx, ny)
    x, y = nx, ny

    if (x, y) in visited:
        t.dot(6, "#b0bec5")      # خانه‌ی تکراری
    else:
        t.dot(10, "#00897b")     # خانه‌ی تازه
        visited.add((x, y))

print("تعداد گام‌ها:", steps)
print("تعداد خانه‌های متفاوت:", len(visited))
turtle.done()
````

trials = [
    [320, 90, 310],
    [300, 1600, 290],
    [305, 298, 312],
    [280, 275, 260],
]
winner = None
for i in range(len(names)):
    perfect = True
    for rt in trials[i]:
        if not (100 <= rt <= 1500):
            perfect = False
            break
    if perfect:
        winner = names[i]
        break
print("اولین شرکت‌کننده‌ی بی‌نقص:", winner)   # پ
