## ۲. آماده کردن بوم نقاشی

سلول زیر یک **تابع کمکی** به نام <code dir="ltr">new_canvas</code> می‌سازد. هر بار که آن را صدا بزنی، یک بوم تازه و یک لاک‌پشت آماده تحویل می‌گیری. **این سلول را یک بار اجرا کن.**

</div>

````python
import turtle

def new_canvas(title="Python Turtle", speed=0, bg="white"):
    """Prepare a fresh canvas and return a turtle."""
    turtle.TurtleScreen._RUNNING = True   # lets us draw again in Jupyter
    screen = turtle.Screen()
    screen.clearscreen()
    screen.title(title)
    screen.bgcolor(bg)
    t = turtle.Turtle()
    t.shape("turtle")
    t.speed(speed)                        # 1 = slow, 10 = fast, 0 = fastest
    return t

print("Ready! Now call new_canvas().")
````

## ۵. چندضلعی‌های منتظم: یک فرمول برای همه

لاک‌پشت برای اینکه به جای اولش برگردد، باید روی هم رفته **۳۶۰ درجه** بچرخد. پس در یک چندضلعی منتظم با n ضلع، باید در هر گوشه ۳۶۰ تقسیم بر n درجه بچرخد.

| شکل | تعداد ضلع | زاویه‌ی چرخش |
|:---|:---:|:---:|
| مثلث | ۳ | ۱۲۰ درجه |
| مربع | ۴ | ۹۰ درجه |
| شش‌ضلعی | ۶ | ۶۰ درجه |
| هشت‌ضلعی | ۸ | ۴۵ درجه |

این بار روی یک **لیست** از تعداد ضلع‌ها و یک **لیست** از رنگ‌ها حلقه می‌زنیم:

</div>

````python
t = new_canvas("Polygons", speed=0)
t.pensize(3)

sides_list = [3, 4, 5, 6, 8]
colors = ["#e53935", "#fb8c00", "#43a047", "#1e88e5", "#8e24aa"]

for i, n in enumerate(sides_list):
    t.color(colors[i])
    angle = 360 / n
    for side in range(n):
        t.forward(80)
        t.left(angle)

turtle.done()
````

برای ستاره‌ی پنج‌پر، در هر نوک **۱۴۴ درجه** می‌چرخیم. پنج بار ۱۴۴ درجه می‌شود ۷۲۰ درجه؛ یعنی لاک‌پشت **دو دور کامل** می‌چرخد.

## ۷. مارپیچ: به‌روزرسانی متغیر در عمل 🌀

در هر دور، طول حرکت را کمی **بیشتر** می‌کنیم. شرط <code dir="ltr">while</code> تضمین می‌کند که حلقه بالاخره تمام شود؛ چون <code dir="ltr">length</code> هر بار **بزرگ‌تر** می‌شود و به ۳۰۰ می‌رسد.

</div>

````python
t = new_canvas("Spiral", speed=0, bg="black")
colors = ["#ff1744", "#ffea00", "#00e676", "#2979ff", "#d500f9"]

length = 5
i = 0
while length < 300:
    t.color(colors[i % len(colors)])   # % makes the colors repeat in a cycle
    t.forward(length)
    t.left(91)
    length += 3                        # remove this line and the loop never ends!
    i += 1

turtle.done()
````

## ۸. حلقه‌ی تو در تو: گل 🌸

- حلقه‌ی **داخلی** یک مربع می‌کشد.
- حلقه‌ی **بیرونی** بعد از هر مربع، لاک‌پشت را ۱۰ درجه می‌چرخاند و این کار را ۳۶ بار تکرار می‌کند؛ چون ۳۶ × ۱۰ = ۳۶۰.

````python
t = new_canvas("Flower", speed=0)
petal_colors = ["#ec407a", "#ab47bc", "#7e57c2", "#5c6bc0"]

for i in range(36):
    t.color(petal_colors[i % len(petal_colors)])
    for side in range(4):
        t.forward(120)
        t.left(90)
    t.right(10)

t.hideturtle()
turtle.done()
````

## ۹. 🧠 پروژه‌ی کوچک: یک نورون بکش!

یک نورون سه بخش اصلی دارد:

1. **جسم سلولی** (soma): بخش گرد مرکزی که هسته در آن است.
2. **دندریت‌ها** (dendrites): شاخه‌هایی که پیام‌ها را از نورون‌های دیگر **دریافت** می‌کنند.
3. **آکسون** (axon): رشته‌ای بلند که پیام را **می‌فرستد**. آکسون معمولاً با لایه‌ای از **میلین** پوشیده شده که سرعت پیام را بالا می‌برد. در انتهای آن **پایانه‌های آکسون** قرار دارند.

</div>

````python
t = new_canvas("A Neuron", speed=0, bg="#fff8f0")
t.hideturtle()

SOMA_X, SOMA_Y, SOMA_R = -150, 0, 35

# 1) Cell body: a filled circle
t.penup()
t.goto(SOMA_X, SOMA_Y - SOMA_R)
t.setheading(0)
t.pendown()
t.color("#6a1b9a", "#ce93d8")
t.begin_fill()
t.circle(SOMA_R)
t.end_fill()

# 2) Dendrites: each dendrite has two small branches (a nested loop!)
dendrite_angles = [100, 140, 180, 220, 260]
t.color("#6a1b9a")
for angle in dendrite_angles:
    t.penup()
    t.goto(SOMA_X, SOMA_Y)
    t.setheading(angle)
    t.forward(SOMA_R)
    t.pendown()
    t.pensize(5)
    t.forward(50)
    t.pensize(2)
    for turn in [-30, 30]:
        t.left(turn)
        t.forward(35)
        t.backward(35)
        t.right(turn)

# 3) Axon with a myelin sheath: alternating yellow and purple segments
t.penup()
t.goto(SOMA_X + SOMA_R, SOMA_Y)
t.setheading(0)
t.pendown()
t.pensize(9)
for segment in range(6):
    if segment % 2 == 0:
        t.color("#fbc02d")      # myelin
    else:
        t.color("#6a1b9a")      # node of Ranvier
    t.forward(45)

# 4) Axon terminals
t.pensize(3)
t.color("#6a1b9a")
for turn in [-40, -15, 15, 40]:
    t.left(turn)
    t.forward(40)
    t.dot(14, "#e53935")
    t.backward(40)
    t.right(turn)

# 5) Labels
labels = [["Dendrites", -260, 80], ["Soma", -150, -70], ["Axon (myelin)", 0, 20], ["Terminals", 190, -70]]
t.color("#263238")
for text, x, y in labels:
    t.penup()
    t.goto(x, y)
    t.write(text, align="center", font=("Arial", 12, "bold"))

turtle.done()
````

## ۱۰. 🧠 یک سیگنال EEG را از روی لیست بکش!

حالا مهم‌ترین مهارت یک دانشمند داده را تمرین می‌کنیم: **نمایش داده**. لیستی از نمونه‌های یک سیگنال می‌سازیم و لاک‌پشت را از نقطه‌ای به نقطه‌ی دیگر می‌بریم:

- محور **افقی** (x) شماره‌ی نمونه، یعنی زمان، است.
- محور **عمودی** (y) ولتاژ است.
- نمونه‌هایی را که از **آستانه** بزرگ‌ترند (احتمالاً آرتیفکت) با **نقطه‌ی قرمز** علامت می‌زنیم.

</div>

````python
import math

# Build a signal: a 10 Hz alpha wave sampled at 250 Hz
fs = 250
signal = []
for i in range(200):
    value = 30 * math.sin(2 * math.pi * 10 * i / fs)
    signal.append(value)

# Add two eye-blink artifacts
signal[60] = 120
signal[61] = 95
signal[140] = -110

THRESHOLD = 80
X_START = -300
X_STEP = 3

t = new_canvas("EEG signal", speed=0, bg="#fafafa")
t.hideturtle()

# Time axis
t.color("#b0bec5")
t.penup()
t.goto(X_START, 0)
t.pendown()
t.goto(X_START + X_STEP * len(signal), 0)

# The signal itself
t.color("#1565c0")
t.pensize(2)
t.penup()
artifacts = 0
for i, v in enumerate(signal):
    x = X_START + i * X_STEP
    t.goto(x, v)
    t.pendown()                  # lines are drawn from the second point on
    if abs(v) > THRESHOLD:
        t.dot(10, "#e53935")
        artifacts += 1

t.penup()
t.goto(0, -170)
t.color("#263238")
t.write("Artifacts: " + str(artifacts), align="center", font=("Arial", 14, "bold"))
turtle.done()
````

## ۱۱. 🎲 قدم‌زدن تصادفی با مجموعه

لاک‌پشت در هر گام **به‌طور تصادفی** یکی از چهار جهت را انتخاب می‌کند. می‌خواهیم بدانیم چند گام طول می‌کشد تا **۶۰ خانه‌ی متفاوت** را ببیند. برای نگه‌داشتن خانه‌های دیده‌شده، **مجموعه** بهترین انتخاب است؛ چون تکرار ندارد و جست‌وجو با <code dir="ltr">in</code> روی آن خیلی سریع است!

این برنامه **همه‌ی** مفهوم‌های دوره را کنار هم دارد: لیست، حلقه‌ی while، به‌روزرسانی متغیر، شرط ترکیبی، اعتبارسنجی، <code dir="ltr">continue</code>، شمارنده‌ی ایمنی و مجموعه.

🧠 **پیوند با علوم اعصاب:** مدل‌های «قدم‌زدن تصادفی» برای توصیف حرکت **مولکول‌ها در مایع بین سلول‌ها** و حتی **ولتاژ غشای نورون** پیش از شلیک به کار می‌روند.

</div>

````python
import random

STEP = 20
LIMIT = 160                     # the turtle must stay inside this border
TARGET = 60                     # number of different cells we want to visit
MAX_STEPS = 5000                # safety counter

moves = [[STEP, 0], [-STEP, 0], [0, STEP], [0, -STEP]]

t = new_canvas("Random walk", speed=0)
t.color("#00897b")

x, y = 0, 0
visited = {(x, y)}              # (x, y) is a tuple: an unchangeable pair that can live in a set
steps = 0

while len(visited) < TARGET and steps < MAX_STEPS:
    steps += 1
    dx, dy = random.choice(moves)
    nx, ny = x + dx, y + dy

    # Validation: is the next cell inside the border?
    if not (-LIMIT <= nx <= LIMIT and -LIMIT <= ny <= LIMIT):
        continue

    t.goto(nx, ny)
    x, y = nx, ny

    if (x, y) in visited:
        t.dot(6, "#b0bec5")      # a cell we have seen before
    else:
        t.dot(10, "#00897b")     # a new cell
        visited.add((x, y))

print("Steps taken:", steps)
print("Different cells visited:", len(visited))
turtle.done()
````
