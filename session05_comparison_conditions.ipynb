## ۱. عملگرهای مقایسه

حاصل هر مقایسه یکی از دو مقدار **بولی** است: <code dir="ltr">True</code> (درست) یا <code dir="ltr">False</code> (نادرست).

| عملگر | معنی | نمونه | نتیجه |
|:---:|:---|:---|:---:|
| <code dir="ltr">==</code> | مساوی است با | <code dir="ltr">5 == 5</code> | <code dir="ltr">True</code> |
| <code dir="ltr">!=</code> | مساوی نیست با | <code dir="ltr">5 != 3</code> | <code dir="ltr">True</code> |
| <code dir="ltr">&lt;</code> | کوچک‌تر از | <code dir="ltr">3 &lt; 2</code> | <code dir="ltr">False</code> |
| <code dir="ltr">&gt;</code> | بزرگ‌تر از | <code dir="ltr">7 &gt; 2</code> | <code dir="ltr">True</code> |
| <code dir="ltr">&lt;=</code> | کوچک‌تر یا مساوی | <code dir="ltr">4 &lt;= 4</code> | <code dir="ltr">True</code> |
| <code dir="ltr">&gt;=</code> | بزرگ‌تر یا مساوی | <code dir="ltr">1 &gt;= 9</code> | <code dir="ltr">False</code> |

</div>

````python
heart_rate = 85

print(heart_rate > 100)
print(heart_rate == 85)
print(heart_rate != 60)

result = heart_rate >= 60
print(result, type(result))
````

### ⚠️ <code dir="ltr">=</code> با <code dir="ltr">==</code> فرق دارد!

- <code dir="ltr">=</code> یعنی **مقداردهی**: «این مقدار را در متغیر بگذار».
- <code dir="ltr">==</code> یعنی **مقایسه**: «آیا این دو مساوی‌اند؟»

نوشتن <code dir="ltr">if x = 5:</code> خطای <code dir="ltr">SyntaxError</code> می‌دهد.

</div>

### مقایسه‌ی رشته‌ها

رشته‌ها هم قابل مقایسه‌اند. <code dir="ltr">==</code> بررسی می‌کند که دو رشته **دقیقاً** یکسان باشند؛ حتی بزرگی و کوچکی حروف مهم است. <code dir="ltr">&lt;</code> و <code dir="ltr">&gt;</code> ترتیب **الفبایی** را می‌سنجند.

</div>

````python
print("Fp1" == "Fp1")
print("Fp1" == "fp1")         # حروف بزرگ و کوچک فرق دارند!
print("fp1".upper() == "FP1") # با upper همه را بزرگ می‌کنیم

print("apple" < "banana")     # a قبل از b می‌آید
````

````python
# حالا می‌توانیم تمرین ۴ جلسه‌ی اول را کامل کنیم: آیا کلمه آینه‌ای است؟
word = "level"
print(word == word[::-1])

word = "python"
print(word == word[::-1])
````

## ۳. شرط‌های ترکیبی: <code dir="ltr">and</code>، <code dir="ltr">or</code> و <code dir="ltr">not</code>

گاهی یک تصمیم به **چند شرط** بستگی دارد.

- <code dir="ltr">A and B</code>: فقط وقتی درست است که **هر دو** درست باشند.
- <code dir="ltr">A or B</code>: وقتی درست است که **دست‌کم یکی** درست باشد.
- <code dir="ltr">not A</code>: مقدار A را **برعکس** می‌کند.

### جدول درستی (Truth Table)

| <code dir="ltr">A</code> | <code dir="ltr">B</code> | <code dir="ltr">A and B</code> | <code dir="ltr">A or B</code> | <code dir="ltr">not A</code> |
|:---:|:---:|:---:|:---:|:---:|
| <code dir="ltr">True</code> | <code dir="ltr">True</code> | <code dir="ltr">True</code> | <code dir="ltr">True</code> | <code dir="ltr">False</code> |
| <code dir="ltr">True</code> | <code dir="ltr">False</code> | <code dir="ltr">False</code> | <code dir="ltr">True</code> | <code dir="ltr">False</code> |
| <code dir="ltr">False</code> | <code dir="ltr">True</code> | <code dir="ltr">False</code> | <code dir="ltr">True</code> | <code dir="ltr">True</code> |
| <code dir="ltr">False</code> | <code dir="ltr">False</code> | <code dir="ltr">False</code> | <code dir="ltr">False</code> | <code dir="ltr">True</code> |

بیا همین جدول را با یک حلقه‌ی تو در تو خود پایتون بسازیم:

</div>

````python
print("A      B      | and    or")
print("--------------+------------")
for a in [True, False]:
    for b in [True, False]:
        print(a, " ", b, " |", a and b, " ", a or b)
````

````python
# 🎢 شرط سوار شدن ترن هوایی: قد دست‌کم ۱۴۰ سانتی‌متر و سن دست‌کم ۱۲ سال
height = 150
age = 11

if height >= 140 and age >= 12:
    print("می‌توانی سوار شوی! 🎢")
else:
    print("متأسفیم، هنوز نه.")
````

## ۴. مقایسه‌ی زنجیره‌ای

در پایتون می‌توانیم مثل ریاضی بنویسیم <code dir="ltr">8 &lt;= f &lt; 13</code>. این دقیقاً یعنی <code dir="ltr">8 &lt;= f and f &lt; 13</code>؛ اما خواناتر است.

<div dir="rtl" style="font-family: Vazirmatn, Vazir, Tahoma, 'Segoe UI', sans-serif; line-height: 2; font-size: 15px; background: #fce4ec; color: #1b1b1b; border-right: 6px solid #c2185b; border-radius: 12px; padding: 14px 20px; margin: 12px 0;">

### 🧠 علوم اعصاب: باندهای فرکانسی امواج مغزی

سیگنال EEG ترکیبی از نوسان‌ها با سرعت‌های مختلف است. دانشمندان این نوسان‌ها را بر اساس **فرکانس** (تعداد نوسان در ثانیه، با واحد هرتز) به چند **باند** تقسیم می‌کنند:

| باند | محدوده‌ی تقریبی | معمولاً مرتبط با |
|:---|:---|:---|
| دلتا (δ) | ۰٫۵ تا ۴ هرتز | خواب عمیق |
| تتا (θ) | ۴ تا ۸ هرتز | خواب‌آلودگی، حافظه |
| آلفا (α) | ۸ تا ۱۳ هرتز | آرامش با چشمان بسته |
| بتا (β) | ۱۳ تا ۳۰ هرتز | تمرکز و فعالیت ذهنی |
| گاما (γ) | بیشتر از ۳۰ هرتز | پردازش‌های پیچیده |

مرز دقیق باندها در مقاله‌های مختلف کمی فرق دارد. در اینجا هر محدوده شامل عدد پایینی است، اما عدد بالایی را شامل نمی‌شود.

</div>

````python
frequencies = [2, 6.5, 10, 13, 22, 40, 0.2]

for f in frequencies:
    if 0.5 <= f < 4:
        band = "دلتا"
    elif 4 <= f < 8:
        band = "تتا"
    elif 8 <= f < 13:
        band = "آلفا"
    elif 13 <= f < 30:
        band = "بتا"
    elif f >= 30:
        band = "گاما"
    else:
        band = "خارج از محدوده‌ی تحلیل"
    print(f, "هرتز ←", band)
````

## ۵. شرط + حلقه = قدرت واقعی 💪

حالا که شرط و حلقه را با هم داریم، دو الگوی بسیار مهم می‌سازیم:

### الگوی فیلتر: نگه‌داشتن عضوهایی که شرط را دارند

</div>

````python
# 🧠 پیدا کردن نمونه‌هایی که از آستانه بالاتر رفته‌اند (احتمالاً یک اسپایک)
voltage = [-65, -60, -20, 30, -55, -70, 25, -62]   # میلی‌ولت
threshold = 0

spikes = []
for i, v in enumerate(voltage):
    if v > threshold:
        spikes.append(i)

print("اندیس اسپایک‌ها:", spikes)
````

### الگوی شمارش شرطی

````python
# 🎵 در جلسه‌ی ۲ قول دادیم: فقط نت‌های «می» را بشماریم
melody = ["می", "ر", "دو", "ر", "می", "می", "می"]

count = 0
for note in melody:
    if note == "می":
        count = count + 1

print("تعداد نت «می»:", count)
print("بررسی با count:", melody.count("می"))
````

### 💻 علوم کامپیوتر: سال کبیسه در تقویم میلادی

یکی از معروف‌ترین شرط‌های ترکیبی در برنامه‌نویسی این است. در تقویم میلادی، سالی **کبیسه** است که:

- بر ۴ بخش‌پذیر باشد **و** بر ۱۰۰ بخش‌پذیر **نباشد**؛
- **یا** بر ۴۰۰ بخش‌پذیر باشد.

عملگر <code dir="ltr">%</code> باقی‌مانده‌ی تقسیم را می‌دهد. پس «<code dir="ltr">y</code> بر ۴ بخش‌پذیر است» یعنی <code dir="ltr">y % 4 == 0</code>.

🗓️ تقویم هجری خورشیدی ما قانون کبیسه‌ی متفاوت و دقیق‌تری دارد که بر پایه‌ی رصد نجومی است.

</div>

````python
years = [1900, 2000, 2024, 2025, 2100]

for y in years:
    is_leap = (y % 4 == 0 and y % 100 != 0) or (y % 400 == 0)
    print(y, "کبیسه است؟", is_leap)
````

### 💡 اولویت عملگرها و پرانتز

پایتون ابتدا <code dir="ltr">not</code>، سپس <code dir="ltr">and</code> و در آخر <code dir="ltr">or</code> را حساب می‌کند. اما برای اینکه کدت **خوانا** باشد و اشتباه نکنی، همیشه از **پرانتز** استفاده کن؛ درست مثل مثال سال کبیسه.

</div>

````python
# 🔐 بررسی قدرت رمز عبور
password = "Neuron2026"

long_enough = len(password) >= 8

has_digit = False
has_upper = False
for ch in password:
    if ch.isdigit():
        has_digit = True
    if ch.isupper():
        has_upper = True

if long_enough and has_digit and has_upper:
    print("رمز قوی است 💪")
elif long_enough and (has_digit or has_upper):
    print("رمز متوسط است 🙂")
else:
    print("رمز ضعیف است 😟")
````

### ⚠️ دام عددهای اعشاری

کامپیوتر عددهای اعشاری را با دقت محدود نگه می‌دارد. به همین دلیل <code dir="ltr">0.1 + 0.2</code> دقیقاً برابر <code dir="ltr">0.3</code> **نمی‌شود**! برای مقایسه‌ی عددهای اعشاری، بررسی کن که **اختلافشان خیلی کوچک** باشد.

</div>

````python
print(0.1 + 0.2)
print(0.1 + 0.2 == 0.3)               # False 😮
print(abs((0.1 + 0.2) - 0.3) < 1e-9)  # True ✅  (abs یعنی قدر مطلق)
````

### 📜 داستان تاریخ: از جورج بول تا لطفی‌زاده

**جورج بول** (George Boole، ۱۸۱۵ تا ۱۸۶۴)، ریاضی‌دان انگلیسی، در سال ۱۸۵۴ کتابی به نام **«قوانین اندیشه»** (The Laws of Thought) منتشر کرد. او نشان داد که منطق را می‌توان مثل جبر با نمادها نوشت؛ با فقط دو مقدار **درست** و **نادرست** و عملگرهایی مثل **و**، **یا** و **نقیض**. امروز به آن **جبر بولی** می‌گوییم و نوع داده‌ی <code dir="ltr">bool</code> در پایتون به افتخار او نام‌گذاری شده است.

حدود ۸۰ سال بعد، دانشجوی جوانی به نام **کلود شانون** (Claude Shannon) در پایان‌نامه‌ی کارشناسی ارشدش در **MIT** (سال ۱۹۳۷) نشان داد که با **مدارهای الکتریکی** ساخته‌شده از کلید و رله می‌توان جبر بولی را پیاده کرد. این ایده پایه‌ی ساخت همه‌ی **پردازنده‌ها**ی امروزی است. درباره‌ی شانون در جلسه‌ی ۹ بیشتر می‌خوانیم.

🇮🇷 اما آیا همه‌چیز در دنیا فقط «درست» یا «نادرست» است؟ آیا هوای ۲۴ درجه «گرم» است یا «سرد»؟ **لطفی‌زاده** (Lotfi A. Zadeh، ۱۹۲۱ تا ۲۰۱۷)، دانشمند ایرانی‌تبار که در باکو به دنیا آمد، در تهران بزرگ شد و در **دانشکده‌ی فنی دانشگاه تهران** مهندسی برق خواند، در سال ۱۹۶۵ در دانشگاه **برکلی** نظریه‌ی **مجموعه‌های فازی** را مطرح کرد. در **منطق فازی**، درستی می‌تواند **درجه** داشته باشد؛ مثلاً ۰٫۷ درست! امروز منطق فازی در کنترل ماشین لباس‌شویی، دوربین‌ها، قطارها و بسیاری از سامانه‌های هوشمند به کار می‌رود.

### 🧩 کوییز سریع: True یا False؟

````python
x = 7
y = 3
print(x > 5 and y > 5)
print(x > 5 or y > 5)
print(not x == 7)
print(1 < y < x < 10)
print(x != y and not (y > x))
````

### ✏️ تمرین ۲ (متوسط): نمره‌ی کلامی

📚 در مدرسه‌های ایران نمره از ۰ تا ۲۰ است. برای هر نمره در لیست زیر، وضعیت را چاپ کن:

| نمره | وضعیت |
|:---|:---|
| ۱۷ تا ۲۰ | خیلی خوب |
| ۱۴ تا کمتر از ۱۷ | خوب |
| ۱۰ تا کمتر از ۱۴ | قابل قبول |
| کمتر از ۱۰ | نیاز به تلاش بیشتر |

</div>

````python
grades = [19.5, 12, 16.75, 8, 14, 20]
# کد خودت را اینجا بنویس ✍️
````

<div dir="rtl" style="font-family: Vazirmatn, Vazir, Tahoma, 'Segoe UI', sans-serif; line-height: 2; font-size: 15px;">

<details>
<summary><b>👀 دیدن پاسخ پیشنهادی</b></summary>

<div dir="ltr" align="left">

```python
grades = [19.5, 12, 16.75, 8, 14, 20]
for g in grades:
    if 17 <= g <= 20:
        status = "خیلی خوب"
    elif 14 <= g < 17:
        status = "خوب"
    elif 10 <= g < 14:
        status = "قابل قبول"
    else:
        status = "نیاز به تلاش بیشتر"
    print(g, "←", status)
```

### ✏️ تمرین ۳ (متوسط): فیلتر آلفا

🧠 لیست زیر فرکانس غالب EEG را در ۱۰ بازه‌ی زمانی نشان می‌دهد. با الگوی فیلتر، لیستی بساز که فقط **اندیس** بازه‌هایی را داشته باشد که در باند **آلفا** (۸ تا کمتر از ۱۳ هرتز) هستند. سپس درصد این بازه‌ها را چاپ کن.

</div>

````python
dominant_freq = [9.5, 10.2, 21.0, 11.8, 4.5, 8.0, 15.3, 12.9, 13.0, 7.9]
# کد خودت را اینجا بنویس ✍️
````

<div dir="rtl" style="font-family: Vazirmatn, Vazir, Tahoma, 'Segoe UI', sans-serif; line-height: 2; font-size: 15px;">

<div dir="ltr" align="left">

```python
dominant_freq = [9.5, 10.2, 21.0, 11.8, 4.5, 8.0, 15.3, 12.9, 13.0, 7.9]
alpha_windows = []
for i, f in enumerate(dominant_freq):
    if 8 <= f < 13:
        alpha_windows.append(i)
print("بازه‌های آلفا:", alpha_windows)      # [0, 1, 3, 5, 7]
percent = len(alpha_windows) / len(dominant_freq) * 100
print("درصد:", percent, "%")               # 50.0 %
```

### ✏️ تمرین ۴ (چالشی): نورون منطقی

🧠💻 یک **نورون مصنوعی ساده** دو ورودی <code dir="ltr">a</code> و <code dir="ltr">b</code> دارد (هر کدام <code dir="ltr">0</code> یا <code dir="ltr">1</code>). نورون ورودی‌ها را جمع می‌کند و اگر مجموع **بزرگ‌تر یا مساوی آستانه** باشد، <code dir="ltr">1</code> شلیک می‌کند؛ وگرنه <code dir="ltr">0</code>.

1. با آستانه‌ی <code dir="ltr">2</code>، برای هر چهار حالت ورودی (با دو حلقه‌ی تو در تو روی <code dir="ltr">[0, 1]</code>) خروجی نورون را چاپ کن. این نورون شبیه کدام عملگر منطقی رفتار می‌کند؟
2. آستانه را <code dir="ltr">1</code> بگذار. حالا شبیه کدام عملگر است؟

(این دقیقاً ایده‌ی **مک‌کالک و پیتس** در سال ۱۹۴۳ است که در جلسه‌ی ۷ درباره‌اش می‌خوانیم!)

</div>

````python
# کد خودت را اینجا بنویس ✍️
````

### 📝 خلاصه‌ی جلسه

- حاصل مقایسه (<code dir="ltr">==</code>، <code dir="ltr">!=</code>، <code dir="ltr">&lt;</code>، <code dir="ltr">&gt;</code>، <code dir="ltr">&lt;=</code>، <code dir="ltr">&gt;=</code>) همیشه <code dir="ltr">True</code> یا <code dir="ltr">False</code> است.
- <code dir="ltr">if</code> / <code dir="ltr">elif</code> / <code dir="ltr">else</code> از بالا به پایین بررسی می‌شوند و فقط **اولین** شرط درست اجرا می‌شود.
- <code dir="ltr">and</code>: هر دو درست باشند. <code dir="ltr">or</code>: دست‌کم یکی درست باشد. <code dir="ltr">not</code>: برعکس.
- مقایسه‌ی زنجیره‌ای مثل <code dir="ltr">8 &lt;= f &lt; 13</code> خوانا و کاربردی است.
- **الگوی فیلتر:** لیست خالی + حلقه + <code dir="ltr">if</code> + <code dir="ltr">append</code>.
- **الگوی شمارش شرطی:** شمارنده + حلقه + <code dir="ltr">if</code>.
- عددهای اعشاری را با <code dir="ltr">==</code> مقایسه نکن؛ اختلافشان را بسنج.
