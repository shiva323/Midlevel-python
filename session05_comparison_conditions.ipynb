## ۱. عملگرهای مقایسه

حاصل هر مقایسه یکی از دو مقدار **بولی** است: <code dir="ltr">True</code> (درست) یا <code dir="ltr">False</code> (نادرست).

| عملگر | معنی | نمونه | نتیجه |
|:---:|:---|:---:|:---:|
| <code dir="ltr">==</code> | مساوی است با | <code dir="ltr">5 == 5</code> | <code dir="ltr">True</code> |
| <code dir="ltr">!=</code> | مساوی نیست با | <code dir="ltr">5 != 3</code> | <code dir="ltr">True</code> |
| <code dir="ltr">&lt;</code> | کوچک‌تر از | <code dir="ltr">3 &lt; 2</code> | <code dir="ltr">False</code> |
| <code dir="ltr">&gt;</code> | بزرگ‌تر از | <code dir="ltr">7 &gt; 2</code> | <code dir="ltr">True</code> |
| <code dir="ltr">&lt;=</code> | کوچک‌تر یا مساوی | <code dir="ltr">4 &lt;= 4</code> | <code dir="ltr">True</code> |
| <code dir="ltr">&gt;=</code> | بزرگ‌تر یا مساوی | <code dir="ltr">1 &gt;= 9</code> | <code dir="ltr">False</code> |

</div>

````python
heart_rate = 85

print(heart_rate > 100)
print(heart_rate == 85)
print(heart_rate != 60)

result = heart_rate >= 60
print(result, type(result))
````

### مقایسه‌ی رشته‌ها

رشته‌ها هم قابل مقایسه‌اند:

- عملگر <code dir="ltr">==</code> بررسی می‌کند که دو رشته **دقیقاً** یکسان باشند؛ حتی بزرگی و کوچکی حروف هم مهم است.

## ۳. شرط‌های ترکیبی: and و or و not

گاهی یک تصمیم به **چند شرط** بستگی دارد:

- عبارت <code dir="ltr">A and B</code> فقط وقتی درست است که **هر دو** درست باشند.
- عبارت <code dir="ltr">A or B</code> وقتی درست است که **دست‌کم یکی** درست باشد.
- عبارت <code dir="ltr">not A</code> مقدار A را **برعکس** می‌کند.

### جدول درستی (Truth Table)

| <code dir="ltr">A</code> | <code dir="ltr">B</code> | <code dir="ltr">A and B</code> | <code dir="ltr">A or B</code> | <code dir="ltr">not A</code> |
|:---:|:---:|:---:|:---:|:---:|
| <code dir="ltr">True</code> | <code dir="ltr">True</code> | <code dir="ltr">True</code> | <code dir="ltr">True</code> | <code dir="ltr">False</code> |
| <code dir="ltr">True</code> | <code dir="ltr">False</code> | <code dir="ltr">False</code> | <code dir="ltr">True</code> | <code dir="ltr">False</code> |
| <code dir="ltr">False</code> | <code dir="ltr">True</code> | <code dir="ltr">False</code> | <code dir="ltr">True</code> | <code dir="ltr">True</code> |
| <code dir="ltr">False</code> | <code dir="ltr">False</code> | <code dir="ltr">False</code> | <code dir="ltr">False</code> | <code dir="ltr">True</code> |

بیا همین جدول را با یک حلقه‌ی تو در تو به دست خود پایتون بسازیم:

</div>

````python
print("A      B      | and    or")
print("--------------+------------")
for a in [True, False]:
    for b in [True, False]:
        print(str(a).ljust(6), str(b).ljust(6), "|", str(a and b).ljust(6), a or b)
````

## ۴. مقایسه‌ی زنجیره‌ای

در پایتون می‌توانیم مثل ریاضی بنویسیم <code dir="ltr">8 &lt;= f &lt; 13</code>. این عبارت دقیقاً همان <code dir="ltr">8 &lt;= f and f &lt; 13</code> است؛ اما خواناتر است.

<div dir="rtl" align="right" style="font-family: Vazirmatn, Vazir, Tahoma, 'Segoe UI', sans-serif; line-height: 2; font-size: 15px; text-align: right; background: #fce4ec; color: #1b1b1b; border-right: 6px solid #c2185b; border-radius: 12px; padding: 14px 20px; margin: 12px 0;">

### 🧠 علوم اعصاب: باندهای فرکانسی امواج مغزی

سیگنال EEG ترکیبی از نوسان‌هایی با سرعت‌های مختلف است. دانشمندان این نوسان‌ها را بر اساس **فرکانس** (تعداد نوسان در ثانیه، با واحد هرتز) به چند **باند** تقسیم می‌کنند:

| باند | نام انگلیسی | محدوده‌ی تقریبی | معمولاً مرتبط با |
|:---|:---:|:---|:---|
| دلتا | delta | ۰٫۵ تا ۴ هرتز | خواب عمیق |
| تتا | theta | ۴ تا ۸ هرتز | خواب‌آلودگی و حافظه |
| آلفا | alpha | ۸ تا ۱۳ هرتز | آرامش با چشمان بسته |
| بتا | beta | ۱۳ تا ۳۰ هرتز | تمرکز و فعالیت ذهنی |
| گاما | gamma | بیشتر از ۳۰ هرتز | پردازش‌های پیچیده |

مرز دقیق باندها در مقاله‌های مختلف کمی فرق دارد. در اینجا هر محدوده عدد پایینی را شامل می‌شود، اما عدد بالایی را نه.

</div>

````python
frequencies = [2, 6.5, 10, 13, 22, 40, 0.2]

for f in frequencies:
    if 0.5 <= f < 4:
        band = "delta"
    elif 4 <= f < 8:
        band = "theta"
    elif 8 <= f < 13:
        band = "alpha"
    elif 13 <= f < 30:
        band = "beta"
    elif f >= 30:
        band = "gamma"
    else:
        band = "outside the analysis range"
    print(f, "Hz ->", band)
````

````python
# Find the samples that crossed the threshold (probably a spike)
voltage = [-65, -60, -20, 30, -55, -70, 25, -62]   # millivolts
threshold = 0

spikes = []
for i, v in enumerate(voltage):
    if v > threshold:
        spikes.append(i)

print("Spike indices:", spikes)
````

### 💻 علوم کامپیوتر: سال کبیسه در تقویم میلادی

یکی از معروف‌ترین شرط‌های ترکیبی در برنامه‌نویسی این است. در تقویم میلادی، سالی **کبیسه** است که:

- بر ۴ بخش‌پذیر باشد **و** بر ۱۰۰ بخش‌پذیر **نباشد**؛
- **یا** بر ۴۰۰ بخش‌پذیر باشد.

عملگر <code dir="ltr">%</code> باقی‌مانده‌ی تقسیم را می‌دهد. پس جمله‌ی «y بر ۴ بخش‌پذیر است» در پایتون می‌شود <code dir="ltr">y % 4 == 0</code>.

🗓️ تقویم هجری خورشیدی ما قانون کبیسه‌ی متفاوت و دقیق‌تری دارد که بر پایه‌ی رصد نجومی است.

</div>

````python
years = [1900, 2000, 2024, 2025, 2100]

for y in years:
    is_leap = (y % 4 == 0 and y % 100 != 0) or (y % 400 == 0)
    print(y, "leap year?", is_leap)
````

### 💡 اولویت عملگرها و پرانتز

پایتون ابتدا <code dir="ltr">not</code>، سپس <code dir="ltr">and</code> و در آخر <code dir="ltr">or</code> را حساب می‌کند. اما برای اینکه کدت **خوانا** باشد و اشتباه نکنی، همیشه از **پرانتز** استفاده کن؛ درست مثل مثال سال کبیسه.

</div>

````python
# Password strength checker
password = "Neuron2026"

long_enough = len(password) >= 8

has_digit = False
has_upper = False
for ch in password:
    if ch.isdigit():
        has_digit = True
    if ch.isupper():
        has_upper = True

if long_enough and has_digit and has_upper:
    print("Strong password")
elif long_enough and (has_digit or has_upper):
    print("Medium password")
else:
    print("Weak password")
````

### ⚠️ دام عددهای اعشاری

کامپیوتر عددهای اعشاری را با دقت محدود نگه می‌دارد. به همین دلیل حاصل <code dir="ltr">0.1 + 0.2</code> دقیقاً برابر با <code dir="ltr">0.3</code> **نمی‌شود**!

````python
print(0.1 + 0.2)
print(0.1 + 0.2 == 0.3)               # False
print(abs((0.1 + 0.2) - 0.3) < 1e-9)  # True  (abs gives the absolute value)
````

### 📜 داستان تاریخ: از جورج بول تا لطفی‌زاده

**جورج بول** (George Boole، ۱۸۱۵ تا ۱۸۶۴)، ریاضی‌دان انگلیسی، در سال ۱۸۵۴ کتابی به نام **«قوانین اندیشه»** (The Laws of Thought) منتشر کرد. او نشان داد که منطق را می‌توان مثل جبر با نمادها نوشت؛ با فقط دو مقدار **درست** و **نادرست** و عملگرهایی مثل **و**، **یا** و **نقیض**. امروز به آن **جبر بولی** می‌گوییم و نوع داده‌ی <code dir="ltr">bool</code> در پایتون به افتخار او نام‌گذاری شده است.

حدود ۸۰ سال بعد، دانشجوی جوانی به نام **کلود شانون** (Claude Shannon) در پایان‌نامه‌ی کارشناسی ارشدش در **MIT** (سال ۱۹۳۷) نشان داد که با **مدارهای الکتریکی** ساخته‌شده از کلید و رله می‌توان جبر بولی را پیاده کرد. این ایده پایه‌ی ساخت همه‌ی **پردازنده‌های** امروزی است. درباره‌ی شانون در جلسه‌ی ۹ بیشتر می‌خوانیم.

🇮🇷 اما آیا همه‌چیز در دنیا فقط «درست» یا «نادرست» است؟ آیا هوای ۲۴ درجه «گرم» است یا «سرد»؟

**لطفی‌زاده** (Lotfi A. Zadeh، ۱۹۲۱ تا ۲۰۱۷)، دانشمند ایرانی‌تبار، در باکو به دنیا آمد، در تهران بزرگ شد و در **دانشکده‌ی فنی دانشگاه تهران** مهندسی برق خواند. او در سال ۱۹۶۵ در دانشگاه **برکلی** نظریه‌ی **مجموعه‌های فازی** را مطرح کرد. در **منطق فازی**، درستی می‌تواند **درجه** داشته باشد؛ مثلاً ۰٫۷ درست! امروز منطق فازی در کنترل ماشین لباس‌شویی، دوربین‌ها، قطارها و بسیاری از سامانه‌های هوشمند به کار می‌رود.
