## ۱. وقتی تعداد تکرار را از قبل نمی‌دانیم

حلقه‌ی <code dir="ltr">for</code> برای وقتی عالی است که **می‌دانیم** چند بار باید تکرار کنیم (مثلاً به تعداد اعضای لیست). اما گاهی فقط می‌دانیم **کی باید بایستیم**:

- «آب بریز **تا وقتی** لیوان پر نشده است.»
- «تمرین کن **تا وقتی** خسته نشده‌ای.»
- «پول پس‌انداز کن **تا وقتی** به قیمت دوچرخه برسی.»

برای این موقعیت‌ها از حلقه‌ی <code dir="ltr">while</code> (به معنی «تا وقتی که») استفاده می‌کنیم:

<div dir="ltr" align="left">

```python
while condition:
    # تا وقتی condition درست است، این بخش تکرار می‌شود
```

</div>

</div>

````python
count = 1               # ① مقدار اولیه

while count <= 5:       # ② شرط
    print("دور شماره‌ی", count)
    count = count + 1   # ③ به‌روزرسانی

print("حلقه تمام شد. مقدار نهایی count:", count)
````

<div dir="rtl" style="font-family: Vazirmatn, Vazir, Tahoma, 'Segoe UI', sans-serif; line-height: 2; font-size: 15px; background: #e8f5e9; color: #1b1b1b; border-right: 6px solid #2e7d32; border-radius: 12px; padding: 14px 20px; margin: 12px 0;">

### 💡 سه جزء طلایی هر حلقه‌ی while

| جزء | سؤالی که باید از خودت بپرسی | در مثال بالا |
|:---|:---|:---|
| ① **مقدار اولیه** | متغیر کنترل حلقه از کجا شروع می‌شود؟ | <code dir="ltr">count = 1</code> |
| ② **شرط** | تا کی باید ادامه دهیم؟ | <code dir="ltr">count &lt;= 5</code> |
| ③ **به‌روزرسانی** | در هر دور، متغیر چطور به سمت **پایان** حرکت می‌کند؟ | <code dir="ltr">count = count + 1</code> |

اگر جزء سوم را فراموش کنی، شرط همیشه درست می‌ماند و حلقه **هرگز تمام نمی‌شود**!

</div>

## ۲. به‌روزرسانی متغیر: شکل‌های کوتاه

به‌روزرسانی متغیرها آن‌قدر رایج است که پایتون برایش **میان‌بُر** دارد:

| شکل کامل | شکل کوتاه | معنی |
|:---|:---|:---|
| <code dir="ltr">x = x + 3</code> | <code dir="ltr">x += 3</code> | ۳ تا اضافه کن |
| <code dir="ltr">x = x - 3</code> | <code dir="ltr">x -= 3</code> | ۳ تا کم کن |
| <code dir="ltr">x = x * 3</code> | <code dir="ltr">x *= 3</code> | ضرب در ۳ |
| <code dir="ltr">x = x / 3</code> | <code dir="ltr">x /= 3</code> | تقسیم بر ۳ |
| <code dir="ltr">x = x // 3</code> | <code dir="ltr">x //= 3</code> | تقسیم صحیح بر ۳ |
| <code dir="ltr">x = x % 3</code> | <code dir="ltr">x %= 3</code> | باقی‌مانده‌ی تقسیم بر ۳ |

</div>

````python
score = 10
score += 5
print("بعد از += 5:", score)
score *= 2
print("بعد از *= 2:", score)
score -= 7
print("بعد از -= 7:", score)
score //= 4
print("بعد از //= 4:", score)
````

## ۳. جدول ردیابی برای while

بیا ببینیم یک عدد چند بار باید **نصف** شود تا به ۱ یا کمتر برسد.

</div>

````python
n = 100
steps = 0

while n > 1:
    n = n / 2
    steps += 1
    print("گام", steps, ": n =", n)

print("تعداد نصف کردن‌ها:", steps)
````

<div dir="rtl" style="font-family: Vazirmatn, Vazir, Tahoma, 'Segoe UI', sans-serif; line-height: 2; font-size: 15px; background: #e3f2fd; color: #1b1b1b; border-right: 6px solid #1565c0; border-radius: 12px; padding: 14px 20px; margin: 12px 0;">

### 💻 علوم کامپیوتر: چرا نصف کردن این‌قدر قدرتمند است؟

اگر یک لیست **مرتب** با ۱۰۰ عضو داشته باشی و بخواهی عددی را در آن پیدا کنی، می‌توانی هر بار عضو وسط را نگاه کنی و **نیمی از لیست را کنار بگذاری**. نتیجه‌ی بالا نشان می‌دهد که حداکثر حدود ۷ گام کافی است!

به این روش **جست‌وجوی دودویی** (binary search) می‌گویند. برای یک میلیون عضو هم فقط حدود ۲۰ گام لازم است. 🤯 بازی «عددی را که در ذهن دارم حدس بزن» را هم با همین ترفند می‌توان خیلی سریع برد.

### 🧠 علوم اعصاب: مدل «جمع کن و شلیک کن»

یک نورون در حالت استراحت **پتانسیل غشایی** حدود <code dir="ltr">-70</code> میلی‌ولت دارد. وقتی از نورون‌های دیگر سیگنال دریافت می‌کند، این ولتاژ کم‌کم بالا می‌رود. اگر ولتاژ به **آستانه** (حدود <code dir="ltr">-55</code> میلی‌ولت) برسد، نورون یک **اسپایک** شلیک می‌کند و ولتاژش دوباره به حالت استراحت برمی‌گردد.

به این مدل ساده، **جمع کن و شلیک کن** (Integrate-and-Fire) می‌گویند. این مدل را نخستین بار **لوئی لاپیک**، دانشمند فرانسوی، در سال ۱۹۰۷ مطرح کرد. ما نمی‌دانیم دقیقاً چند گام طول می‌کشد تا نورون شلیک کند؛ پس <code dir="ltr">while</code> بهترین انتخاب است!

</div>

````python
voltage = -70      # میلی‌ولت (پتانسیل استراحت)
threshold = -55    # آستانه‌ی شلیک
input_per_ms = 2   # افزایش ولتاژ در هر میلی‌ثانیه
time_ms = 0

while voltage < threshold:
    voltage += input_per_ms
    time_ms += 1
    print("t =", time_ms, "ms   V =", voltage, "mV")

print("⚡ اسپایک! نورون بعد از", time_ms, "میلی‌ثانیه شلیک کرد.")
````

## ۴. حلقه‌ی بی‌نهایت ♾️

اگر شرط حلقه‌ی <code dir="ltr">while</code> **هرگز نادرست نشود**، حلقه برای همیشه ادامه پیدا می‌کند. به این وضعیت **حلقه‌ی بی‌نهایت** (infinite loop) می‌گوییم. برنامه هنگ می‌کند، پردازنده داغ می‌شود و هیچ خروجی مفیدی نمی‌گیریم.

### نمونه‌های رایج

<div dir="ltr" align="left">

```python
# ❌ نمونه‌ی ۱: به‌روزرسانی فراموش شده
count = 1
while count <= 5:
    print(count)
    # count += 1  ← جا افتاده!

# ❌ نمونه‌ی ۲: به‌روزرسانی در جهت اشتباه
count = 1
while count <= 5:
    print(count)
    count -= 1     # count به جای بزرگ شدن، کوچک می‌شود!

# ❌ نمونه‌ی ۳: شرطی که هیچ‌وقت نادرست نمی‌شود
x = 1
while x != 10:
    x += 2         # x فقط فرد است: 1، 3، 5، 7، 9، 11، ... و هرگز ۱۰ نمی‌شود!
```

## ۵. پنج راه برای جلوگیری از حلقه‌ی بی‌نهایت 🛡️

1. **به‌روزرسانی را فراموش نکن.** قبل از نوشتن بدنه‌ی حلقه، اول خط به‌روزرسانی را بنویس.
2. **جهت حرکت را بررسی کن.** آیا هر به‌روزرسانی، متغیر را **به سمت** نادرست شدن شرط می‌برد؟
3. **از <code dir="ltr">&lt;</code> یا <code dir="ltr">&gt;</code> به جای <code dir="ltr">!=</code> استفاده کن.** در نمونه‌ی ۳، اگر شرط <code dir="ltr">x &lt; 10</code> بود، حلقه حتماً تمام می‌شد.
4. **یک شمارنده‌ی ایمنی بگذار.** حداکثر تعداد دور را محدود کن (مثال زیر).
5. **اگر تعداد تکرار معلوم است، از <code dir="ltr">for</code> استفاده کن.** حلقه‌ی <code dir="ltr">for</code> روی لیست یا <code dir="ltr">range</code> همیشه تمام می‌شود.

</div>

````python
# 🛡️ شمارنده‌ی ایمنی: حتی اگر شرط اصلی اشتباه باشد، حلقه بیشتر از ۲۰ دور نمی‌چرخد.
x = 1
safety = 0
MAX_STEPS = 20

while x != 10 and safety < MAX_STEPS:
    x += 2
    safety += 1

if safety == MAX_STEPS:
    print("⚠️ حلقه به سقف ایمنی رسید! احتمالاً شرط حلقه ایراد دارد. x =", x)
else:
    print("حلقه به‌درستی تمام شد.")
````

### 🔬 ریاضی: حدس کولاتز

این قانون ساده را در نظر بگیر و با یک عدد مثبت شروع کن:

- اگر عدد **زوج** است، آن را **نصف** کن.
- اگر عدد **فرد** است، آن را **سه برابر کن و یکی اضافه کن**.
- این کار را **تا وقتی** به ۱ برسی، تکرار کن.

ریاضی‌دان آلمانی **لوتار کولاتز** در سال ۱۹۳۷ حدس زد که با **هر** عدد مثبتی شروع کنیم، بالاخره به ۱ می‌رسیم. کامپیوترها این حدس را برای اعداد بسیار بسیار بزرگی آزموده‌اند، اما **هنوز هیچ‌کس نتوانسته آن را ثابت کند**! 😲

چون نمی‌دانیم این حلقه برای همه‌ی اعداد تمام می‌شود یا نه، یک شمارنده‌ی ایمنی هم می‌گذاریم.

</div>

````python
n = 27
steps = 0
path = [n]

while n != 1 and steps < 1000:
    if n % 2 == 0:
        n //= 2
    else:
        n = 3 * n + 1
    steps += 1
    path.append(n)

print("تعداد گام‌ها:", steps)
print("بزرگ‌ترین عددی که در مسیر دیدیم:", max(path))
print("ده عدد اول مسیر:", path[:10])
````

## ۶. <code dir="ltr">for</code> یا <code dir="ltr">while</code>؟

| موقعیت | انتخاب بهتر | نمونه |
|:---|:---:|:---|
| روی همه‌ی اعضای یک لیست کار داریم. | <code dir="ltr">for</code> | میانگین زمان‌های واکنش |
| تعداد تکرار را از قبل می‌دانیم. | <code dir="ltr">for</code> با <code dir="ltr">range</code> | جدول ضرب ۱ تا ۱۰ |
| فقط **شرط توقف** را می‌دانیم. | <code dir="ltr">while</code> | نورون تا رسیدن به آستانه |
| باید تا وقتی ورودی **معتبر** نیست، دوباره بپرسیم. | <code dir="ltr">while</code> | جلسه‌ی ۸ 😉 |

هر حلقه‌ی <code dir="ltr">for</code> را می‌توان با <code dir="ltr">while</code> نوشت. ببین:

</div>

````python
channels = ["Fp1", "C3", "O1"]

# با for
for ch in channels:
    print("for:", ch)

# همان کار با while: مقدار اولیه، شرط و به‌روزرسانی را خودمان می‌نویسیم.
i = 0
while i < len(channels):
    print("while:", channels[i])
    i += 1
````

### 📜 داستان تاریخ: ادسخر دایکسترا و برنامه‌نویسی ساخت‌یافته

**ادسخر دایکسترا** (Edsger W. Dijkstra، ۱۹۳۰ تا ۲۰۰۲)، دانشمند هلندی علوم کامپیوتر، از نخستین کسانی بود که به برنامه‌نویسی به چشم یک **دانش دقیق** نگاه کرد.

- در سال ۱۹۵۶، در حالی که با نامزدش در کافه‌ای در آمستردام نشسته بود، در حدود ۲۰ دقیقه و بدون کاغذ و قلم، الگوریتمی برای پیدا کردن **کوتاه‌ترین مسیر** بین دو شهر طراحی کرد. امروز **الگوریتم دایکسترا** در مسیریاب‌ها و شبکه‌های اینترنت به کار می‌رود. 🗺️
- در سال ۱۹۶۸ نامه‌ی مشهوری با عنوان **«دستور goto زیان‌آور است»** منتشر کرد و از **برنامه‌نویسی ساخت‌یافته** دفاع کرد؛ یعنی برنامه‌ها باید با ساختارهای روشنی مثل **شرط** و **حلقه** نوشته شوند تا بتوان درباره‌ی درستی‌شان فکر کرد.
- او تأکید می‌کرد که برنامه‌نویس باید برای هر حلقه **ثابت کند** که بالاخره **تمام می‌شود**؛ دقیقاً همان کاری که در این جلسه با «به‌روزرسانی به سمت پایان» کردیم.
- در سال ۱۹۷۲ **جایزه‌ی تورینگ** را دریافت کرد.

یکی از جمله‌های معروف او این است: **«آزمایش کردن برنامه می‌تواند وجود باگ را نشان دهد، اما هرگز نمی‌تواند نبودن باگ را ثابت کند.»** 🐞

### 🧩 کوییز سریع: این حلقه چند بار اجرا می‌شود؟

برای هر حلقه حدس بزن چند بار <code dir="ltr">print</code> اجرا می‌شود. کدام‌ها **بی‌نهایت** هستند؟ (سلول را اجرا **نکن**!)

<div dir="ltr" align="left">

```python
# الف
k = 10
while k > 0:
    print(k)
    k -= 3

# ب
k = 0
while k < 10:
    print(k)
```

### ✏️ تمرین ۴ (چالشی): نورونِ نشت‌دار

- در هر میلی‌ثانیه، ولتاژ به اندازه‌ی <code dir="ltr">input_per_ms = 2</code> بالا می‌رود.

2. حالا <code dir="ltr">input_per_ms = 1</code> بگذار. چه اتفاقی می‌افتد؟ 🤔 (اینجا شمارنده‌ی ایمنی جانت را نجات می‌دهد!)

```python
rest = -70
threshold = -55
input_per_ms = 2
voltage = rest
time_ms = 0
MAX_TIME = 500

while voltage < threshold and time_ms < MAX_TIME:
    voltage += input_per_ms
    voltage -= 0.1 * (voltage - rest)
    time_ms += 1

if voltage >= threshold:
    print("⚡ شلیک در", time_ms, "میلی‌ثانیه")
else:
    print("نورون هرگز شلیک نکرد. ولتاژ نهایی:", round(voltage, 2))
```

### 📝 خلاصه‌ی جلسه

- <code dir="ltr">while condition:</code> تا وقتی شرط درست است، تکرار می‌کند.
- هر حلقه‌ی while سه جزء دارد: **مقدار اولیه**، **شرط** و **به‌روزرسانی**.
- میان‌بُرهای به‌روزرسانی: <code dir="ltr">+=</code>، <code dir="ltr">-=</code>، <code dir="ltr">*=</code>، <code dir="ltr">/=</code>، <code dir="ltr">//=</code> و <code dir="ltr">%=</code>.
- حلقه‌ی بی‌نهایت وقتی رخ می‌دهد که شرط هیچ‌وقت نادرست نشود. با دکمه‌ی ■ یا دو بار کلید <code dir="ltr">I</code> متوقفش کن.
- برای جلوگیری از آن: به‌روزرسانی را درست بنویس، از <code dir="ltr">&lt;</code> و <code dir="ltr">&gt;</code> به جای <code dir="ltr">!=</code> استفاده کن و **شمارنده‌ی ایمنی** بگذار.
- تعداد تکرار معلوم است؟ <code dir="ltr">for</code>. فقط شرط توقف معلوم است؟ <code dir="ltr">while</code>.
