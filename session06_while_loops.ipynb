## ۱. وقتی تعداد تکرار را از قبل نمی‌دانیم

حلقه‌ی <code dir="ltr">for</code> برای وقتی عالی است که **می‌دانیم** چند بار باید تکرار کنیم؛ مثلاً به تعداد اعضای لیست. اما گاهی فقط می‌دانیم **کی باید بایستیم**:

- «آب بریز **تا وقتی** لیوان پر نشده است.»
- «تمرین کن **تا وقتی** خسته نشده‌ای.»
- «پول پس‌انداز کن **تا وقتی** به قیمت دوچرخه برسی.»

برای این موقعیت‌ها از حلقه‌ی <code dir="ltr">while</code> (به معنی «تا وقتی که») استفاده می‌کنیم:

<div dir="ltr" align="left">

```python
while condition:
    # repeats as long as condition is True
```

</div>

</div>

````python
count = 1               # 1) initial value

while count <= 5:       # 2) condition
    print("Round", count)
    count = count + 1   # 3) update

print("Loop finished. Final value of count:", count)
````

<div dir="rtl" align="right" style="font-family: Vazirmatn, Vazir, Tahoma, 'Segoe UI', sans-serif; line-height: 2; font-size: 15px; text-align: right; background: #e8f5e9; color: #1b1b1b; border-right: 6px solid #2e7d32; border-radius: 12px; padding: 14px 20px; margin: 12px 0;">

### 💡 سه جزء طلایی هر حلقه‌ی while

| جزء | سؤالی که باید از خودت بپرسی | در مثال بالا |
|:---|:---|:---|
| ۱. **مقدار اولیه** | متغیر کنترل حلقه از کجا شروع می‌شود؟ | <code dir="ltr">count = 1</code> |
| ۲. **شرط** | تا کی باید ادامه دهیم؟ | <code dir="ltr">count &lt;= 5</code> |
| ۳. **به‌روزرسانی** | در هر دور، متغیر چطور به سمت **پایان** حرکت می‌کند؟ | <code dir="ltr">count = count + 1</code> |

اگر جزء سوم را فراموش کنی، شرط همیشه درست می‌ماند و حلقه **هرگز تمام نمی‌شود**!

</div>

## ۲. به‌روزرسانی متغیر: شکل‌های کوتاه

به‌روزرسانی متغیرها آن‌قدر رایج است که پایتون برایش **میان‌بُر** دارد:

| شکل کامل | شکل کوتاه | معنی |
|:---|:---|:---|
| <code dir="ltr">x = x + 3</code> | <code dir="ltr">x += 3</code> | ۳ تا اضافه کن |
| <code dir="ltr">x = x - 3</code> | <code dir="ltr">x -= 3</code> | ۳ تا کم کن |
| <code dir="ltr">x = x * 3</code> | <code dir="ltr">x *= 3</code> | ضرب در ۳ |
| <code dir="ltr">x = x / 3</code> | <code dir="ltr">x /= 3</code> | تقسیم بر ۳ |
| <code dir="ltr">x = x // 3</code> | <code dir="ltr">x //= 3</code> | تقسیم صحیح بر ۳ |
| <code dir="ltr">x = x % 3</code> | <code dir="ltr">x %= 3</code> | باقی‌مانده‌ی تقسیم بر ۳ |

</div>

````python
score = 10
score += 5
print("after += 5:", score)
score *= 2
print("after *= 2:", score)
score -= 7
print("after -= 7:", score)
score //= 4
print("after //= 4:", score)
````

## ۳. جدول ردیابی برای while

بیا ببینیم یک عدد چند بار باید **نصف** شود تا به ۱ یا کمتر برسد.

</div>

````python
n = 100
steps = 0

while n > 1:
    n = n / 2
    steps += 1
    print("step", steps, ": n =", n)

print("Number of halvings:", steps)
````

<div dir="rtl" align="right" style="font-family: Vazirmatn, Vazir, Tahoma, 'Segoe UI', sans-serif; line-height: 2; font-size: 15px; text-align: right; background: #e3f2fd; color: #1b1b1b; border-right: 6px solid #1565c0; border-radius: 12px; padding: 14px 20px; margin: 12px 0;">

### 💻 علوم کامپیوتر: چرا نصف کردن این‌قدر قدرتمند است؟

فرض کن یک لیست **مرتب** با ۱۰۰ عضو داری و می‌خواهی عددی را در آن پیدا کنی. می‌توانی هر بار عضو وسط را نگاه کنی و **نیمی از لیست را کنار بگذاری**. نتیجه‌ی بالا نشان می‌دهد که حداکثر حدود ۷ گام کافی است!

به این روش **جست‌وجوی دودویی** (binary search) می‌گویند. برای یک میلیون عضو هم فقط حدود ۲۰ گام لازم است. 🤯 بازی «عددی را که در ذهن دارم حدس بزن» را هم با همین ترفند می‌توان خیلی سریع برد.

</div>

<div dir="rtl" align="right" style="font-family: Vazirmatn, Vazir, Tahoma, 'Segoe UI', sans-serif; line-height: 2; font-size: 15px; text-align: right; background: #fce4ec; color: #1b1b1b; border-right: 6px solid #c2185b; border-radius: 12px; padding: 14px 20px; margin: 12px 0;">

### 🧠 علوم اعصاب: مدل «جمع کن و شلیک کن»

یک نورون در حالت استراحت **پتانسیل غشایی** حدود منفی ۷۰ میلی‌ولت دارد. وقتی از نورون‌های دیگر سیگنال دریافت می‌کند، این ولتاژ کم‌کم بالا می‌رود. اگر ولتاژ به **آستانه** (حدود منفی ۵۵ میلی‌ولت) برسد، نورون یک **اسپایک** شلیک می‌کند و ولتاژش دوباره به حالت استراحت برمی‌گردد.

به این مدل ساده، **جمع کن و شلیک کن** (Integrate-and-Fire) می‌گویند. این مدل را نخستین بار **لوئی لاپیک**، دانشمند فرانسوی، در سال ۱۹۰۷ مطرح کرد.

ما نمی‌دانیم دقیقاً چند گام طول می‌کشد تا نورون شلیک کند؛ پس <code dir="ltr">while</code> بهترین انتخاب است!

</div>

````python
voltage = -70      # millivolts (resting potential)
threshold = -55    # firing threshold
input_per_ms = 2   # voltage increase per millisecond
time_ms = 0

while voltage < threshold:
    voltage += input_per_ms
    time_ms += 1
    print("t =", time_ms, "ms   V =", voltage, "mV")

print("Spike! The neuron fired after", time_ms, "ms.")
````

## ۴. حلقه‌ی بی‌نهایت ♾️

اگر شرط حلقه‌ی <code dir="ltr">while</code> **هرگز نادرست نشود**، حلقه برای همیشه ادامه پیدا می‌کند. به این وضعیت **حلقه‌ی بی‌نهایت** (infinite loop) می‌گوییم. برنامه هنگ می‌کند، پردازنده داغ می‌شود و هیچ خروجی مفیدی نمی‌گیریم.

### نمونه‌های رایج

<div dir="ltr" align="left">

```python
# Example 1: the update is missing
count = 1
while count <= 5:
    print(count)
    # count += 1  <- forgotten!

# Example 2: the update goes the wrong way
count = 1
while count <= 5:
    print(count)
    count -= 1     # count gets smaller instead of bigger!

# Example 3: a condition that can never become False
x = 1
while x != 10:
    x += 2         # x is always odd: 1, 3, 5, 7, 9, 11, ... it never equals 10!
```

## ۵. پنج راه برای جلوگیری از حلقه‌ی بی‌نهایت 🛡️

1. **به‌روزرسانی را فراموش نکن.** قبل از نوشتن بدنه‌ی حلقه، اول خط به‌روزرسانی را بنویس.
2. **جهت حرکت را بررسی کن.** آیا هر به‌روزرسانی، متغیر را **به سمت** نادرست شدن شرط می‌برد؟
3. **به جای <code dir="ltr">!=</code> از <code dir="ltr">&lt;</code> یا <code dir="ltr">&gt;</code> استفاده کن.** در نمونه‌ی ۳، اگر شرط <code dir="ltr">x &lt; 10</code> بود، حلقه حتماً تمام می‌شد.
4. **یک شمارنده‌ی ایمنی بگذار.** حداکثر تعداد دورها را محدود کن (مثال زیر).
5. **اگر تعداد تکرار معلوم است، از <code dir="ltr">for</code> استفاده کن.** حلقه‌ی <code dir="ltr">for</code> روی لیست یا <code dir="ltr">range</code> همیشه تمام می‌شود.

</div>

````python
# Safety counter: even if the main condition is wrong, the loop runs at most 20 times.
x = 1
safety = 0
MAX_STEPS = 20

while x != 10 and safety < MAX_STEPS:
    x += 2
    safety += 1

if safety == MAX_STEPS:
    print("Warning: the loop hit its safety limit! The condition is probably wrong. x =", x)
else:
    print("The loop finished normally.")
````

### 🔬 ریاضی: حدس کولاتز

این قانون ساده را در نظر بگیر و با یک عدد مثبت شروع کن:

- اگر عدد **زوج** است، آن را **نصف** کن.
- اگر عدد **فرد** است، آن را **سه برابر کن و یکی اضافه کن**.
- این کار را **تا وقتی** به ۱ برسی، تکرار کن.

ریاضی‌دان آلمانی **لوتار کولاتز** در سال ۱۹۳۷ حدس زد که با **هر** عدد مثبتی شروع کنیم، بالاخره به ۱ می‌رسیم. کامپیوترها این حدس را برای اعداد بسیار بزرگی آزموده‌اند، اما **هنوز هیچ‌کس نتوانسته آن را ثابت کند**! 😲

چون نمی‌دانیم این حلقه برای همه‌ی اعداد تمام می‌شود یا نه، یک شمارنده‌ی ایمنی هم می‌گذاریم.

</div>

````python
n = 27
steps = 0
path = [n]

while n != 1 and steps < 1000:
    if n % 2 == 0:
        n //= 2
    else:
        n = 3 * n + 1
    steps += 1
    path.append(n)

print("Number of steps:", steps)
print("Largest number on the path:", max(path))
print("First ten numbers:", path[:10])
````

## ۶. حلقه‌ی for یا حلقه‌ی while؟

| موقعیت | انتخاب بهتر | نمونه |
|:---|:---:|:---|
| روی همه‌ی اعضای یک لیست کار داریم. | <code dir="ltr">for</code> | میانگین زمان‌های واکنش |
| تعداد تکرار را از قبل می‌دانیم. | <code dir="ltr">for</code> با <code dir="ltr">range</code> | جدول ضرب ۱ تا ۱۰ |
| فقط **شرط توقف** را می‌دانیم. | <code dir="ltr">while</code> | نورون تا رسیدن به آستانه |
| باید تا وقتی ورودی **معتبر** نیست، دوباره بپرسیم. | <code dir="ltr">while</code> | جلسه‌ی ۸ 😉 |

هر حلقه‌ی <code dir="ltr">for</code> را می‌توان با <code dir="ltr">while</code> هم نوشت. ببین:

</div>

````python
channels = ["Fp1", "C3", "O1"]

# With for
for ch in channels:
    print("for:", ch)

# The same thing with while: we write the initial value, condition and update ourselves.
i = 0
while i < len(channels):
    print("while:", channels[i])
    i += 1
````

### 📜 داستان تاریخ: ادسخر دایکسترا و برنامه‌نویسی ساخت‌یافته

**ادسخر دایکسترا** (Edsger W. Dijkstra، ۱۹۳۰ تا ۲۰۰۲)، دانشمند هلندی علوم کامپیوتر، از نخستین کسانی بود که به برنامه‌نویسی به چشم یک **دانش دقیق** نگاه کرد.

- در سال ۱۹۵۶، در حالی که با نامزدش در کافه‌ای در آمستردام نشسته بود، در حدود ۲۰ دقیقه و بدون کاغذ و قلم، الگوریتمی برای پیدا کردن **کوتاه‌ترین مسیر** بین دو شهر طراحی کرد. امروز **الگوریتم دایکسترا** در مسیریاب‌ها و شبکه‌های اینترنت به کار می‌رود. 🗺️
- در سال ۱۹۶۸ نامه‌ی مشهوری با عنوان **«دستور goto زیان‌آور است»** منتشر کرد و از **برنامه‌نویسی ساخت‌یافته** دفاع کرد؛ یعنی برنامه‌ها باید با ساختارهای روشنی مثل **شرط** و **حلقه** نوشته شوند تا بتوان درباره‌ی درستی‌شان فکر کرد.
- او تأکید می‌کرد که برنامه‌نویس باید برای هر حلقه **ثابت کند** که بالاخره **تمام می‌شود**؛ دقیقاً همان کاری که در این جلسه با «به‌روزرسانی به سمت پایان» کردیم.
- او در سال ۱۹۷۲ **جایزه‌ی تورینگ** را دریافت کرد.

یکی از جمله‌های معروف او این است: **«آزمایش کردن برنامه می‌تواند وجود باگ را نشان دهد، اما هرگز نمی‌تواند نبودن باگ را ثابت کند.»** 🐞

voltage = rest

    voltage += input_per_ms
    voltage -= 0.1 * (voltage - rest)
    time_ms += 1

    print("Spike at", time_ms, "ms")
else:
    print("The neuron never fired. Final voltage:", round(voltage, 2))

### 📝 خلاصه‌ی جلسه

- حلقه‌ی <code dir="ltr">while condition:</code> تا وقتی شرط درست است تکرار می‌کند.
- هر حلقه‌ی while سه جزء دارد: **مقدار اولیه**، **شرط** و **به‌روزرسانی**.
- میان‌بُرهای به‌روزرسانی عبارت‌اند از <code dir="ltr">+=</code>، <code dir="ltr">-=</code>، <code dir="ltr">*=</code>، <code dir="ltr">/=</code>، <code dir="ltr">//=</code> و <code dir="ltr">%=</code>.
- حلقه‌ی بی‌نهایت وقتی رخ می‌دهد که شرط هیچ‌وقت نادرست نشود. با دکمه‌ی ■ یا دو بار کلید I متوقفش کن.
- برای جلوگیری از آن، به‌روزرسانی را درست بنویس، به جای <code dir="ltr">!=</code> از <code dir="ltr">&lt;</code> و <code dir="ltr">&gt;</code> استفاده کن و **شمارنده‌ی ایمنی** بگذار.
- اگر تعداد تکرار معلوم است از <code dir="ltr">for</code> استفاده کن؛ اگر فقط شرط توقف معلوم است، از <code dir="ltr">while</code>.
