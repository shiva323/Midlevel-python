### 🎯 اهداف این جلسه

در پایان این جلسه می‌توانی:

1. با حلقه‌ی <code dir="ltr">for</code> روی **همه‌ی اعضای یک لیست** حرکت کنی.
2. نقش **دونقطه** <code dir="ltr">:</code> و **تورفتگی** (indentation) را در پایتون توضیح بدهی.
3. با الگوی **جمع‌کننده** (accumulator) مجموع و میانگین یک لیست را حساب کنی.
4. با الگوی **شمارنده** تعداد چیزی را بشماری.
5. روی حروف یک **رشته** حلقه بزنی.
6. با جدول **ردیابی** (trace table) دنبال کنی که حلقه در هر دور چه کار می‌کند.

</div>

<div dir="rtl" style="font-family: Vazirmatn, Vazir, Tahoma, 'Segoe UI', sans-serif; line-height: 2; font-size: 15px; background: #e0f7fa; color: #1b1b1b; border-right: 6px solid #00838f; border-radius: 12px; padding: 14px 20px; margin: 12px 0;">

- اندیس از **صفر** شروع می‌شود: <code dir="ltr">channels[0]</code> یعنی <code dir="ltr">"Fp1"</code>.
- <code dir="ltr">channels[-1]</code> آخرین عضو است و <code dir="ltr">len(channels)</code> تعداد اعضا را می‌دهد.

❓ **سؤال گرم‌کردن:** <code dir="ltr">channels[1:]</code> چه چیزی را برمی‌گرداند؟

## ۱. مشکل: کارهای تکراری

فرض کن می‌خواهیم نام هر کانال EEG را با یک پیام چاپ کنیم. بدون حلقه باید این‌طور بنویسیم:

</div>

````python
channels = ["Fp1", "Fp2", "C3", "C4", "O1", "O2"]

print("در حال بررسی کانال", channels[0])
print("در حال بررسی کانال", channels[1])
print("در حال بررسی کانال", channels[2])
print("در حال بررسی کانال", channels[3])
print("در حال بررسی کانال", channels[4])
print("در حال بررسی کانال", channels[5])
````

<div dir="rtl" style="font-family: Vazirmatn, Vazir, Tahoma, 'Segoe UI', sans-serif; line-height: 2; font-size: 15px;">

اگر ۶۴ کانال داشتیم چه؟ یا اگر تعداد کانال‌ها را از قبل نمی‌دانستیم؟ 😩

برنامه‌نویس‌ها یک قانون طلایی دارند: **«خودت را تکرار نکن!»** (به انگلیسی: Don't Repeat Yourself یا به اختصار DRY). برای کارهای تکراری از **حلقه** استفاده می‌کنیم.

## ۲. ساختار حلقه‌ی <code dir="ltr">for</code>

<div dir="ltr" align="left">

```python
for item in my_list:
    # کارهایی که برای هر عضو انجام می‌شود
```

</div>

این را این‌طور بخوان: **«برای هر <code dir="ltr">item</code> در <code dir="ltr">my_list</code>، کارهای زیر را انجام بده.»**

- <code dir="ltr">for</code> و <code dir="ltr">in</code> کلمه‌های کلیدی پایتون هستند.
- <code dir="ltr">item</code> یک **متغیر حلقه** است. در هر دور، یکی از اعضای لیست در آن قرار می‌گیرد. اسمش را خودمان انتخاب می‌کنیم.
- **دونقطه** <code dir="ltr">:</code> در انتهای خط اول اجباری است.
- خط‌هایی که با **تورفتگی** (معمولاً ۴ فاصله) نوشته شده‌اند، **بدنه‌ی حلقه** هستند و در هر دور اجرا می‌شوند.

</div>

````python
channels = ["Fp1", "Fp2", "C3", "C4", "O1", "O2"]

for channel in channels:
    print("در حال بررسی کانال", channel)

print("✅ بررسی همه‌ی کانال‌ها تمام شد.")
````

<div dir="rtl" style="font-family: Vazirmatn, Vazir, Tahoma, 'Segoe UI', sans-serif; line-height: 2; font-size: 15px; background: #e8f5e9; color: #1b1b1b; border-right: 6px solid #2e7d32; border-radius: 12px; padding: 14px 20px; margin: 12px 0;">

### 💡 تورفتگی در پایتون معنا دارد

خط آخر برنامه‌ی بالا (<code dir="ltr">print("✅ ...")</code>) تورفتگی **ندارد**؛ پس جزء حلقه نیست و فقط **یک بار**، بعد از تمام شدن حلقه اجرا می‌شود.

در بسیاری از زبان‌های دیگر، تورفتگی فقط برای زیبایی است؛ اما در پایتون **بخشی از دستور زبان** است و تعیین می‌کند کدام خط‌ها داخل حلقه هستند.

</div>

<div dir="rtl" style="font-family: Vazirmatn, Vazir, Tahoma, 'Segoe UI', sans-serif; line-height: 2; font-size: 15px;">

## ۳. حلقه دقیقاً چه کار می‌کند؟ جدول ردیابی

بیا قدم‌به‌قدم ببینیم در هر دور حلقه چه اتفاقی می‌افتد. لیست سیاره‌های سنگی را در نظر بگیر:

</div>

````python
rocky_planets = ["عطارد", "زهره", "زمین", "مریخ"]

for planet in rocky_planets:
    print("🪐", planet)
````

<div dir="rtl" style="font-family: Vazirmatn, Vazir, Tahoma, 'Segoe UI', sans-serif; line-height: 2; font-size: 15px;">

| دور حلقه | مقدار <code dir="ltr">planet</code> | خروجی |
|:---:|:---:|:---|
| ۱ | <code dir="ltr">"عطارد"</code> | 🪐 عطارد |
| ۲ | <code dir="ltr">"زهره"</code> | 🪐 زهره |
| ۳ | <code dir="ltr">"زمین"</code> | 🪐 زمین |
| ۴ | <code dir="ltr">"مریخ"</code> | 🪐 مریخ |
| — | اعضای لیست تمام شد | حلقه پایان می‌یابد |

تعداد دورهای حلقه **دقیقاً برابر** با تعداد اعضای لیست است.

</div>

<div dir="rtl" style="font-family: Vazirmatn, Vazir, Tahoma, 'Segoe UI', sans-serif; line-height: 2; font-size: 15px; background: #fce4ec; color: #1b1b1b; border-right: 6px solid #c2185b; border-radius: 12px; padding: 14px 20px; margin: 12px 0;">

### 🧠 مثال علوم اعصاب: تبدیل واحد زمان واکنش

در آزمایش‌های **زمان واکنش** (reaction time)، شرکت‌کننده باید به محض دیدن یک نور روی صفحه، دکمه‌ای را فشار دهد. نرم‌افزار، زمان واکنش را بر حسب **میلی‌ثانیه** ثبت می‌کند. بیا آن‌ها را به **ثانیه** تبدیل کنیم (هر ثانیه ۱۰۰۰ میلی‌ثانیه است).

</div>

````python
reaction_times_ms = [312, 287, 345, 298, 401]

for rt in reaction_times_ms:
    seconds = rt / 1000
    print(rt, "میلی‌ثانیه =", seconds, "ثانیه")
````

<div dir="rtl" style="font-family: Vazirmatn, Vazir, Tahoma, 'Segoe UI', sans-serif; line-height: 2; font-size: 15px;">

## ۴. الگوی جمع‌کننده (Accumulator)

یکی از مهم‌ترین کاربردهای حلقه، **جمع کردن** اعضای لیست است. ایده ساده است:

1. **پیش از حلقه** یک متغیر به نام <code dir="ltr">total</code> می‌سازیم و مقدارش را <code dir="ltr">0</code> می‌گذاریم.
2. **در هر دور**، عضو فعلی را به <code dir="ltr">total</code> اضافه می‌کنیم.
3. **بعد از حلقه**، <code dir="ltr">total</code> مجموع همه‌ی اعضاست.

</div>

````python
reaction_times_ms = [312, 287, 345, 298, 401]

total = 0
for rt in reaction_times_ms:
    total = total + rt
    print("عدد فعلی:", rt, "  ← مجموع تا اینجا:", total)

average = total / len(reaction_times_ms)
print("--------------------")
print("مجموع:", total)
print("میانگین زمان واکنش:", average, "میلی‌ثانیه")
````

<div dir="rtl" style="font-family: Vazirmatn, Vazir, Tahoma, 'Segoe UI', sans-serif; line-height: 2; font-size: 15px; background: #e8f5e9; color: #1b1b1b; border-right: 6px solid #2e7d32; border-radius: 12px; padding: 14px 20px; margin: 12px 0;">

### 💡 خط <code dir="ltr">total = total + rt</code> را چطور بخوانیم؟

در ریاضی، <code dir="ltr">total = total + rt</code> بی‌معنی است! اما در برنامه‌نویسی، علامت <code dir="ltr">=</code> یعنی **«مقدار سمت راست را حساب کن و در متغیر سمت چپ بگذار»**.

پس این خط می‌گوید: «مقدار فعلی <code dir="ltr">total</code> را بردار، <code dir="ltr">rt</code> را به آن اضافه کن و نتیجه را دوباره در <code dir="ltr">total</code> بگذار». در جلسه‌ی ۶ شکل کوتاه‌تر آن یعنی <code dir="ltr">total += rt</code> را هم یاد می‌گیریم.

</div>

<div dir="rtl" style="font-family: Vazirmatn, Vazir, Tahoma, 'Segoe UI', sans-serif; line-height: 2; font-size: 15px; background: #e3f2fd; color: #1b1b1b; border-right: 6px solid #1565c0; border-radius: 12px; padding: 14px 20px; margin: 12px 0;">

### 💻 مثال علوم کامپیوتر: وزن همینگ

در علوم کامپیوتر، **تعداد بیت‌های ۱** در یک دنباله‌ی بیت را **وزن همینگ** (Hamming weight) می‌نامند. این عدد در رمزنگاری و تشخیص خطا کاربرد دارد. (درباره‌ی ریچارد همینگ در جلسه‌ی ۸ بیشتر می‌خوانیم.)

چون بیت‌ها فقط <code dir="ltr">0</code> یا <code dir="ltr">1</code> هستند، کافی است همه را با هم جمع کنیم!

</div>

````python
bits = [1, 0, 1, 1, 0, 0, 1, 0]

ones = 0
for bit in bits:
    ones = ones + bit

print("وزن همینگ این بایت:", ones)
````

<div dir="rtl" style="font-family: Vazirmatn, Vazir, Tahoma, 'Segoe UI', sans-serif; line-height: 2; font-size: 15px;">

## ۵. الگوی شمارنده

شمارنده شبیه جمع‌کننده است؛ با این تفاوت که در هر دور، به جای خود عضو، **عدد ۱** را اضافه می‌کنیم. این کار وقتی معنا پیدا می‌کند که بخواهیم چیزی را بشماریم.

</div>

````python
# 🎵 تعداد نت‌های یک ملودی
melody = ["می", "ر", "دو", "ر", "می", "می", "می"]

count = 0
for note in melody:
    count = count + 1

print("این ملودی", count, "نت دارد.")
print("بررسی با len:", len(melody))
````

<div dir="rtl" style="font-family: Vazirmatn, Vazir, Tahoma, 'Segoe UI', sans-serif; line-height: 2; font-size: 15px;">

البته برای شمردن همه‌ی اعضا، <code dir="ltr">len</code> کافی است. شمارنده وقتی واقعاً به کار می‌آید که بخواهیم فقط عضوهایی با **شرط خاص** را بشماریم (مثلاً فقط نت‌های «می»). این را در جلسه‌ی ۵ با <code dir="ltr">if</code> یاد می‌گیریم.

## ۶. ساختن متن در حلقه

می‌توانیم با جمع کردن رشته‌ها (<code dir="ltr">+</code>) در حلقه، یک متن بلندتر بسازیم:

</div>

````python
# 🧬 بازهای یک رشته‌ی DNA را پشت سر هم می‌چسبانیم و با خط تیره جدا می‌کنیم.
dna = ["A", "T", "G", "C", "G", "A"]

sequence = ""
for base in dna:
    sequence = sequence + base + "-"

print(sequence)
````

<div dir="rtl" style="font-family: Vazirmatn, Vazir, Tahoma, 'Segoe UI', sans-serif; line-height: 2; font-size: 15px;">

## ۷. حلقه روی یک رشته

حلقه‌ی <code dir="ltr">for</code> فقط مخصوص لیست نیست. یک **رشته** هم دنباله‌ای از حروف است و می‌توانیم روی حروفش حلقه بزنیم:

</div>

````python
word = "NEURON"

for letter in word:
    print(letter, "→", letter.lower())
````

<div dir="rtl" style="font-family: Vazirmatn, Vazir, Tahoma, 'Segoe UI', sans-serif; line-height: 2; font-size: 15px; background: #f1f8e9; color: #1b1b1b; border-right: 6px solid #558b2f; border-radius: 12px; padding: 14px 20px; margin: 12px 0;">

### 🔬 مثال فیزیک و محیط زیست: تبدیل دما

دمای هوای یک هفته بر حسب **سلسیوس** ثبت شده است. فرمول تبدیل به **فارنهایت** این است:

$$F = C \times 1.8 + 32$$

</div>

````python
temps_c = [18, 21, 25, 30, 27, 22, 19]

for c in temps_c:
    f = c * 1.8 + 32
    print(c, "°C  =", round(f, 1), "°F")
````

<div dir="rtl" style="font-family: Vazirmatn, Vazir, Tahoma, 'Segoe UI', sans-serif; line-height: 2; font-size: 15px; background: #fff3e0; color: #1b1b1b; border-right: 6px solid #ef6c00; border-radius: 12px; padding: 14px 20px; margin: 12px 0;">

### ⚠️ خطاهای رایج در حلقه‌ی for

1. **فراموش کردن دونقطه:** <code dir="ltr">for x in nums</code> ← خطای <code dir="ltr">SyntaxError</code>
2. **فراموش کردن تورفتگی:** اگر بدنه‌ی حلقه تورفتگی نداشته باشد، خطای <code dir="ltr">IndentationError</code> می‌گیری.
3. **تورفتگی اشتباه:** اگر <code dir="ltr">print</code> نتیجه‌ی نهایی را داخل حلقه بگذاری، به جای یک بار، در هر دور چاپ می‌شود.
4. **تغییر متغیر حلقه، لیست را عوض نمی‌کند:** سلول زیر را ببین.

</div>

````python
numbers = [1, 2, 3]

for n in numbers:
    n = n * 10   # فقط متغیر n عوض می‌شود، نه خود لیست!

print(numbers)   # هنوز [1, 2, 3] است
````

### 📜 داستان تاریخ: از خوارزمی تا تورینگ

**محمد بن موسی خوارزمی** (حدود ۷۸۰ تا ۸۵۰ میلادی)، دانشمند ایرانی اهل **خوارزم**، در «بیت‌الحکمه»ی بغداد کار می‌کرد. او کتاب‌هایی درباره‌ی حساب با ارقام هندی و حل معادله‌ها نوشت. وقتی کتاب‌هایش به لاتین ترجمه شد، نامش به صورت **Algoritmi** نوشته شد و واژه‌ی **الگوریتم** (algorithm) از همین‌جا آمد. واژه‌ی **جبر** (algebra) هم از نام کتاب او، «الجبر و المقابله»، گرفته شده است.

الگوریتم یعنی **دستورالعملی گام‌به‌گام برای حل یک مسئله**؛ و حلقه یکی از مهم‌ترین ابزارهای هر الگوریتم است.

بیش از هزار سال بعد، ریاضی‌دان انگلیسی **آلن تورینگ** (Alan Turing، ۱۹۱۲ تا ۱۹۵۴) در سال ۱۹۳۶ ماشینی خیالی را تعریف کرد که امروز **ماشین تورینگ** نام دارد. این ماشین یک **نوار** بلند دارد که به خانه‌های پشت سر هم تقسیم شده و در هر خانه یک نماد نوشته شده است. ماشین **خانه‌به‌خانه** جلو می‌رود، نماد را می‌خواند و طبق قوانینش تصمیم می‌گیرد؛ درست شبیه حلقه‌ای که روی اعضای یک لیست حرکت می‌کند! 🎞️

تورینگ در جنگ جهانی دوم در **بلچلی پارک** به شکستن رمز ماشین **انیگما**ی آلمان کمک کرد. در سال ۱۹۵۰ هم مقاله‌ای نوشت که «**آزمون تورینگ**» را برای سنجش هوش ماشین پیشنهاد می‌داد. معتبرترین جایزه‌ی علوم کامپیوتر، **جایزه‌ی تورینگ**، به نام اوست.

</div>

<div dir="rtl" style="font-family: Vazirmatn, Vazir, Tahoma, 'Segoe UI', sans-serif; line-height: 2; font-size: 15px; background: #e0f2f1; color: #1b1b1b; border-right: 6px solid #00695c; border-radius: 12px; padding: 14px 20px; margin: 12px 0;">

### 🧩 کوییز سریع: خروجی را حدس بزن!

قبل از اجرا، روی کاغذ بنویس این برنامه چه چیزی چاپ می‌کند. حواست به تورفتگی خط آخر باشد!

</div>

````python
scores = [4, 7, 1]
total = 0
for s in scores:
    total = total + s
    print("در حلقه:", total)
print("پایان:", total)
````

### ✏️ تمرین ۳ (متوسط): میانگین نرخ شلیک نورون

🧠 نورون‌ها با فرستادن پالس‌های الکتریکی کوتاهی به نام **اسپایک** (spike) با هم حرف می‌زنند. تعداد اسپایک‌ها در هر ثانیه را **نرخ شلیک** (firing rate) می‌نامند و واحد آن **هرتز** (Hz) است.

لیست زیر نرخ شلیک یک نورون را در ۶ آزمایش نشان می‌دهد. با الگوی جمع‌کننده، **مجموع** و **میانگین** نرخ شلیک را حساب و چاپ کن. (از تابع آماده‌ی <code dir="ltr">sum</code> استفاده نکن!)

### ✏️ تمرین ۵ (چالشی): برعکس کردن کلمه با حلقه

بدون استفاده از برش <code dir="ltr">[::-1]</code>، با یک حلقه‌ی <code dir="ltr">for</code> کلمه‌ی <code dir="ltr">"PYTHON"</code> را برعکس کن.

💡 **راهنمایی:** هر حرف جدید را به **اول** متن قبلی بچسبان، نه به آخرش: <code dir="ltr">reversed_word = letter + reversed_word</code>

یک جدول ردیابی هم روی کاغذ بکش تا ببینی در هر دور، مقدار <code dir="ltr">reversed_word</code> چه می‌شود.

</div>

````python
# کد خودت را اینجا بنویس ✍️
````

### 📝 خلاصه‌ی جلسه

- حلقه‌ی <code dir="ltr">for item in my_list:</code> برای **هر عضو** لیست یک بار اجرا می‌شود.
- **دونقطه** و **تورفتگی** تعیین می‌کنند کدام خط‌ها داخل حلقه هستند.
- **الگوی جمع‌کننده:** <code dir="ltr">total = 0</code> پیش از حلقه، <code dir="ltr">total = total + x</code> داخل حلقه.
- **الگوی شمارنده:** <code dir="ltr">count = 0</code> پیش از حلقه، <code dir="ltr">count = count + 1</code> داخل حلقه.
- روی **رشته** هم می‌توان حلقه زد: <code dir="ltr">for letter in "ABC":</code>
- تغییر دادن متغیر حلقه، خود لیست را تغییر **نمی‌دهد**.
