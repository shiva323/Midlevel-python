### 🎯 اهداف این جلسه

در پایان این جلسه می‌توانی:

1. با حلقه‌ی <code dir="ltr">for</code> روی **همه‌ی اعضای یک لیست** حرکت کنی.
2. نقش **دونقطه** و **تورفتگی** (indentation) را در پایتون توضیح بدهی.
3. با الگوی **جمع‌کننده** (accumulator) مجموع و میانگین یک لیست را حساب کنی.
4. با الگوی **شمارنده** تعداد چیزی را بشماری.
5. روی حروف یک **رشته** حلقه بزنی.
6. با **جدول ردیابی** (trace table) دنبال کنی که حلقه در هر دور چه کاری انجام می‌دهد.

</div>

## ۱. مشکل: کارهای تکراری

فرض کن می‌خواهیم نام هر کانال EEG را همراه با یک پیام چاپ کنیم. بدون حلقه باید این‌طور بنویسیم:

</div>

````python
channels = ["Fp1", "Fp2", "C3", "C4", "O1", "O2"]

print("Checking channel", channels[0])
print("Checking channel", channels[1])
print("Checking channel", channels[2])
print("Checking channel", channels[3])
print("Checking channel", channels[4])
print("Checking channel", channels[5])
````

<div dir="rtl" align="right" style="font-family: Vazirmatn, Vazir, Tahoma, 'Segoe UI', sans-serif; line-height: 2; font-size: 15px; text-align: right;">

اگر ۶۴ کانال داشتیم چه؟ یا اگر تعداد کانال‌ها را از قبل نمی‌دانستیم؟ 😩

برنامه‌نویس‌ها یک قانون طلایی دارند: **«خودت را تکرار نکن!»** (به انگلیسی: Don't Repeat Yourself یا به اختصار DRY). برای کارهای تکراری از **حلقه** استفاده می‌کنیم.

## ۲. ساختار حلقه‌ی for

<div dir="ltr" align="left">

```python
for item in my_list:
    # code that runs once for every item
```

</div>

این ساختار را این‌طور بخوان: **«برای هر <code dir="ltr">item</code> در <code dir="ltr">my_list</code>، کارهای زیر را انجام بده.»**

- کلمه‌های <code dir="ltr">for</code> و <code dir="ltr">in</code> کلمه‌های کلیدی پایتون هستند.
- متغیر <code dir="ltr">item</code> **متغیر حلقه** است. در هر دور، یکی از اعضای لیست در آن قرار می‌گیرد. نامش را خودمان انتخاب می‌کنیم.
- **دونقطه** در انتهای خط اول اجباری است.
- خط‌هایی که با **تورفتگی** (معمولاً ۴ فاصله) نوشته شده‌اند، **بدنه‌ی حلقه** هستند و در هر دور اجرا می‌شوند.

</div>

````python
channels = ["Fp1", "Fp2", "C3", "C4", "O1", "O2"]

for channel in channels:
    print("Checking channel", channel)

print("All channels checked.")
````

### 💡 تورفتگی در پایتون معنا دارد

خط آخر برنامه‌ی بالا تورفتگی **ندارد**؛ پس جزء حلقه نیست و فقط **یک بار**، بعد از تمام شدن حلقه اجرا می‌شود.

در بسیاری از زبان‌های دیگر، تورفتگی فقط برای زیبایی است؛ اما در پایتون **بخشی از دستور زبان** است و تعیین می‌کند کدام خط‌ها داخل حلقه هستند.

</div>

<div dir="rtl" align="right" style="font-family: Vazirmatn, Vazir, Tahoma, 'Segoe UI', sans-serif; line-height: 2; font-size: 15px; text-align: right;">

## ۳. حلقه دقیقاً چه کار می‌کند؟ جدول ردیابی

بیا قدم‌به‌قدم ببینیم در هر دور حلقه چه اتفاقی می‌افتد. لیست سیاره‌های سنگی را در نظر بگیر:

</div>

````python
rocky_planets = ["Mercury", "Venus", "Earth", "Mars"]

for planet in rocky_planets:
    print("Planet:", planet)
````

<div dir="rtl" align="right" style="font-family: Vazirmatn, Vazir, Tahoma, 'Segoe UI', sans-serif; line-height: 2; font-size: 15px; text-align: right;">

| دور حلقه | مقدار متغیر <code dir="ltr">planet</code> | خروجی |
|:---:|:---:|:---:|
| ۱ | <code dir="ltr">"Mercury"</code> | <code dir="ltr">Planet: Mercury</code> |
| ۲ | <code dir="ltr">"Venus"</code> | <code dir="ltr">Planet: Venus</code> |
| ۳ | <code dir="ltr">"Earth"</code> | <code dir="ltr">Planet: Earth</code> |
| ۴ | <code dir="ltr">"Mars"</code> | <code dir="ltr">Planet: Mars</code> |
| — | اعضای لیست تمام شد | حلقه پایان می‌یابد |

تعداد دورهای حلقه **دقیقاً برابر** با تعداد اعضای لیست است.

</div>

<div dir="rtl" align="right" style="font-family: Vazirmatn, Vazir, Tahoma, 'Segoe UI', sans-serif; line-height: 2; font-size: 15px; text-align: right; background: #fce4ec; color: #1b1b1b; border-right: 6px solid #c2185b; border-radius: 12px; padding: 14px 20px; margin: 12px 0;">

### 🧠 مثال علوم اعصاب: تبدیل واحد زمان واکنش

در آزمایش‌های **زمان واکنش** (reaction time)، شرکت‌کننده باید به محض دیدن یک نور روی صفحه، دکمه‌ای را فشار دهد. نرم‌افزار زمان واکنش را بر حسب **میلی‌ثانیه** ثبت می‌کند.

بیا این زمان‌ها را به **ثانیه** تبدیل کنیم. هر ثانیه ۱۰۰۰ میلی‌ثانیه است.

</div>

````python
reaction_times_ms = [312, 287, 345, 298, 401]

for rt in reaction_times_ms:
    seconds = rt / 1000
    print(rt, "ms =", seconds, "s")
````

<div dir="rtl" align="right" style="font-family: Vazirmatn, Vazir, Tahoma, 'Segoe UI', sans-serif; line-height: 2; font-size: 15px; text-align: right;">

## ۴. الگوی جمع‌کننده (Accumulator)

یکی از مهم‌ترین کاربردهای حلقه، **جمع کردن** اعضای لیست است. ایده ساده است:

1. **پیش از حلقه** متغیری به نام <code dir="ltr">total</code> می‌سازیم و مقدارش را صفر می‌گذاریم.
2. **در هر دور** عضو فعلی را به <code dir="ltr">total</code> اضافه می‌کنیم.
3. **بعد از حلقه**، متغیر <code dir="ltr">total</code> مجموع همه‌ی اعضاست.

</div>

````python
reaction_times_ms = [312, 287, 345, 298, 401]

total = 0
for rt in reaction_times_ms:
    total = total + rt
    print("current value:", rt, " -> running total:", total)

average = total / len(reaction_times_ms)
print("--------------------")
print("Total:", total)
print("Average reaction time:", average, "ms")
````

<div dir="rtl" align="right" style="font-family: Vazirmatn, Vazir, Tahoma, 'Segoe UI', sans-serif; line-height: 2; font-size: 15px; text-align: right; background: #e8f5e9; color: #1b1b1b; border-right: 6px solid #2e7d32; border-radius: 12px; padding: 14px 20px; margin: 12px 0;">

### 💡 خط <code dir="ltr">total = total + rt</code> را چطور بخوانیم؟

در ریاضی، این خط بی‌معنی است! اما در برنامه‌نویسی، علامت مساوی یعنی **«مقدار سمت راست را حساب کن و در متغیر سمت چپ بگذار»**.

پس این خط می‌گوید: «مقدار فعلی <code dir="ltr">total</code> را بردار، <code dir="ltr">rt</code> را به آن اضافه کن و نتیجه را دوباره در <code dir="ltr">total</code> بگذار».

در جلسه‌ی ۶ شکل کوتاه‌تر آن، یعنی <code dir="ltr">total += rt</code>، را هم یاد می‌گیریم.

</div>

<div dir="rtl" align="right" style="font-family: Vazirmatn, Vazir, Tahoma, 'Segoe UI', sans-serif; line-height: 2; font-size: 15px; text-align: right; background: #e3f2fd; color: #1b1b1b; border-right: 6px solid #1565c0; border-radius: 12px; padding: 14px 20px; margin: 12px 0;">

### 💻 مثال علوم کامپیوتر: وزن همینگ

در علوم کامپیوتر، **تعداد بیت‌های ۱** در یک دنباله‌ی بیت را **وزن همینگ** (Hamming weight) می‌نامند. این عدد در رمزنگاری و تشخیص خطا کاربرد دارد. درباره‌ی ریچارد همینگ در جلسه‌ی ۸ بیشتر می‌خوانیم.

چون بیت‌ها فقط صفر یا یک هستند، کافی است همه را با هم جمع کنیم!

</div>

````python
bits = [1, 0, 1, 1, 0, 0, 1, 0]

ones = 0
for bit in bits:
    ones = ones + bit

print("Hamming weight of this byte:", ones)
````

<div dir="rtl" align="right" style="font-family: Vazirmatn, Vazir, Tahoma, 'Segoe UI', sans-serif; line-height: 2; font-size: 15px; text-align: right;">

## ۵. الگوی شمارنده

شمارنده شبیه جمع‌کننده است؛ با این تفاوت که در هر دور، به جای خود عضو، **عدد ۱** را اضافه می‌کنیم.

</div>

````python
# How many notes does a melody have?
melody = ["mi", "re", "do", "re", "mi", "mi", "mi"]

count = 0
for note in melody:
    count = count + 1

print("This melody has", count, "notes.")
print("Check with len:", len(melody))
````

<div dir="rtl" align="right" style="font-family: Vazirmatn, Vazir, Tahoma, 'Segoe UI', sans-serif; line-height: 2; font-size: 15px; text-align: right;">

البته برای شمردن همه‌ی اعضا، تابع <code dir="ltr">len</code> کافی است. شمارنده وقتی واقعاً به کار می‌آید که بخواهیم فقط عضوهایی با **شرط خاص** را بشماریم؛ مثلاً فقط نت‌های «می». این را در جلسه‌ی ۵ با دستور <code dir="ltr">if</code> یاد می‌گیریم.

## ۶. ساختن متن در حلقه

می‌توانیم با به هم چسباندن رشته‌ها در حلقه، یک متن بلندتر بسازیم:

</div>

````python
# Join the bases of a DNA strand, separated by dashes
dna = ["A", "T", "G", "C", "G", "A"]

sequence = ""
for base in dna:
    sequence = sequence + base + "-"

print(sequence)
````

## ۷. حلقه روی یک رشته

حلقه‌ی <code dir="ltr">for</code> فقط مخصوص لیست نیست. یک **رشته** هم دنباله‌ای از حروف است و می‌توانیم روی حروفش حلقه بزنیم:

</div>

````python
word = "NEURON"

for letter in word:
    print(letter, "->", letter.lower())
````

<div dir="rtl" align="right" style="font-family: Vazirmatn, Vazir, Tahoma, 'Segoe UI', sans-serif; line-height: 2; font-size: 15px; text-align: right; background: #f1f8e9; color: #1b1b1b; border-right: 6px solid #558b2f; border-radius: 12px; padding: 14px 20px; margin: 12px 0;">

### 🔬 مثال فیزیک و محیط زیست: تبدیل دما

دمای هوای یک هفته بر حسب **سلسیوس** ثبت شده است. فرمول تبدیل آن به **فارنهایت** این است:

$$F = C \times 1.8 + 32$$

</div>

````python
temps_c = [18, 21, 25, 30, 27, 22, 19]

for c in temps_c:
    f = c * 1.8 + 32
    print(c, "C =", round(f, 1), "F")
````

<div dir="rtl" align="right" style="font-family: Vazirmatn, Vazir, Tahoma, 'Segoe UI', sans-serif; line-height: 2; font-size: 15px; text-align: right; background: #fff3e0; color: #1b1b1b; border-right: 6px solid #ef6c00; border-radius: 12px; padding: 14px 20px; margin: 12px 0;">

### ⚠️ خطاهای رایج در حلقه‌ی for

1. **فراموش کردن دونقطه:** اگر بنویسی <code dir="ltr">for x in nums</code> و دونقطه را جا بیندازی، خطای <code dir="ltr">SyntaxError</code> می‌گیری.
2. **فراموش کردن تورفتگی:** اگر بدنه‌ی حلقه تورفتگی نداشته باشد، خطای <code dir="ltr">IndentationError</code> می‌گیری.
3. **تورفتگی اشتباه:** اگر دستور چاپ نتیجه‌ی نهایی را داخل حلقه بگذاری، به جای یک بار، در هر دور چاپ می‌شود.
4. **تغییر متغیر حلقه، لیست را عوض نمی‌کند:** سلول زیر را ببین.

</div>

````python
numbers = [1, 2, 3]

for n in numbers:
    n = n * 10   # only the loop variable n changes, not the list itself!

print(numbers)   # still [1, 2, 3]
````

### 📜 داستان تاریخ: از خوارزمی تا تورینگ

**محمد بن موسی خوارزمی** (حدود ۷۸۰ تا ۸۵۰ میلادی)، دانشمند ایرانی اهل **خوارزم**، در «بیت‌الحکمه»ی بغداد کار می‌کرد. او کتاب‌هایی درباره‌ی حساب با ارقام هندی و حل معادله‌ها نوشت.

- وقتی کتاب‌هایش به لاتین ترجمه شد، نامش به صورت **Algoritmi** نوشته شد و واژه‌ی **الگوریتم** از همین‌جا آمد.
- واژه‌ی **جبر** (algebra) هم از نام کتاب او، «الجبر و المقابله»، گرفته شده است.

الگوریتم یعنی **دستورالعملی گام‌به‌گام برای حل یک مسئله**؛ و حلقه یکی از مهم‌ترین ابزارهای هر الگوریتم است.

بیش از هزار سال بعد، ریاضی‌دان انگلیسی **آلن تورینگ** (Alan Turing، ۱۹۱۲ تا ۱۹۵۴) در سال ۱۹۳۶ ماشینی خیالی را تعریف کرد که امروز **ماشین تورینگ** نام دارد. این ماشین یک **نوار** بلند دارد که به خانه‌های پشت سر هم تقسیم شده و در هر خانه یک نماد نوشته شده است. ماشین **خانه‌به‌خانه** جلو می‌رود، نماد را می‌خواند و طبق قوانینش تصمیم می‌گیرد؛ درست شبیه حلقه‌ای که روی اعضای یک لیست حرکت می‌کند! 🎞️

تورینگ در جنگ جهانی دوم در **بلچلی پارک** به شکستن رمز ماشین **انیگما**ی آلمان کمک کرد. در سال ۱۹۵۰ هم مقاله‌ای نوشت که «**آزمون تورینگ**» را برای سنجش هوش ماشین پیشنهاد می‌داد. معتبرترین جایزه‌ی علوم کامپیوتر، **جایزه‌ی تورینگ**، به نام اوست.

</div>

<div dir="rtl" align="right" style="font-family: Vazirmatn, Vazir, Tahoma, 'Segoe UI', sans-serif; line-height: 2; font-size: 15px; text-align: right; background: #e0f2f1; color: #1b1b1b; border-right: 6px solid #00695c; border-radius: 12px; padding: 14px 20px; margin: 12px 0;">

### 🧩 کوییز سریع: خروجی را حدس بزن!

قبل از اجرا، روی کاغذ بنویس این برنامه چه چیزی چاپ می‌کند. حواست به تورفتگی خط آخر باشد!

</div>

````python
scores = [4, 7, 1]
total = 0
for s in scores:
    total = total + s
    print("inside loop:", total)
print("end:", total)
````

### 📝 خلاصه‌ی جلسه

- حلقه‌ی <code dir="ltr">for item in my_list:</code> برای **هر عضو** لیست یک بار اجرا می‌شود.
