### 🎯 اهداف این جلسه

در پایان این جلسه می‌توانی:

1. با دستور <code dir="ltr">break</code> یک حلقه را **زودتر از موعد** متوقف کنی.
2. با دستور <code dir="ltr">continue</code> بقیه‌ی یک دور را **رد کنی** و به دور بعد بروی.

4. ورودی کاربر را **اعتبارسنجی** کنی؛ هم **نوع** آن (آیا عدد است؟) و هم **محدوده‌اش** (آیا معقول است؟).
5. شرط‌های ترکیبی را برای اعتبارسنجی به کار ببری.
6. ایده‌ی **بیت توازن** برای تشخیص خطا در داده‌ها را توضیح بدهی.

## ۱. دستور break: همین الان از حلقه بیرون برو! 🚪

گاهی پیش از تمام شدن حلقه به جوابمان رسیده‌ایم و ادامه دادن فقط وقت تلف کردن است. دستور <code dir="ltr">break</code> حلقه را **فوراً** متوقف می‌کند و اجرای برنامه از **اولین خط بعد از حلقه** ادامه پیدا می‌کند.

</div>

````python
# Find the FIRST spike: the moment the voltage crosses the threshold
voltage = [-70, -68, -65, -60, -52, 30, -75, -70, 25, -72]   # millivolts
threshold = -55

for t, v in enumerate(voltage):
    print("checking moment", t, "with voltage", v)
    if v > threshold:
        print("First spike found at moment", t)
        break

print("Search finished.")
````

<div dir="rtl" align="right" style="font-family: Vazirmatn, Vazir, Tahoma, 'Segoe UI', sans-serif; line-height: 2; font-size: 15px; text-align: right;">

به خروجی دقت کن: بعد از پیدا کردن اسپایک در لحظه‌ی ۴، حلقه **بقیه‌ی لیست را بررسی نکرد**. در یک ثبت واقعی با میلیون‌ها نمونه، این صرفه‌جویی خیلی مهم است.

## ۲. دستور continue: این یکی را رد کن! ⏭️

دستور <code dir="ltr">continue</code> بقیه‌ی بدنه‌ی حلقه را **در همین دور** نادیده می‌گیرد و مستقیم به **دور بعد** می‌رود. حلقه متوقف **نمی‌شود**.

</div>

<div dir="rtl" align="right" style="font-family: Vazirmatn, Vazir, Tahoma, 'Segoe UI', sans-serif; line-height: 2; font-size: 15px; text-align: right; background: #fce4ec; color: #1b1b1b; border-right: 6px solid #c2185b; border-radius: 12px; padding: 14px 20px; margin: 12px 0;">

### 🧠 علوم اعصاب: حذف آرتیفکت

سیگنال EEG خیلی ضعیف است؛ فقط چند ده میکروولت. وقتی شرکت‌کننده **پلک می‌زند**، **دندان‌هایش را به هم فشار می‌دهد** یا تکان می‌خورد، ولتاژهای بسیار بزرگی ثبت می‌شود که از مغز نیامده‌اند. به این‌ها **آرتیفکت** (artifact) می‌گویند.

یکی از ساده‌ترین روش‌ها برای پاک‌سازی این است که نمونه‌هایی را کنار بگذاریم که اندازه‌شان (قدر مطلقشان) از یک حد مشخص، مثلاً **۱۰۰ میکروولت**، بیشتر است.

</div>

````python
eeg = [12, -8, 15, 250, 10, -6, -310, 9, 14, -11]   # microvolts
LIMIT = 100

clean = []
rejected = 0

for sample in eeg:
    if abs(sample) > LIMIT:
        rejected += 1
        continue            # skip this sample and move on to the next one
    clean.append(sample)

print("Clean data:", clean)
print("Rejected samples:", rejected)
print("Mean of clean data:", round(sum(clean) / len(clean), 2))
````

<div dir="rtl" align="right" style="font-family: Vazirmatn, Vazir, Tahoma, 'Segoe UI', sans-serif; line-height: 2; font-size: 15px; text-align: right; background: #e8f5e9; color: #1b1b1b; border-right: 6px solid #2e7d32; border-radius: 12px; padding: 14px 20px; margin: 12px 0;">

### 💡 تفاوت break و continue در یک نگاه

| دستور | چه می‌کند؟ | تشبیه |
|:---|:---|:---|
| <code dir="ltr">break</code> | کل حلقه را **تمام** می‌کند. | از صف بیرون می‌آیی و به خانه می‌روی. 🏠 |
| <code dir="ltr">continue</code> | فقط **دور فعلی** را رد می‌کند. | از روی یک پله‌ی خراب می‌پری و بالا رفتن را ادامه می‌دهی. 🪜 |

</div>

````python
print("with continue:")
for n in range(1, 8):
    if n == 4:
        continue
    print(n, end=" ")

print("\nwith break:")
for n in range(1, 8):
    if n == 4:
        break
    print(n, end=" ")
````

<div dir="rtl" align="right" style="font-family: Vazirmatn, Vazir, Tahoma, 'Segoe UI', sans-serif; line-height: 2; font-size: 15px; text-align: right; background: #fff3e0; color: #1b1b1b; border-right: 6px solid #ef6c00; border-radius: 12px; padding: 14px 20px; margin: 12px 0;">

### ⚠️ دستور break فقط داخلی‌ترین حلقه را می‌شکند

در حلقه‌های تو در تو، <code dir="ltr">break</code> فقط از **همان حلقه‌ای** بیرون می‌آید که مستقیماً داخلش نوشته شده است؛ حلقه‌ی بیرونی به کارش ادامه می‌دهد.

</div>

````python
# For each channel, only check samples up to the first bad one.
channels = ["Fp1", "C3", "O1"]
data = [
    [10, 12, 500, 11],
    [5, 6, 7, 8],
    [-400, 3, 4, 5],
]

for i in range(len(channels)):
    good = 0
    for sample in data[i]:
        if abs(sample) > 100:
            break           # only the inner loop stops
        good += 1
    print(channels[i], ": good samples before the first bad one =", good)
````

## ۳. الگوی while True به همراه break

عبارت <code dir="ltr">while True</code> یعنی «برای همیشه تکرار کن»! 😮 این حلقه عمداً بی‌نهایت است، اما **داخل آن** یک <code dir="ltr">break</code> می‌گذاریم تا در لحظه‌ی مناسب بیرون بیاییم. این الگو وقتی مفید است که شرط توقف را فقط **وسط** حلقه بتوانیم بررسی کنیم.

</div>

<div dir="rtl" align="right" style="font-family: Vazirmatn, Vazir, Tahoma, 'Segoe UI', sans-serif; line-height: 2; font-size: 15px; text-align: right; background: #fff3e0; color: #1b1b1b; border-right: 6px solid #ef6c00; border-radius: 12px; padding: 14px 20px; margin: 12px 0;">

### ⚠️ قانون ایمنی while True

هر وقت <code dir="ltr">while True</code> نوشتی، **بلافاصله** از خودت بپرس: «آیا مطمئنم که بالاخره به <code dir="ltr">break</code> می‌رسم؟» اگر مطمئن نیستی، یک **شمارنده‌ی ایمنی** اضافه کن (جلسه‌ی ۶).

</div>

````python
# Number guessing game: the user's guesses are simulated with a list
secret = 42
guesses = [50, 30, 45, 42, 10]   # guesses the user "typed"

attempt = 0
while True:
    guess = guesses[attempt]
    attempt += 1
    if guess == secret:
        print("Well done! You guessed it on attempt", attempt)
        break
    elif guess < secret:
        print(guess, "-> guess higher")
    else:
        print(guess, "-> guess lower")
````

## ۴. اعتبارسنجی (Validation): به ورودی اعتماد نکن! 🛡️

یکی از قانون‌های مهم برنامه‌نویسی این است: **هرگز کورکورانه به ورودی کاربر اعتماد نکن.** کاربر ممکن است:

- به جای عدد، حرف تایپ کند؛ مثلاً <code dir="ltr">"fifteen"</code>.
- عدد منفی یا خیلی بزرگ وارد کند؛ مثلاً <code dir="ltr">-5</code> یا <code dir="ltr">999</code>.
- چیزی وارد نکند و فقط کلید Enter را بزند.

**اعتبارسنجی** یعنی بررسی اینکه ورودی **قابل قبول** است یا نه؛ پیش از اینکه از آن استفاده کنیم. اعتبارسنجی معمولاً دو مرحله دارد:

1. **بررسی نوع:** آیا ورودی عدد است؟ متد <code dir="ltr">isdigit</code> بررسی می‌کند که رشته فقط از رقم ساخته شده باشد.
2. **بررسی محدوده:** آیا عدد در بازه‌ی معقول است؟ اینجا **شرط‌های ترکیبی** جلسه‌ی ۵ به کار می‌آیند.

</div>

````python
print("123".isdigit())    # True
print("12a".isdigit())    # False
print("".isdigit())       # False
print("-5".isdigit())     # False  (the minus sign is not a digit!)
print("۱۵".isdigit())     # True   (Persian digits count as digits too, and int() understands them)
print(int("۱۵") + 1)      # 16
````

### نسخه‌ی واقعی با input

سلول زیر واقعاً از تو سؤال می‌پرسد. اجرایش کن و ورودی‌های نادرست مختلف را امتحان کن! 🧪

🧠 **سناریو:** یک پژوهش علوم اعصاب فقط شرکت‌کننده‌های **۱۲ تا ۱۸ ساله** را می‌پذیرد.

</div>

````python
while True:
    text = input("Enter the participant's age (12 to 18): ")

    if not text.isdigit():
        print("Please enter a number only.")
        continue

    age = int(text)

    if not (12 <= age <= 18):
        print("This study only accepts ages 12 to 18.")
        continue

    break   # if we get here, the input is valid

print("Saved. Participant age:", age)
````

### نسخه‌ی شبیه‌سازی‌شده

برای اینکه بتوانیم برنامه را با ورودی‌های مختلف **آزمایش** کنیم، ورودی‌ها را از یک لیست می‌خوانیم. ببین هر ورودی چطور بررسی می‌شود:

</div>

````python
fake_inputs = ["fifteen", "", "-3", "25", "9", "15", "16"]

for text in fake_inputs:
    print("input:", repr(text).ljust(10), end=" -> ")

    if not text.isdigit():
        print("not a number")
        continue

    age = int(text)
    if not (12 <= age <= 18):
        print("out of range")
        continue

    print("valid! age =", age)
    break
````

### 💻 علوم کامپیوتر: سه بار فرصت ورود

سیستم‌های امنیتی معمولاً فقط **تعداد محدودی** فرصت برای وارد کردن رمز می‌دهند. این کار جلوی کسی را می‌گیرد که می‌خواهد با امتحان کردن رمزهای زیاد وارد حساب شود. به این حمله **حمله‌ی جست‌وجوی فراگیر** (brute force) می‌گویند. در کد زیر، <code dir="ltr">break</code> و یک شمارنده با هم کار می‌کنند.

</div>

````python
correct_password = "Brain2026"
attempts = ["brain2026", "Brain 2026", "Brain2026"]   # typed passwords (simulated)
MAX_TRIES = 3

logged_in = False
for i in range(MAX_TRIES):
    typed = attempts[i]
    if typed == correct_password:
        logged_in = True
        break
    print("Wrong password. Tries left:", MAX_TRIES - i - 1)

if logged_in:
    print("Welcome!")
else:
    print("Account locked for 15 minutes.")
````

## ۵. اعتبارسنجی داده: بیت توازن

اعتبارسنجی فقط برای ورودی کاربر نیست. وقتی داده از طریق **کابل**، **وای‌فای** یا **حافظه** جابه‌جا می‌شود، ممکن است یک بیت به خاطر نویز به اشتباه از ۰ به ۱ (یا برعکس) تغییر کند.

یکی از ساده‌ترین راه‌های تشخیص این خطا، **بیت توازن** (parity bit) است:

1. فرستنده یک بیت اضافه به داده می‌چسباند تا **تعداد کل ۱ها زوج** شود.
2. گیرنده تعداد ۱ها را می‌شمارد.
3. اگر تعداد **فرد** بود، گیرنده می‌فهمد که خطایی رخ داده است!

</div>

````python
def count_ones(bits):
    total = 0
    for b in bits:
        total += b
    return total

# Sender
data = [1, 0, 1, 1, 0, 1, 0]
parity = count_ones(data) % 2        # if the number of ones is odd, the parity bit is 1
packet = data + [parity]
print("Sent packet:", packet)

# Receiver: three packets arrived; one is intact and two have noise
received_packets = [
    [1, 0, 1, 1, 0, 1, 0, 0],
    [1, 0, 0, 1, 0, 1, 0, 0],   # one bit flipped
    [1, 1, 0, 1, 0, 1, 0, 0],   # two bits flipped!
]

for p in received_packets:
    if count_ones(p) % 2 == 0:
        print(p, "-> looks OK")
    else:
        print(p, "-> error detected!")
````

### 📜 داستان تاریخ: ریچارد همینگ و ماشینی که خطایش را درست می‌کرد

**ریچارد همینگ** (Richard Hamming، ۱۹۱۵ تا ۱۹۹۸)، ریاضی‌دان آمریکایی، در دهه‌ی ۱۹۴۰ در **آزمایشگاه‌های بل** کار می‌کرد. کامپیوترهای آن زمان برنامه‌ها را از روی **کارت‌های پانچ‌شده** می‌خواندند و گاهی در خواندن یک بیت اشتباه می‌کردند.

همینگ فقط آخر هفته‌ها اجازه داشت از کامپیوتر استفاده کند. وقتی ماشین خطایی را **تشخیص** می‌داد، کار او را متوقف می‌کرد و همینگ باید تا هفته‌ی بعد صبر می‌کرد! او با عصبانیت از خودش پرسید: **«اگر ماشین می‌تواند تشخیص دهد که خطایی رخ داده، چرا نمی‌تواند بفهمد خطا کجاست و آن را درست کند؟»** 😤

نتیجه‌ی این عصبانیت، **کدهای همینگ** بود که در سال ۱۹۵۰ منتشر شد. همینگ با گذاشتن چند بیت توازن در جایگاه‌های هوشمندانه، روشی ساخت که نه‌تنها خطای یک‌بیتی را **تشخیص** می‌دهد، بلکه **محل آن را پیدا و اصلاحش** می‌کند.

امروز کدهای همینگ در **حافظه‌ی کامپیوترهای سرور** به کار می‌روند. کار او همچنین راه را برای **کدهای تصحیح خطای** پیشرفته‌تری باز کرد که در **ارتباط با ماهواره‌ها** و **کد QR** استفاده می‌شوند؛ به همین دلیل یک کد QR حتی اگر بخشی از آن پاره شود، باز هم خوانده می‌شود!

همینگ در سال ۱۹۶۸ **جایزه‌ی تورینگ** را دریافت کرد. «وزن همینگ» جلسه‌ی ۲ هم به نام اوست.

### 🧩 کوییز سریع: خروجی را حدس بزن!

پیش از اجرای سلول زیر، مقدار نهایی <code dir="ltr">total</code> را حدس بزن.

</div>

````python
total = 0
for n in [5, -2, 8, 0, 3, -7, 4]:
    if n < 0:
        continue
    if n == 0:
        break
    total += n
print(total)
````

### ✏️ تمرین ۴ (چالشی): آیا عدد اول است؟

💻 عدد **اول** عددی بزرگ‌تر از ۱ است که فقط بر ۱ و خودش بخش‌پذیر است؛ مثل ۲، ۳، ۵، ۷ و ۱۱. عددهای اول در **رمزنگاری** اینترنت نقش اساسی دارند. 🔐

برای هر عدد در لیست زیر بررسی کن که اول است یا نه:

- برای هر عدد n، مقسوم‌علیه‌ها را از ۲ به بعد امتحان کن.
- **به محض پیدا کردن اولین مقسوم‌علیه**، با <code dir="ltr">break</code> بیرون بیا؛ دیگر لازم نیست ادامه بدهی!

🌟 **چالش بیشتر:** چرا کافی است مقسوم‌علیه‌ها را فقط تا **جذر** n امتحان کنیم؟

</div>

````python
numbers = [2, 9, 13, 21, 29, 97, 100]
# Write your code here
````

```python
numbers = [2, 9, 13, 21, 29, 97, 100]
for n in numbers:
    is_prime = n > 1
    d = 2
    while d * d <= n:          # only up to the square root of n
        if n % d == 0:
            is_prime = False
            break
        d += 1
    print(n, "is prime" if is_prime else "is not prime")
# If n = a * b and both a and b were bigger than the square root of n,
# their product would be bigger than n. So at least one divisor is
# always less than or equal to the square root of n.
```

### 📝 خلاصه‌ی جلسه

- دستور <code dir="ltr">break</code> حلقه را **کامل** متوقف می‌کند؛ در حلقه‌ی تو در تو فقط داخلی‌ترین حلقه را.
- دستور <code dir="ltr">continue</code> بقیه‌ی **همین دور** را رد می‌کند و به دور بعد می‌رود.
- الگوی <code dir="ltr">while True</code> به همراه <code dir="ltr">break</code> برای وقتی است که شرط توقف وسط حلقه بررسی می‌شود؛ حواست به ایمنی‌اش باشد.
- **اعتبارسنجی:** اول **نوع** را بررسی کن (با <code dir="ltr">isdigit</code>)، بعد **محدوده** را (مثلاً <code dir="ltr">12 &lt;= age &lt;= 18</code>).
- **بیت توازن** با شمردن ۱ها خطاهای یک‌بیتی را تشخیص می‌دهد، اما همه‌ی خطاها را نه.
