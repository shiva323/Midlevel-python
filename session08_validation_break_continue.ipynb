### 🎯 اهداف این جلسه

در پایان این جلسه می‌توانی:

1. با <code dir="ltr">break</code> یک حلقه را **زودتر از موعد** متوقف کنی.
2. با <code dir="ltr">continue</code> بقیه‌ی یک دور را **رد کنی** و به دور بعد بروی.
3. الگوی <code dir="ltr">while True</code> + <code dir="ltr">break</code> را بشناسی و **بی‌خطر** از آن استفاده کنی.
4. ورودی کاربر را **اعتبارسنجی** کنی: هم **نوع** آن (آیا عدد است؟) و هم **محدوده**‌اش (آیا معقول است؟).
5. شرط‌های ترکیبی را برای اعتبارسنجی به کار ببری.

## ۱. <code dir="ltr">break</code>: همین الان از حلقه بیرون برو! 🚪

گاهی پیش از تمام شدن حلقه، به جوابمان رسیده‌ایم و ادامه دادن فقط وقت تلف کردن است. دستور <code dir="ltr">break</code> حلقه را **فوراً** متوقف می‌کند و اجرای برنامه از **اولین خط بعد از حلقه** ادامه پیدا می‌کند.

</div>

````python
# 🧠 پیدا کردن «اولین» اسپایک: لحظه‌ای که ولتاژ از آستانه بالاتر می‌رود
voltage = [-70, -68, -65, -60, -52, 30, -75, -70, 25, -72]   # میلی‌ولت
threshold = -55

for t, v in enumerate(voltage):
    print("بررسی لحظه‌ی", t, "با ولتاژ", v)
    if v > threshold:
        print("⚡ اولین اسپایک در لحظه‌ی", t, "پیدا شد!")
        break

print("جست‌وجو تمام شد.")
````

<div dir="rtl" style="font-family: Vazirmatn, Vazir, Tahoma, 'Segoe UI', sans-serif; line-height: 2; font-size: 15px;">

به خروجی دقت کن: بعد از پیدا کردن اسپایک در لحظه‌ی ۴، حلقه **بقیه‌ی لیست را بررسی نکرد**. در یک ثبت واقعی با میلیون‌ها نمونه، این صرفه‌جویی خیلی مهم است.

## ۲. <code dir="ltr">continue</code>: این یکی را رد کن! ⏭️

دستور <code dir="ltr">continue</code> بقیه‌ی بدنه‌ی حلقه را **در همین دور** نادیده می‌گیرد و مستقیم به **دور بعد** می‌رود. حلقه متوقف **نمی‌شود**.

</div>

<div dir="rtl" style="font-family: Vazirmatn, Vazir, Tahoma, 'Segoe UI', sans-serif; line-height: 2; font-size: 15px; background: #fce4ec; color: #1b1b1b; border-right: 6px solid #c2185b; border-radius: 12px; padding: 14px 20px; margin: 12px 0;">

### 🧠 علوم اعصاب: حذف آرتیفکت

سیگنال EEG خیلی ضعیف است (چند ده میکروولت). وقتی شرکت‌کننده **پلک می‌زند**، **دندان‌هایش را به هم فشار می‌دهد** یا تکان می‌خورد، ولتاژهای بسیار بزرگی ثبت می‌شود که از مغز نیامده‌اند. به این‌ها **آرتیفکت** (artifact) می‌گویند.

یکی از ساده‌ترین روش‌ها برای پاک‌سازی، کنار گذاشتن نمونه‌هایی است که اندازه‌شان (قدر مطلقشان) از یک حد مشخص، مثلاً **۱۰۰ میکروولت**، بیشتر است.

</div>

````python
eeg = [12, -8, 15, 250, 10, -6, -310, 9, 14, -11]   # میکروولت
LIMIT = 100

clean = []
rejected = 0

for sample in eeg:
    if abs(sample) > LIMIT:
        rejected += 1
        continue            # این نمونه را رد کن و برو سراغ نمونه‌ی بعدی
    clean.append(sample)

print("داده‌ی تمیز:", clean)
print("تعداد نمونه‌های ردشده:", rejected)
print("میانگین داده‌ی تمیز:", round(sum(clean) / len(clean), 2))
````

<div dir="rtl" style="font-family: Vazirmatn, Vazir, Tahoma, 'Segoe UI', sans-serif; line-height: 2; font-size: 15px; background: #e8f5e9; color: #1b1b1b; border-right: 6px solid #2e7d32; border-radius: 12px; padding: 14px 20px; margin: 12px 0;">

### 💡 تفاوت break و continue در یک نگاه

| دستور | چه می‌کند؟ | تشبیه |
|:---|:---|:---|
| <code dir="ltr">break</code> | کل حلقه را **تمام** می‌کند. | از صف بیرون می‌آیی و به خانه می‌روی. 🏠 |
| <code dir="ltr">continue</code> | فقط **دور فعلی** را رد می‌کند. | از روی یک پله‌ی خراب می‌پری و بالا رفتن را ادامه می‌دهی. 🪜 |

</div>

````python
print("با continue:")
for n in range(1, 8):
    if n == 4:
        continue
    print(n, end=" ")

print("\nبا break:")
for n in range(1, 8):
    if n == 4:
        break
    print(n, end=" ")
````

<div dir="rtl" style="font-family: Vazirmatn, Vazir, Tahoma, 'Segoe UI', sans-serif; line-height: 2; font-size: 15px; background: #fff3e0; color: #1b1b1b; border-right: 6px solid #ef6c00; border-radius: 12px; padding: 14px 20px; margin: 12px 0;">

### ⚠️ break فقط داخلی‌ترین حلقه را می‌شکند

در حلقه‌های تو در تو، <code dir="ltr">break</code> فقط از **همان حلقه‌ای** بیرون می‌آید که مستقیماً داخلش نوشته شده است؛ حلقه‌ی بیرونی به کارش ادامه می‌دهد.

</div>

````python
# برای هر کانال، فقط تا اولین نمونه‌ی خراب را بررسی می‌کنیم.
channels = ["Fp1", "C3", "O1"]
data = [
    [10, 12, 500, 11],
    [5, 6, 7, 8],
    [-400, 3, 4, 5],
]

for i in range(len(channels)):
    good = 0
    for sample in data[i]:
        if abs(sample) > 100:
            break           # فقط حلقه‌ی داخلی تمام می‌شود
        good += 1
    print(channels[i], ": تعداد نمونه‌های سالم پیش از اولین خرابی =", good)
````

<div dir="rtl" style="font-family: Vazirmatn, Vazir, Tahoma, 'Segoe UI', sans-serif; line-height: 2; font-size: 15px;">

## ۳. الگوی <code dir="ltr">while True</code> + <code dir="ltr">break</code>

<code dir="ltr">while True</code> یعنی «برای همیشه تکرار کن»! 😮 این حلقه عمداً بی‌نهایت است، اما **داخل آن** یک <code dir="ltr">break</code> می‌گذاریم تا در لحظه‌ی مناسب بیرون بیاییم. این الگو وقتی مفید است که شرط توقف را فقط **وسط** حلقه بتوانیم بررسی کنیم.

</div>

<div dir="rtl" style="font-family: Vazirmatn, Vazir, Tahoma, 'Segoe UI', sans-serif; line-height: 2; font-size: 15px; background: #fff3e0; color: #1b1b1b; border-right: 6px solid #ef6c00; border-radius: 12px; padding: 14px 20px; margin: 12px 0;">

### ⚠️ قانون ایمنی while True

هر وقت <code dir="ltr">while True</code> نوشتی، **بلافاصله** از خودت بپرس: «آیا مطمئنم که بالاخره به <code dir="ltr">break</code> می‌رسم؟» اگر مطمئن نیستی، یک **شمارنده‌ی ایمنی** اضافه کن (جلسه‌ی ۶).

</div>

````python
# 🎯 حدس عدد: پاسخ‌های کاربر را از یک لیست شبیه‌سازی می‌کنیم
secret = 42
guesses = [50, 30, 45, 42, 10]   # حدس‌هایی که کاربر «وارد کرده است»

attempt = 0
while True:
    guess = guesses[attempt]
    attempt += 1
    if guess == secret:
        print("🎉 آفرین! در تلاش", attempt, "درست حدس زدی.")
        break
    elif guess < secret:
        print(guess, "← بزرگ‌تر حدس بزن ⬆️")
    else:
        print(guess, "← کوچک‌تر حدس بزن ⬇️")
````

## ۴. اعتبارسنجی (Validation): به ورودی اعتماد نکن! 🛡️

یکی از قانون‌های مهم برنامه‌نویسی این است: **هرگز به ورودی کاربر کورکورانه اعتماد نکن.** کاربر ممکن است:

- به جای عدد، حرف تایپ کند: <code dir="ltr">"پانزده"</code>
- عدد منفی یا خیلی بزرگ وارد کند: <code dir="ltr">-5</code> یا <code dir="ltr">999</code>
- چیزی وارد نکند و فقط Enter بزند: <code dir="ltr">""</code>

**اعتبارسنجی** یعنی بررسی اینکه ورودی **قابل قبول** است یا نه؛ پیش از اینکه از آن استفاده کنیم. معمولاً دو مرحله دارد:

1. **بررسی نوع:** آیا ورودی عدد است؟ متد <code dir="ltr">isdigit()</code> بررسی می‌کند که رشته فقط از رقم ساخته شده باشد.
2. **بررسی محدوده:** آیا عدد در بازه‌ی معقول است؟ اینجا **شرط‌های ترکیبی** جلسه‌ی ۵ به کار می‌آیند.

</div>

````python
print("123".isdigit())    # True
print("12a".isdigit())    # False
print("".isdigit())       # False
print("-5".isdigit())     # False  (علامت منفی رقم نیست!)
print("۱۵".isdigit())     # True   (رقم‌های فارسی هم رقم حساب می‌شوند و int آن‌ها را می‌فهمد)
print(int("۱۵") + 1)      # 16
````

<div dir="rtl" style="font-family: Vazirmatn, Vazir, Tahoma, 'Segoe UI', sans-serif; line-height: 2; font-size: 15px;">

### نسخه‌ی واقعی با <code dir="ltr">input</code>

سلول زیر واقعاً از تو سؤال می‌پرسد. اجرایش کن و ورودی‌های نادرست مختلف را امتحان کن! 🧪

🧠 **سناریو:** یک پژوهش علوم اعصاب فقط شرکت‌کننده‌های **۱۲ تا ۱۸ ساله** را می‌پذیرد.

</div>

````python
while True:
    text = input("سن شرکت‌کننده را وارد کن (۱۲ تا ۱۸): ")

    if not text.isdigit():
        print("❌ لطفاً فقط عدد وارد کن.")
        continue

    age = int(text)

    if not (12 <= age <= 18):
        print("❌ این پژوهش فقط برای سن ۱۲ تا ۱۸ سال است.")
        continue

    break   # اگر به اینجا رسیدیم، ورودی معتبر است

print("✅ ثبت شد. سن شرکت‌کننده:", age)
````

<div dir="rtl" style="font-family: Vazirmatn, Vazir, Tahoma, 'Segoe UI', sans-serif; line-height: 2; font-size: 15px;">

### نسخه‌ی شبیه‌سازی‌شده

برای اینکه بتوانیم برنامه را با ورودی‌های مختلف **آزمایش** کنیم، ورودی‌ها را از یک لیست می‌خوانیم. ببین هر ورودی چطور بررسی می‌شود:

</div>

````python
fake_inputs = ["پانزده", "", "-3", "25", "9", "15", "16"]

for text in fake_inputs:
    print("ورودی:", repr(text), end="  →  ")

    if not text.isdigit():
        print("❌ عدد نیست")
        continue

    age = int(text)
    if not (12 <= age <= 18):
        print("❌ خارج از محدوده")
        continue

    print("✅ معتبر! سن =", age)
    break
````

### 💻 علوم کامپیوتر: سه بار فرصت ورود

سیستم‌های امنیتی معمولاً فقط **تعداد محدودی** فرصت برای وارد کردن رمز می‌دهند. این کار جلوی کسی را می‌گیرد که می‌خواهد با امتحان کردن رمزهای زیاد، وارد حساب شود (به این حمله **حمله‌ی جست‌وجوی فراگیر** یا brute force می‌گویند). اینجا <code dir="ltr">break</code> و یک شمارنده با هم کار می‌کنند.

</div>

````python
correct_password = "Brain2026"
attempts = ["brain2026", "Brain 2026", "Brain2026"]   # رمزهای واردشده (شبیه‌سازی)
MAX_TRIES = 3

logged_in = False
for i in range(MAX_TRIES):
    typed = attempts[i]
    if typed == correct_password:
        logged_in = True
        break
    print("❌ رمز نادرست است. فرصت باقی‌مانده:", MAX_TRIES - i - 1)

if logged_in:
    print("🔓 خوش آمدی!")
else:
    print("🔒 حساب به مدت ۱۵ دقیقه قفل شد.")
````

## ۵. اعتبارسنجی داده: بیت توازن

اعتبارسنجی فقط برای ورودی کاربر نیست. وقتی داده از طریق **کابل**، **وای‌فای** یا **حافظه** جابه‌جا می‌شود، ممکن است به خاطر نویز، یک بیت به اشتباه از <code dir="ltr">0</code> به <code dir="ltr">1</code> (یا برعکس) تغییر کند.

یکی از ساده‌ترین راه‌های تشخیص این خطا، **بیت توازن** (parity bit) است. فرستنده یک بیت اضافه به داده می‌چسباند تا **تعداد کل ۱ها زوج** شود. گیرنده تعداد ۱ها را می‌شمارد؛ اگر **فرد** بود، می‌فهمد که خطایی رخ داده است!

</div>

````python
def count_ones(bits):
    total = 0
    for b in bits:
        total += b
    return total

# 📤 فرستنده
data = [1, 0, 1, 1, 0, 1, 0]
parity = count_ones(data) % 2        # اگر تعداد ۱ها فرد باشد، بیت توازن ۱ می‌شود
packet = data + [parity]
print("بسته‌ی ارسالی:", packet)

# 📥 گیرنده: سه بسته دریافت شده است؛ یکی سالم و دو تا با نویز
received_packets = [
    [1, 0, 1, 1, 0, 1, 0, 0],
    [1, 0, 0, 1, 0, 1, 0, 0],   # یک بیت عوض شده
    [1, 1, 0, 1, 0, 1, 0, 0],   # دو بیت عوض شده!
]

for p in received_packets:
    if count_ones(p) % 2 == 0:
        print(p, "→ ✅ سالم به نظر می‌رسد")
    else:
        print(p, "→ ❌ خطا تشخیص داده شد!")
````

### 📜 داستان تاریخ: ریچارد همینگ و ماشینی که خطایش را درست می‌کرد

**ریچارد همینگ** (Richard Hamming، ۱۹۱۵ تا ۱۹۹۸)، ریاضی‌دان آمریکایی، در دهه‌ی ۱۹۴۰ در **آزمایشگاه‌های بل** کار می‌کرد. کامپیوترهای آن زمان برنامه‌ها را از روی **کارت‌های پانچ‌شده** می‌خواندند و گاهی در خواندن یک بیت اشتباه می‌کردند.

همینگ فقط آخر هفته‌ها اجازه داشت از کامپیوتر استفاده کند. وقتی ماشین خطایی را **تشخیص** می‌داد، کار او را متوقف می‌کرد و همینگ باید تا هفته‌ی بعد صبر می‌کرد! او با عصبانیت از خودش پرسید: **«اگر ماشین می‌تواند تشخیص دهد که خطایی رخ داده، چرا نمی‌تواند بفهمد خطا کجاست و آن را درست کند؟»** 😤

نتیجه‌ی این عصبانیت، **کدهای همینگ** بود که در سال ۱۹۵۰ منتشر شد. همینگ با گذاشتن چند بیت توازن در جایگاه‌های هوشمندانه، روشی ساخت که نه‌تنها خطای یک‌بیتی را **تشخیص** می‌دهد، بلکه **محل آن را پیدا و اصلاحش** می‌کند.

امروز کدهای همینگ در **حافظه‌ی کامپیوترهای سرور** به کار می‌روند. کار او همچنین راه را برای **کدهای تصحیح خطای** پیشرفته‌تری باز کرد که در **ارتباط با ماهواره‌ها** و **کد QR** استفاده می‌شوند؛ به همین دلیل یک کد QR حتی اگر بخشی از آن پاره شود، باز هم خوانده می‌شود! همینگ در سال ۱۹۶۸ **جایزه‌ی تورینگ** را دریافت کرد. «وزن همینگ» جلسه‌ی ۲ هم به نام اوست.

### ✏️ تمرین ۴ (چالشی): آیا عدد اول است؟

💻 عدد **اول** عددی بزرگ‌تر از ۱ است که فقط بر ۱ و خودش بخش‌پذیر است (مثل ۲، ۳، ۵، ۷، ۱۱). عددهای اول در **رمزنگاری** اینترنت نقش اساسی دارند. 🔐

برای هر عدد در لیست زیر بررسی کن که اول است یا نه. برای هر عدد <code dir="ltr">n</code>، مقسوم‌علیه‌ها را از <code dir="ltr">2</code> تا <code dir="ltr">n - 1</code> امتحان کن و **به محض پیدا کردن اولین مقسوم‌علیه**، با <code dir="ltr">break</code> بیرون بیا (دیگر لازم نیست ادامه بدهی!).

🌟 **چالش بیشتر:** چرا کافی است مقسوم‌علیه‌ها را فقط تا **جذر** <code dir="ltr">n</code> امتحان کنیم؟

</div>

````python
numbers = [2, 9, 13, 21, 29, 97, 100]
# کد خودت را اینجا بنویس ✍️
````

<div dir="rtl" style="font-family: Vazirmatn, Vazir, Tahoma, 'Segoe UI', sans-serif; line-height: 2; font-size: 15px;">

<div dir="ltr" align="left">

```python
numbers = [2, 9, 13, 21, 29, 97, 100]
for n in numbers:
    is_prime = n > 1
    d = 2
    while d * d <= n:          # فقط تا جذر n
        if n % d == 0:
            is_prime = False
            break
        d += 1
    print(n, "اول است" if is_prime else "اول نیست")
# اگر n = a × b و هر دو از جذر n بزرگ‌تر باشند، حاصل‌ضربشان از n بیشتر می‌شود؛
# پس دست‌کم یکی از مقسوم‌علیه‌ها حتماً کوچک‌تر یا مساوی جذر n است.
```

### 📝 خلاصه‌ی جلسه

- <code dir="ltr">break</code> حلقه را **کامل** متوقف می‌کند؛ در حلقه‌ی تو در تو فقط داخلی‌ترین حلقه را.
- <code dir="ltr">continue</code> بقیه‌ی **همین دور** را رد می‌کند و به دور بعد می‌رود.
- <code dir="ltr">while True</code> + <code dir="ltr">break</code> برای وقتی است که شرط توقف وسط حلقه بررسی می‌شود؛ حواست به ایمنی‌اش باشد.
- **اعتبارسنجی:** اول **نوع** را بررسی کن (<code dir="ltr">isdigit</code>)، بعد **محدوده** را (<code dir="ltr">12 &lt;= age &lt;= 18</code>).
- **بیت توازن** با شمردن ۱ها خطاهای یک‌بیتی را تشخیص می‌دهد، اما همه‌ی خطاها را نه.
