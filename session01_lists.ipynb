### 🎯 اهداف این جلسه

در پایان این جلسه می‌توانی:

1. توضیح بدهی که **چرا** به لیست نیاز داریم.
2. یک لیست **بسازی** و آن را چاپ کنی.
3. با **اندیس** (شماره‌ی خانه) به هر عضو لیست دسترسی پیدا کنی؛ هم با اندیس مثبت و هم با اندیس منفی.
4. با تابع <code dir="ltr">len</code> **طول** لیست را به دست بیاوری.
5. یک عضو لیست را **تغییر** بدهی.
6. با **برش** (slicing) تکه‌ای از لیست را جدا کنی.
7. با عملگر <code dir="ltr">in</code> بررسی کنی که چیزی در لیست **هست یا نه**.

## ۱. چرا به لیست نیاز داریم؟

فرض کن در یک آزمایشگاه علوم اعصاب کار می‌کنی. دستگاه **EEG** (نوار مغزی) در پنج لحظه‌ی پشت سر هم، ولتاژ مغز را بر حسب میکروولت اندازه گرفته است. اگر بخواهیم این داده‌ها را با متغیرهای معمولی نگه داریم، کدمان این‌طور می‌شود:

</div>

## ۲. ساختن لیست‌های مختلف

یک لیست می‌تواند عدد صحیح، عدد اعشاری، رشته (متن) یا حتی ترکیبی از این‌ها را نگه دارد. به چند نمونه از رشته‌های مختلف علمی نگاه کن:

</div>

### 💻 دانستنی کامپیوتری: بایت چیست؟

کامپیوتر همه‌چیز را با دو رقم <code dir="ltr">0</code> و <code dir="ltr">1</code> نگه می‌دارد. به هر کدام از این رقم‌ها **بیت** (bit) می‌گوییم. هشت بیت کنار هم یک **بایت** (byte) می‌سازند.

مثلاً بایت <code dir="ltr">[0, 1, 0, 0, 0, 0, 0, 1]</code> در کامپیوتر نشان‌دهنده‌ی حرف **A** است. در جلسه‌ی ۴ یاد می‌گیریم چطور این بیت‌ها را به عدد تبدیل کنیم!

## ۳. اندیس: شماره‌ی هر خانه

هر عضو لیست یک **شماره‌ی خانه** دارد که به آن **اندیس** (index) می‌گوییم.

⚠️ نکته‌ی بسیار مهم: در پایتون **شمارش از صفر شروع می‌شود**، نه از یک!

| عضو | <code dir="ltr">"frontal"</code> | <code dir="ltr">"parietal"</code> | <code dir="ltr">"temporal"</code> | <code dir="ltr">"occipital"</code> |
|:---:|:---:|:---:|:---:|:---:|
| **اندیس مثبت** | <code dir="ltr">0</code> | <code dir="ltr">1</code> | <code dir="ltr">2</code> | <code dir="ltr">3</code> |
| **اندیس منفی** | <code dir="ltr">-4</code> | <code dir="ltr">-3</code> | <code dir="ltr">-2</code> | <code dir="ltr">-1</code> |

برای خواندن یک عضو، نام لیست را می‌نویسیم و اندیس را داخل کروشه می‌گذاریم:

</div>

````python
brain_lobes = ["frontal", "parietal", "temporal", "occipital"]

print(brain_lobes[0])   # first item
print(brain_lobes[1])   # second item
print(brain_lobes[3])   # fourth (last) item
````

<div dir="rtl" align="right" style="font-family: Vazirmatn, Vazir, Tahoma, 'Segoe UI', sans-serif; line-height: 2; font-size: 15px; text-align: right;">

### اندیس منفی: شمردن از آخر

اندیس منفی از **انتهای** لیست می‌شمارد:

- اندیس <code dir="ltr">-1</code> یعنی آخرین عضو.
- اندیس <code dir="ltr">-2</code> یعنی یکی مانده به آخر.

این ترفند وقتی به درد می‌خورد که طول لیست را نمی‌دانیم، ولی آخرین عضو را لازم داریم.

</div>

````python
planets = ["Mercury", "Venus", "Earth", "Mars", "Jupiter", "Saturn", "Uranus", "Neptune"]

print("Farthest planet:", planets[-1])
print("Second to last: ", planets[-2])
print("Our home:       ", planets[2])
````

سلول زیر را اجرا کن تا این خطا را از نزدیک ببینی. نترس؛ خواندن پیام خطا یکی از مهارت‌های مهم برنامه‌نویس‌هاست! 🕵️

## ۴. طول لیست با تابع <code dir="ltr">len</code>

</div>

````python
eeg_channels = ["Fp1", "Fp2", "C3", "C4", "O1", "O2"]
one_byte = [0, 1, 0, 0, 0, 0, 0, 1]

print("Number of EEG channels:", len(eeg_channels))
print("Number of bits in a byte:", len(one_byte))
print("Length of an empty list:", len([]))
````

### 💡 رابطه‌ی طلایی

اگر طول لیستی <code dir="ltr">n</code> باشد، اندیس‌های مثبت آن از <code dir="ltr">0</code> تا <code dir="ltr">n - 1</code> هستند. پس آخرین عضو همیشه در خانه‌ی <code dir="ltr">my_list[len(my_list) - 1]</code> است؛ یعنی همان <code dir="ltr">my_list[-1]</code>.

## ۵. تغییر دادن عضوهای لیست

لیست **تغییرپذیر** (mutable) است؛ یعنی بعد از ساختنش می‌توانیم عضوهایش را عوض کنیم. کافی است با اندیس، مقدار تازه‌ای به یک خانه بدهیم.

🧠 **مثال:** فرض کن الکترود <code dir="ltr">C3</code> در حین ثبت جابه‌جا شده و آن را با الکترود <code dir="ltr">Cz</code> عوض کرده‌ایم:

</div>

````python
eeg_channels = ["Fp1", "Fp2", "C3", "C4", "O1", "O2"]
print("Before:", eeg_channels)

eeg_channels[2] = "Cz"
print("After: ", eeg_channels)
````

````python
# Turn on the last bit of a byte
one_byte = [0, 1, 0, 0, 0, 0, 0, 0]
one_byte[-1] = 1
print(one_byte)
````

## ۶. برش (Slicing): جدا کردن تکه‌ای از لیست

گاهی فقط بخشی از لیست را لازم داریم. با **برش** می‌توانیم یک تکه را جدا کنیم:

<div dir="ltr" align="left">

```python
my_list[start:stop]
```

</div>

- عدد اول (<code dir="ltr">start</code>) اندیسی است که برش **از آن شروع** می‌شود. خودش داخل برش هست.
- عدد دوم (<code dir="ltr">stop</code>) اندیسی است که برش **قبل از آن تمام** می‌شود. خودش داخل برش **نیست**.
- اگر عدد اول را ننویسیم، برش از اول لیست شروع می‌شود.
- اگر عدد دوم را ننویسیم، برش تا آخر لیست ادامه پیدا می‌کند.

</div>

````python
# Voltage recorded at 10 consecutive moments (microvolts)
eeg = [5, 12, 18, 9, -4, -11, -6, 3, 14, 20]

print("First three samples:", eeg[0:3])
print("Index 4 to 6:       ", eeg[4:7])
print("Index 7 to the end: ", eeg[7:])
print("All but the last 2: ", eeg[:-2])
print("A full copy:        ", eeg[:])
````

<div dir="rtl" align="right" style="font-family: Vazirmatn, Vazir, Tahoma, 'Segoe UI', sans-serif; line-height: 2; font-size: 15px; text-align: right;">

### گام (step)

برش می‌تواند عدد سومی هم داشته باشد که **گام** حرکت را تعیین می‌کند:

<div dir="ltr" align="left">

```python
my_list[start:stop:step]
```

</div>

</div>

````python
eeg = [5, 12, 18, 9, -4, -11, -6, 3, 14, 20]

print("Every other sample:", eeg[::2])
print("Reversed:          ", eeg[::-1])
````

<div dir="rtl" align="right" style="font-family: Vazirmatn, Vazir, Tahoma, 'Segoe UI', sans-serif; line-height: 2; font-size: 15px; text-align: right; background: #fce4ec; color: #1b1b1b; border-right: 6px solid #c2185b; border-radius: 12px; padding: 14px 20px; margin: 12px 0;">

### 🧠 علوم اعصاب: کاهش نرخ نمونه‌برداری

دانشمندان علوم اعصاب گاهی داده‌ی EEG را با سرعت خیلی بالا ثبت می‌کنند (مثلاً ۱۰۰۰ نمونه در ثانیه)، ولی برای بعضی تحلیل‌ها این‌همه نمونه لازم نیست.

یکی از ساده‌ترین ایده‌ها این است که فقط **یکی در میان** نمونه‌ها را نگه دارند تا حجم داده نصف شود. به این کار **کاهش نرخ نمونه‌برداری** (downsampling) می‌گویند؛ همان کاری که با <code dir="ltr">eeg[::2]</code> کردیم!

البته در دنیای واقعی، پیش از این کار یک فیلتر هم روی سیگنال اعمال می‌کنند تا اطلاعات مهم از دست نرود.

</div>

````python
# A neat trick: reversing a DNA strand
dna = ["A", "T", "G", "C", "G", "A"]
print("Original:", dna)
print("Reversed:", dna[::-1])
````

<div dir="rtl" align="right" style="font-family: Vazirmatn, Vazir, Tahoma, 'Segoe UI', sans-serif; line-height: 2; font-size: 15px; text-align: right;">

## ۷. آیا این عضو در لیست هست؟

عملگر <code dir="ltr">in</code> بررسی می‌کند که یک مقدار در لیست وجود دارد یا نه، و عملگر <code dir="ltr">not in</code> برعکس آن را بررسی می‌کند. جواب هر دو یکی از این دو مقدار است:

- مقدار <code dir="ltr">True</code> یعنی «درست».
- مقدار <code dir="ltr">False</code> یعنی «نادرست».

</div>

````python
eeg_channels = ["Fp1", "Fp2", "C3", "C4", "O1", "O2"]

print("O1" in eeg_channels)       # Was electrode O1 recorded?
print("Pz" in eeg_channels)       # Was electrode Pz recorded?
print("Pz" not in eeg_channels)   # Was Pz left out?
````

<div dir="rtl" align="right" style="font-family: Vazirmatn, Vazir, Tahoma, 'Segoe UI', sans-serif; line-height: 2; font-size: 15px; text-align: right;">

## ۸. لیست درون لیست (یک نگاه کوتاه)

عضو یک لیست می‌تواند خودش یک لیست باشد! مثلاً داده‌های دو شرکت‌کننده در یک آزمایش زمان واکنش:

</div>

````python
# Each inner list: [name, first reaction time, second reaction time] in milliseconds
experiment = [
    ["Maryam", 310, 295],
    ["Ali", 350, 330],
]

print(experiment[0])      # everything about Maryam
print(experiment[1][0])   # name of the second person
print(experiment[0][2])   # Maryam's second reaction time
````

### 💡 یادت باشد

عبارت <code dir="ltr">experiment[1][0]</code> را این‌طور بخوان: «از لیست بیرونی عضو شماره‌ی ۱ را بردار، سپس از آن عضو شماره‌ی ۰ را بردار».

در جلسه‌ی ۷ با حلقه‌های تو در تو حسابی با این ساختار کار می‌کنیم.

### 📜 داستان تاریخ: ایدا لاولیس و چارلز ببیج

حدود ۲۰۰ سال پیش، ریاضی‌دان انگلیسی **چارلز ببیج** (Charles Babbage، ۱۷۹۱ تا ۱۸۷۱) طرح ماشینی به نام **موتور تحلیلی** (Analytical Engine) را کشید. این ماشین مکانیکی بود و با چرخ‌دنده کار می‌کرد، اما ایده‌هایش شبیه کامپیوترهای امروزی بود:

- بخشی به نام **انبار** (the Store) داشت که قرار بود هزار عدد را در ستون‌های شماره‌دار نگه دارد؛ تقریباً مثل یک لیست بزرگ با اندیس!
- بخشی به نام **آسیاب** (the Mill) داشت که محاسبه‌ها را انجام می‌داد؛ چیزی شبیه پردازنده‌ی امروزی.

**ایدا لاولیس** (Ada Lovelace، ۱۸۱۵ تا ۱۸۵۲)، ریاضی‌دان جوان انگلیسی، در سال ۱۸۴۳ یادداشت‌های مفصلی درباره‌ی این ماشین نوشت. او در یکی از این یادداشت‌ها (یادداشت G) روشی گام‌به‌گام برای محاسبه‌ی دنباله‌ای از اعداد به نام **اعداد برنولی** توضیح داد. بسیاری او را **نخستین برنامه‌نویس جهان** می‌دانند.
