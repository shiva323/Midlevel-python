### 🎯 اهداف این جلسه

در پایان این جلسه می‌توانی:

1. توضیح بدهی **چرا** به لیست نیاز داریم.
2. یک لیست **بسازی** و آن را چاپ کنی.
3. با **اندیس** (شماره‌ی خانه) به هر عضو لیست دسترسی پیدا کنی؛ هم با اندیس مثبت و هم با اندیس منفی.
4. با تابع <code dir="ltr">len</code> **طول** لیست را به دست بیاوری.
5. یک عضو لیست را **تغییر** بدهی.
6. با **برش** (slicing) تکه‌ای از لیست را جدا کنی.
7. با <code dir="ltr">in</code> بررسی کنی که چیزی در لیست **هست یا نه**.

</div>

## ۱. چرا به لیست نیاز داریم؟

فرض کن در یک آزمایشگاه علوم اعصاب کار می‌کنی و دستگاه **EEG** (نوار مغزی) در پنج لحظه‌ی پشت سر هم، ولتاژ مغز را بر حسب میکروولت اندازه گرفته است. اگر بخواهیم با متغیرهای معمولی این داده‌ها را نگه داریم، کدمان این‌طور می‌شود:

</div>

````python
sample1 = 12.5
sample2 = -3.1
sample3 = 8.0
sample4 = 15.2
sample5 = -7.4

print(sample1, sample2, sample3, sample4, sample5)
````

<div dir="rtl" style="font-family: Vazirmatn, Vazir, Tahoma, 'Segoe UI', sans-serif; line-height: 2; font-size: 15px;">

این روش برای پنج عدد قابل تحمل است؛ اما یک دستگاه EEG واقعی در **هر ثانیه** صدها نمونه ثبت می‌کند! 😵

راه‌حل پایتون، **لیست** (<code dir="ltr">list</code>) است: یک متغیر که می‌تواند **چندین مقدار را به ترتیب** در خودش نگه دارد. لیست را با **کروشه** <code dir="ltr">[ ]</code> می‌سازیم و اعضای آن را با **ویرگول** <code dir="ltr">,</code> از هم جدا می‌کنیم.

</div>

````python
eeg_samples = [12.5, -3.1, 8.0, 15.2, -7.4]

print(eeg_samples)
print(type(eeg_samples))
````

<div dir="rtl" style="font-family: Vazirmatn, Vazir, Tahoma, 'Segoe UI', sans-serif; line-height: 2; font-size: 15px; background: #e8f5e9; color: #1b1b1b; border-right: 6px solid #2e7d32; border-radius: 12px; padding: 14px 20px; margin: 12px 0;">

### 💡 لیست را مثل یک قطار تصور کن

لیست شبیه یک **قطار** است: هر واگن یک مقدار را حمل می‌کند و واگن‌ها به ترتیب پشت سر هم قرار دارند. ترتیب واگن‌ها مهم است و تا وقتی خودمان آن را تغییر ندهیم، ثابت می‌ماند.

🚂 <code dir="ltr">[ 12.5 | -3.1 | 8.0 | 15.2 | -7.4 ]</code>

</div>

<div dir="rtl" style="font-family: Vazirmatn, Vazir, Tahoma, 'Segoe UI', sans-serif; line-height: 2; font-size: 15px;">

## ۲. ساختن لیست‌های مختلف

یک لیست می‌تواند عدد صحیح، عدد اعشاری، رشته (متن) یا حتی ترکیبی از این‌ها را نگه دارد. به چند نمونه از رشته‌های مختلف علمی نگاه کن:

</div>

````python
# 🧠 علوم اعصاب: چهار لوب اصلی قشر مغز
brain_lobes = ["پیشانی", "آهیانه", "گیجگاهی", "پس‌سری"]

# 🧠 نام الکترودهای EEG (طبق سیستم بین‌المللی ۱۰-۲۰)
eeg_channels = ["Fp1", "Fp2", "C3", "C4", "O1", "O2"]

# 💻 علوم کامپیوتر: یک بایت از ۸ بیت ساخته می‌شود
one_byte = [0, 1, 0, 0, 0, 0, 0, 1]

# 🧬 زیست‌شناسی: تکه‌ای از یک رشته‌ی DNA
dna = ["A", "T", "G", "C", "G", "A"]

# 🪐 نجوم: سیاره‌های منظومه‌ی شمسی به ترتیب فاصله از خورشید
planets = ["عطارد", "زهره", "زمین", "مریخ", "مشتری", "زحل", "اورانوس", "نپتون"]

# 🎲 لیست ترکیبی: نام، سن و قد یک شرکت‌کننده در آزمایش
participant = ["نیما", 15, 1.68]

# 📭 لیست خالی: بعداً آن را پر می‌کنیم
empty_list = []

print(brain_lobes)
print(one_byte)
print(participant)
print(empty_list)
````

<div dir="rtl" style="font-family: Vazirmatn, Vazir, Tahoma, 'Segoe UI', sans-serif; line-height: 2; font-size: 15px; background: #e8f5e9; color: #1b1b1b; border-right: 6px solid #2e7d32; border-radius: 12px; padding: 14px 20px; margin: 12px 0;">

### 💡 این <code dir="ltr">\u200c</code> در خروجی چیست؟

شاید در خروجی بالا به جای «پس‌سری» چیزی شبیه <code dir="ltr">'پس\u200cسری'</code> دیده باشی. نگران نباش! <code dir="ltr">\u200c</code> همان **نیم‌فاصله**‌ی فارسی است. وقتی پایتون یک **لیست** را چاپ می‌کند، نویسه‌های نامرئی را با کدشان نشان می‌دهد تا برنامه‌نویس بداند آنجا چیزی هست. متن تو کاملاً سالم است؛ اگر خود عضو را چاپ کنی (<code dir="ltr">print(brain_lobes[3])</code>)، آن را درست می‌بینی.

</div>

<div dir="rtl" style="font-family: Vazirmatn, Vazir, Tahoma, 'Segoe UI', sans-serif; line-height: 2; font-size: 15px; background: #e3f2fd; color: #1b1b1b; border-right: 6px solid #1565c0; border-radius: 12px; padding: 14px 20px; margin: 12px 0;">

### 💻 دانستنی کامپیوتری: بایت چیست؟

کامپیوتر همه‌چیز را با دو رقم <code dir="ltr">0</code> و <code dir="ltr">1</code> نگه می‌دارد. به هر کدام از این رقم‌ها **بیت** (bit) می‌گوییم. هشت بیت کنار هم یک **بایت** (byte) می‌سازند.

مثلاً بایت <code dir="ltr">[0, 1, 0, 0, 0, 0, 0, 1]</code> در کامپیوتر نشان‌دهنده‌ی حرف **A** است. در جلسه‌ی ۴ یاد می‌گیریم چطور این بیت‌ها را به عدد تبدیل کنیم!

</div>

<div dir="rtl" style="font-family: Vazirmatn, Vazir, Tahoma, 'Segoe UI', sans-serif; line-height: 2; font-size: 15px;">

## ۳. اندیس: شماره‌ی هر خانه

هر عضو لیست یک **شماره‌ی خانه** دارد که به آن **اندیس** (index) می‌گوییم. نکته‌ی بسیار مهم این است که در پایتون **شمارش از صفر شروع می‌شود**، نه از یک!

| عضو | <code dir="ltr">"پیشانی"</code> | <code dir="ltr">"آهیانه"</code> | <code dir="ltr">"گیجگاهی"</code> | <code dir="ltr">"پس‌سری"</code> |
|:---:|:---:|:---:|:---:|:---:|
| **اندیس مثبت** | <code dir="ltr">0</code> | <code dir="ltr">1</code> | <code dir="ltr">2</code> | <code dir="ltr">3</code> |
| **اندیس منفی** | <code dir="ltr">-4</code> | <code dir="ltr">-3</code> | <code dir="ltr">-2</code> | <code dir="ltr">-1</code> |

برای خواندن یک عضو، نام لیست را می‌نویسیم و اندیس را داخل کروشه می‌گذاریم:

</div>

````python
brain_lobes = ["پیشانی", "آهیانه", "گیجگاهی", "پس‌سری"]

print(brain_lobes[0])   # اولین عضو
print(brain_lobes[1])   # دومین عضو
print(brain_lobes[3])   # چهارمین (آخرین) عضو
````

<div dir="rtl" style="font-family: Vazirmatn, Vazir, Tahoma, 'Segoe UI', sans-serif; line-height: 2; font-size: 15px;">

### اندیس منفی: شمردن از آخر

اندیس منفی از **انتهای** لیست می‌شمارد. <code dir="ltr">-1</code> یعنی آخرین عضو و <code dir="ltr">-2</code> یعنی یکی مانده به آخر. این ترفند وقتی به درد می‌خورد که طول لیست را نمی‌دانیم ولی آخرین عضو را لازم داریم.

</div>

````python
planets = ["عطارد", "زهره", "زمین", "مریخ", "مشتری", "زحل", "اورانوس", "نپتون"]

print("دورترین سیاره:", planets[-1])
print("یکی مانده به آخر:", planets[-2])
print("خانه‌ی ما:", planets[2])
````

<div dir="rtl" style="font-family: Vazirmatn, Vazir, Tahoma, 'Segoe UI', sans-serif; line-height: 2; font-size: 15px; background: #fff3e0; color: #1b1b1b; border-right: 6px solid #ef6c00; border-radius: 12px; padding: 14px 20px; margin: 12px 0;">

### ⚠️ خطای رایج: اندیس خارج از محدوده

اگر اندیسی بخواهی که در لیست وجود ندارد، پایتون خطای <code dir="ltr">IndexError</code> می‌دهد. لیست <code dir="ltr">brain_lobes</code> چهار عضو دارد، پس بزرگ‌ترین اندیس آن <code dir="ltr">3</code> است، نه <code dir="ltr">4</code>.

سلول زیر را اجرا کن تا این خطا را از نزدیک ببینی. نترس؛ خواندن پیام خطا یکی از مهارت‌های مهم برنامه‌نویس‌هاست! 🕵️

</div>

````python
brain_lobes = ["پیشانی", "آهیانه", "گیجگاهی", "پس‌سری"]
print(brain_lobes[4])
````

<div dir="rtl" style="font-family: Vazirmatn, Vazir, Tahoma, 'Segoe UI', sans-serif; line-height: 2; font-size: 15px;">

## ۴. طول لیست با <code dir="ltr">len</code>

تابع <code dir="ltr">len</code> (کوتاه‌شده‌ی length به معنی «طول») تعداد اعضای لیست را برمی‌گرداند.

</div>

````python
eeg_channels = ["Fp1", "Fp2", "C3", "C4", "O1", "O2"]
one_byte = [0, 1, 0, 0, 0, 0, 0, 1]

print("تعداد کانال‌های EEG:", len(eeg_channels))
print("تعداد بیت‌های یک بایت:", len(one_byte))
print("طول لیست خالی:", len([]))
````

<div dir="rtl" style="font-family: Vazirmatn, Vazir, Tahoma, 'Segoe UI', sans-serif; line-height: 2; font-size: 15px; background: #e8f5e9; color: #1b1b1b; border-right: 6px solid #2e7d32; border-radius: 12px; padding: 14px 20px; margin: 12px 0;">

### 💡 رابطه‌ی طلایی

اگر طول لیستی <code dir="ltr">n</code> باشد، اندیس‌های مثبت آن از <code dir="ltr">0</code> تا <code dir="ltr">n - 1</code> هستند. پس آخرین عضو همیشه در <code dir="ltr">my_list[len(my_list) - 1]</code> است؛ که همان <code dir="ltr">my_list[-1]</code> است!

</div>

<div dir="rtl" style="font-family: Vazirmatn, Vazir, Tahoma, 'Segoe UI', sans-serif; line-height: 2; font-size: 15px;">

## ۵. تغییر دادن عضوهای لیست

لیست **تغییرپذیر** (mutable) است؛ یعنی بعد از ساختن آن می‌توانیم عضوهایش را عوض کنیم. کافی است با اندیس، مقدار جدیدی به یک خانه بدهیم.

🧠 **مثال:** فرض کن الکترود <code dir="ltr">C3</code> در حین ثبت جابه‌جا شده و آن را با الکترود <code dir="ltr">Cz</code> عوض کرده‌ایم:

</div>

````python
eeg_channels = ["Fp1", "Fp2", "C3", "C4", "O1", "O2"]
print("قبل:", eeg_channels)

eeg_channels[2] = "Cz"
print("بعد:", eeg_channels)
````

````python
# 💻 روشن کردن آخرین بیت یک بایت
one_byte = [0, 1, 0, 0, 0, 0, 0, 0]
one_byte[-1] = 1
print(one_byte)
````

<div dir="rtl" style="font-family: Vazirmatn, Vazir, Tahoma, 'Segoe UI', sans-serif; line-height: 2; font-size: 15px;">

## ۶. برش (Slicing): جدا کردن تکه‌ای از لیست

گاهی فقط بخشی از لیست را لازم داریم. با **برش** می‌توانیم یک تکه را جدا کنیم:

<div dir="ltr" align="left">

```python
my_list[start:stop]
```

</div>

- <code dir="ltr">start</code>: اندیسی که برش **از آن شروع** می‌شود (خودش داخل برش هست).
- <code dir="ltr">stop</code>: اندیسی که برش **قبل از آن تمام** می‌شود (خودش داخل برش **نیست**).

📌 اگر <code dir="ltr">start</code> را ننویسیم، از اول لیست شروع می‌شود و اگر <code dir="ltr">stop</code> را ننویسیم، تا آخر لیست ادامه می‌یابد.

</div>

````python
# ولتاژ ثبت‌شده در ۱۰ لحظه‌ی پشت سر هم (میکروولت)
eeg = [5, 12, 18, 9, -4, -11, -6, 3, 14, 20]

print("سه نمونه‌ی اول:", eeg[0:3])
print("از اندیس ۴ تا ۶:", eeg[4:7])
print("از اندیس ۷ تا آخر:", eeg[7:])
print("همه به جز دو تای آخر:", eeg[:-2])
print("یک کپی کامل:", eeg[:])
````

<div dir="rtl" style="font-family: Vazirmatn, Vazir, Tahoma, 'Segoe UI', sans-serif; line-height: 2; font-size: 15px;">

### گام (step)

برش می‌تواند عدد سومی هم داشته باشد که **گام** حرکت را تعیین می‌کند: <code dir="ltr">my_list[start:stop:step]</code>.

</div>

````python
eeg = [5, 12, 18, 9, -4, -11, -6, 3, 14, 20]

print("یکی در میان:", eeg[::2])
print("برعکس:", eeg[::-1])
````

<div dir="rtl" style="font-family: Vazirmatn, Vazir, Tahoma, 'Segoe UI', sans-serif; line-height: 2; font-size: 15px; background: #fce4ec; color: #1b1b1b; border-right: 6px solid #c2185b; border-radius: 12px; padding: 14px 20px; margin: 12px 0;">

### 🧠 علوم اعصاب: کاهش نرخ نمونه‌برداری

دانشمندان علوم اعصاب گاهی داده‌ی EEG را با سرعت خیلی بالا ثبت می‌کنند (مثلاً ۱۰۰۰ نمونه در ثانیه)، ولی برای بعضی تحلیل‌ها این‌همه نمونه لازم نیست. یکی از ساده‌ترین ایده‌ها این است که فقط **یکی در میان** نمونه‌ها را نگه دارند تا حجم داده نصف شود. به این کار **کاهش نرخ نمونه‌برداری** (downsampling) می‌گویند.

همین کاری که با <code dir="ltr">eeg[::2]</code> کردیم! البته در دنیای واقعی، پیش از این کار یک فیلتر هم روی سیگنال اعمال می‌کنند تا اطلاعات مهم از دست نرود.

</div>

````python
# 🧬 یک ترفند جالب: برعکس کردن یک رشته‌ی DNA
dna = ["A", "T", "G", "C", "G", "A"]
print("رشته‌ی اصلی:", dna)
print("رشته‌ی برعکس:", dna[::-1])
````

<div dir="rtl" style="font-family: Vazirmatn, Vazir, Tahoma, 'Segoe UI', sans-serif; line-height: 2; font-size: 15px;">

## ۷. آیا این عضو در لیست هست؟ (<code dir="ltr">in</code> و <code dir="ltr">not in</code>)

عملگر <code dir="ltr">in</code> بررسی می‌کند که یک مقدار در لیست وجود دارد یا نه. جواب آن <code dir="ltr">True</code> (درست) یا <code dir="ltr">False</code> (نادرست) است.

</div>

````python
eeg_channels = ["Fp1", "Fp2", "C3", "C4", "O1", "O2"]

print("O1" in eeg_channels)       # آیا الکترود O1 ثبت شده؟
print("Pz" in eeg_channels)       # آیا الکترود Pz ثبت شده؟
print("Pz" not in eeg_channels)   # آیا Pz ثبت نشده؟
````

<div dir="rtl" style="font-family: Vazirmatn, Vazir, Tahoma, 'Segoe UI', sans-serif; line-height: 2; font-size: 15px;">

## ۸. لیست درون لیست (یک نگاه کوتاه)

عضو یک لیست می‌تواند خودش یک لیست باشد! مثلاً داده‌های دو شرکت‌کننده در یک آزمایش زمان واکنش:

</div>

````python
# هر لیست داخلی: [نام، زمان واکنش اول، زمان واکنش دوم] بر حسب میلی‌ثانیه
experiment = [
    ["مریم", 310, 295],
    ["علی", 350, 330],
]

print(experiment[0])      # کل اطلاعات مریم
print(experiment[1][0])   # نام نفر دوم
print(experiment[0][2])   # دومین زمان واکنش مریم
````

### 📜 داستان تاریخ: ایدا لاولیس و چارلز ببیج

حدود ۲۰۰ سال پیش، ریاضی‌دان انگلیسی **چارلز ببیج** (Charles Babbage، ۱۷۹۱ تا ۱۸۷۱) طرح ماشینی به نام **موتور تحلیلی** (Analytical Engine) را کشید. این ماشین مکانیکی بود و با چرخ‌دنده کار می‌کرد، اما ایده‌هایش شبیه کامپیوترهای امروزی بود:

- بخشی به نام **انبار** (the Store) داشت که قرار بود هزار عدد را در ستون‌های شماره‌دار نگه دارد؛ تقریباً مثل یک لیست بزرگ با اندیس!
- بخشی به نام **آسیاب** (the Mill) داشت که محاسبه‌ها را انجام می‌داد؛ چیزی شبیه پردازنده‌ی امروزی.

**ایدا لاولیس** (Ada Lovelace، ۱۸۱۵ تا ۱۸۵۲)، ریاضی‌دان جوان انگلیسی، در سال ۱۸۴۳ یادداشت‌های مفصلی درباره‌ی این ماشین نوشت. در یکی از این یادداشت‌ها (یادداشت G) روشی گام‌به‌گام برای محاسبه‌ی دنباله‌ای از اعداد به نام **اعداد برنولی** توضیح داد. بسیاری او را **نخستین برنامه‌نویس جهان** می‌دانند.

نکته‌ی شگفت‌انگیز این است که ایدا حدس زده بود چنین ماشینی روزی می‌تواند به جای عدد با **نماد**‌ها کار کند و حتی **موسیقی** بسازد. امروز زبان برنامه‌نویسی **Ada** به افتخار او نام‌گذاری شده است. 👩‍💻


</div>

### 📝 خلاصه‌ی جلسه

| مفهوم | نمونه | توضیح |
|:---|:---|:---|
| ساختن لیست | <code dir="ltr">x = [3, 7, 1]</code> | با کروشه و ویرگول |
| اندیس مثبت | <code dir="ltr">x[0]</code> | شمارش از صفر |
| اندیس منفی | <code dir="ltr">x[-1]</code> | آخرین عضو |
| طول | <code dir="ltr">len(x)</code> | تعداد اعضا |
| تغییر عضو | <code dir="ltr">x[1] = 9</code> | لیست تغییرپذیر است |
