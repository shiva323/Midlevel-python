## ۱. مجموعه چیست؟

**مجموعه** (set) گروهی از عضوهاست که دو ویژگی مهم دارد:

1. **بدون تکرار:** هر عضو فقط **یک بار** در مجموعه می‌آید.
2. **بدون ترتیب:** عضوها جایگاه یا اندیس ندارند؛ پس عبارتی مثل <code dir="ltr">my_set[0]</code> معنا ندارد!

مجموعه را با **آکولاد** <code dir="ltr">{ }</code> می‌سازیم:

</div>

````python
# Neurotransmitters: chemicals that neurons use to send messages
neurotransmitters = {"dopamine", "serotonin", "glutamate", "GABA", "dopamine"}

print(neurotransmitters)        # the duplicate dopamine is gone!
print(len(neurotransmitters))
print(type(neurotransmitters))
````

## ۳. تغییر دادن مجموعه

| متد | کار | اگر عضو وجود نداشته باشد؟ |
|:---|:---|:---|
| <code dir="ltr">add(x)</code> | افزودن x | اگر از قبل باشد، اتفاقی نمی‌افتد. |
| <code dir="ltr">remove(x)</code> | حذف x | خطای <code dir="ltr">KeyError</code> می‌دهد. ❌ |
| <code dir="ltr">discard(x)</code> | حذف x | بی‌سروصدا هیچ کاری نمی‌کند. ✅ |

</div>

````python
recorded = {"Fp1", "Fp2", "C3", "C4"}

recorded.add("O1")
recorded.add("C3")          # already there; nothing changes
print(sorted(recorded))

recorded.discard("T7")      # T7 is not there, but there is no error
recorded.remove("Fp2")
print(sorted(recorded))
````

## ۴. بررسی عضویت با in: سریع مثل برق ⚡

عملگر <code dir="ltr">in</code> روی مجموعه هم کار می‌کند. نکته‌ی جالب این است که روی مجموعه **خیلی سریع‌تر** از لیست است!

- برای پیدا کردن چیزی در **لیست**، پایتون باید عضوها را **یکی‌یکی** نگاه کند. در لیستی با یک میلیون عضو، ممکن است یک میلیون بار مقایسه لازم باشد.
- مجموعه از روشی به نام **درهم‌سازی** (hashing) استفاده می‌کند و تقریباً **مستقیم** به جای درست می‌رود؛ مثل اینکه بدانی کتاب در کدام قفسه‌ی کتابخانه است.

بیا خودمان آزمایش کنیم: ⏱️

</div>

````python
import time

big_list = list(range(2_000_000))
big_set = set(big_list)
target = 1_999_999          # worst case for the list: the last item

start = time.perf_counter()
for _ in range(20):
    found = target in big_list
list_time = time.perf_counter() - start

start = time.perf_counter()
for _ in range(20):
    found = target in big_set
set_time = time.perf_counter() - start

print("Search time in the list:", round(list_time, 4), "s")
print("Search time in the set: ", round(set_time, 6), "s")
print("The set was about", int(list_time / set_time), "times faster!")
````

## ۵. عملیات مجموعه‌ها: ریاضی در پایتون

احتمالاً در ریاضی با **نمودار ون** آشنا شده‌ای. پایتون همه‌ی آن عملیات را دارد:

| عملیات | نماد پایتون | متد | معنی |
|:---|:---:|:---|:---|
| اجتماع (∪) | `A \| B` | <code dir="ltr">A.union(B)</code> | عضوهایی که **دست‌کم در یکی** هستند |
| اشتراک (∩) | <code dir="ltr">A &amp; B</code> | <code dir="ltr">A.intersection(B)</code> | عضوهایی که **در هر دو** هستند |
| تفاضل (−) | <code dir="ltr">A - B</code> | <code dir="ltr">A.difference(B)</code> | عضوهایی که **در A هستند ولی در B نه** |
| تفاضل متقارن (△) | <code dir="ltr">A ^ B</code> | <code dir="ltr">A.symmetric_difference(B)</code> | عضوهایی که **فقط در یکی** هستند |

🔎 **مثال:** اگر <code dir="ltr">A = {1, 2, 3, 4}</code> و <code dir="ltr">B = {3, 4, 5}</code> باشد:

- حاصل اجتماع <code dir="ltr">{1, 2, 3, 4, 5}</code> است.
- حاصل اشتراک <code dir="ltr">{3, 4}</code> است.
- حاصل تفاضل A منهای B برابر <code dir="ltr">{1, 2}</code> است.
- حاصل تفاضل متقارن <code dir="ltr">{1, 2, 5}</code> است.

</div>

### 🧠 علوم اعصاب: کدام نورون‌ها در هر دو کار فعال‌اند؟

پژوهشگران فعالیت گروهی از نورون‌ها را در دو کار ثبت کرده‌اند: **کار الف** (دیدن چهره) و **کار ب** (شنیدن صدای آشنا). هر نورون یک شماره دارد. سؤال‌های جالب این‌ها هستند:

- کدام نورون‌ها در **هر دو** کار فعال بوده‌اند؟ احتمالاً این نورون‌ها در **شناختن آدم‌ها** نقش دارند!
- کدام نورون‌ها **فقط** به چهره پاسخ داده‌اند؟
- روی هم رفته، چند نورون **دست‌کم در یک کار** فعال بوده‌اند؟

🤯 **دانستنی:** در سال ۲۰۰۵، پژوهشگران در مغز چند بیمار نورون‌هایی پیدا کردند که فقط به **یک آدم مشهور خاص** پاسخ می‌دادند؛ چه عکسش را می‌دیدند و چه نامش را می‌خواندند! بعدها به این نورون‌ها **سلول‌های مفهوم** (concept cells) گفتند.

</div>

````python
face_task  = {3, 7, 12, 15, 21, 28, 33, 40}
voice_task = {7, 9, 15, 21, 25, 40, 44}

both      = face_task & voice_task
any_task  = face_task | voice_task
face_only = face_task - voice_task
one_only  = face_task ^ voice_task

print("Active in both tasks:    ", sorted(both))
print("Active in at least one:  ", sorted(any_task), "-> count:", len(any_task))
print("Only in the face task:   ", sorted(face_only))
print("In exactly one task:     ", sorted(one_only))
````

````python
# Channels needed for the analysis vs. channels that were actually recorded
required = {"Fp1", "Fp2", "F3", "F4", "C3", "C4", "P3", "P4", "O1", "O2"}
recorded = {"Fp1", "Fp2", "F3", "C3", "C4", "P3", "P4", "O1", "O2", "EOG"}

missing = required - recorded
extra = recorded - required

print("Missing channels:", sorted(missing))
print("Extra channels:  ", sorted(extra))

if len(missing) == 0:
    print("All required channels were recorded.")
else:
    print("Before the analysis we must deal with", len(missing), "missing channel(s).")
````

## ۶. زیرمجموعه و ابرمجموعه

- عبارت <code dir="ltr">A &lt;= B</code> (یا <code dir="ltr">A.issubset(B)</code>) می‌پرسد: آیا **همه‌ی** عضوهای A در B هستند؟
- عبارت <code dir="ltr">A &gt;= B</code> (یا <code dir="ltr">A.issuperset(B)</code>) می‌پرسد: آیا A **همه‌ی** عضوهای B را دارد؟
- عبارت <code dir="ltr">A.isdisjoint(B)</code> می‌پرسد: آیا A و B **هیچ** عضو مشترکی ندارند؟

</div>

````python
# A pangram is a sentence that uses every letter of the English alphabet
alphabet = set("abcdefghijklmnopqrstuvwxyz")

sentences = [
    "The quick brown fox jumps over the lazy dog",
    "Python is fun",
    "Pack my box with five dozen liquor jugs",
]

for s in sentences:
    letters = set(s.lower())
    if alphabet <= letters:
        print("Pangram:     ", s)
    else:
        missing = alphabet - letters
        print("Not a pangram:", s, "-> missing letters:", len(missing))
````

````python
# Check that a username only uses allowed characters
allowed = set("abcdefghijklmnopqrstuvwxyz0123456789_")

usernames = ["neuro_kid", "Shiva!", "brain2026", "hello world"]

for u in usernames:
    if set(u) <= allowed:
        print("OK:    ", u)
    else:
        bad = set(u) - allowed
        print("Not OK:", u, "-> forbidden characters:", sorted(bad))
````

## ۷. حلقه روی مجموعه

می‌توانیم با <code dir="ltr">for</code> روی مجموعه حلقه بزنیم، اما ترتیب **تضمین‌شده نیست**. اگر ترتیب مهم است، اول با <code dir="ltr">sorted</code> مرتبش کن.

</div>

````python
friends_of_sara = {"Nima", "Elham", "Kian", "Roya"}
friends_of_reza = {"Kian", "Mani", "Roya", "Bahar"}

mutual = friends_of_sara & friends_of_reza
print("Mutual friends of Sara and Reza:")
for friend in sorted(mutual):
    print("  -", friend)

suggestions = friends_of_reza - friends_of_sara
print("Friend suggestions for Sara:", sorted(suggestions))
````

### 📜 داستان تاریخ: کانتور، بی‌نهایت‌ها و پایگاه داده‌ها

**گئورگ کانتور** (Georg Cantor، ۱۸۴۵ تا ۱۹۱۸)، ریاضی‌دان آلمانی، در دهه‌های ۱۸۷۰ و ۱۸۸۰ **نظریه‌ی مجموعه‌ها** را پایه‌گذاری کرد. او با ایده‌ای ساده اما انقلابی کار کرد: دو مجموعه **هم‌اندازه‌اند** اگر بتوان عضوهایشان را **یک‌به‌یک** با هم جفت کرد.

با همین ایده، کانتور به نتیجه‌ای حیرت‌انگیز رسید: **همه‌ی بی‌نهایت‌ها هم‌اندازه نیستند!** ♾️ مثلاً بی‌نهایتِ **عددهای حقیقی** (همه‌ی عددهای اعشاری) از بی‌نهایتِ **عددهای طبیعی** (۱، ۲، ۳ و ...) **بزرگ‌تر** است.

بسیاری از ریاضی‌دانان آن زمان ایده‌هایش را نپذیرفتند، اما امروز نظریه‌ی مجموعه‌ها **زبان پایه‌ی تقریباً همه‌ی ریاضیات** است. ریاضی‌دان بزرگ، **داوید هیلبرت**، گفته بود: «هیچ‌کس ما را از بهشتی که کانتور برایمان ساخته بیرون نخواهد کرد.»

### ✏️ تمرین ۴ (چالشی): شبکه‌ی مغز با مجموعه‌ها

🧠 در جلسه‌ی ۷، اتصال‌های مغز را با یک ماتریس نشان دادیم. راه دیگر این است که برای هر ناحیه، **مجموعه‌ی همسایه‌هایش** را نگه داریم. با داده‌ی زیر:

1. همسایه‌های **مشترک** ناحیه‌ی frontal و ناحیه‌ی parietal را پیدا کن.
2. برای هر **جفت** ناحیه (بدون تکرار)، تعداد همسایه‌های مشترکشان را چاپ کن.
3. 🌟 ناحیه‌هایی که با ناحیه‌ی occipital **مستقیماً** وصل نیستند، اما **همسایه‌ی مشترک** با آن دارند، کدام‌اند؟ این همان ایده‌ی «پیشنهاد دوستی» است!

</div>

````python
region_names = ["frontal", "parietal", "temporal", "occipital", "hippocampus"]
neighbors = [
    {"parietal", "temporal", "hippocampus"},              # frontal
    {"frontal", "temporal", "occipital"},                 # parietal
    {"frontal", "parietal", "occipital", "hippocampus"},  # temporal
    {"parietal", "temporal"},                             # occipital
    {"frontal", "temporal"},                              # hippocampus
]
# Write your code here
````

```python
print("Shared by frontal and parietal:", neighbors[0] & neighbors[1])   # {'temporal'}

for i in range(len(region_names)):
    for j in range(i + 1, len(region_names)):
        common = neighbors[i] & neighbors[j]
        print(region_names[i], "and", region_names[j], "->", len(common))

target = region_names.index("occipital")
suggestions = set()
for i in range(len(region_names)):
    if i == target:
        continue
    name = region_names[i]
    if name in neighbors[target]:
        continue                       # already directly connected
    if len(neighbors[i] & neighbors[target]) > 0:
        suggestions.add(name)
print("Suggestions:", sorted(suggestions))        # frontal, hippocampus
```

### 📝 خلاصه‌ی جلسه

- مجموعه **بدون تکرار** و **بدون ترتیب** است؛ مثل <code dir="ltr">s = {1, 2, 3}</code>. مجموعه‌ی خالی را با <code dir="ltr">set()</code> می‌سازیم.
- تابع <code dir="ltr">set</code> تکراری‌های یک لیست را حذف می‌کند.
- متد <code dir="ltr">add</code> برای افزودن است؛ <code dir="ltr">remove</code> (با خطا) و <code dir="ltr">discard</code> (بدون خطا) برای حذف.
- عملگر <code dir="ltr">in</code> روی مجموعه **خیلی سریع‌تر** از لیست است.
- نمادهای <code dir="ltr">|</code>، <code dir="ltr">&amp;</code>، <code dir="ltr">-</code> و <code dir="ltr">^</code> به ترتیب اجتماع، اشتراک، تفاضل و تفاضل متقارن هستند.
- عبارت <code dir="ltr">A &lt;= B</code> یعنی A زیرمجموعه‌ی B است.
