## ۱. مجموعه چیست؟

**مجموعه** (<code dir="ltr">set</code>) گروهی از عضوهاست که دو ویژگی مهم دارد:

1. **بدون تکرار:** هر عضو فقط **یک بار** در مجموعه می‌آید.
2. **بدون ترتیب:** عضوها جایگاه یا اندیس ندارند؛ پس <code dir="ltr">my_set[0]</code> معنا ندارد!

مجموعه را با **آکولاد** <code dir="ltr">{ }</code> می‌سازیم:

</div>

````python
# 🧠 انتقال‌دهنده‌های عصبی (مواد شیمیایی که نورون‌ها با آن‌ها پیام می‌فرستند)
neurotransmitters = {"دوپامین", "سروتونین", "گلوتامات", "گابا", "دوپامین"}

print(neurotransmitters)        # دوپامینِ تکراری حذف شده است!
print(len(neurotransmitters))
print(type(neurotransmitters))
````

### 💡 ترتیب چاپ را جدی نگیر

## ۲. حذف تکراری‌ها: پرکاربردترین ترفند مجموعه

تابع <code dir="ltr">set(...)</code> هر لیست یا رشته‌ای را به مجموعه تبدیل می‌کند و تکراری‌ها خودبه‌خود حذف می‌شوند.

## ۳. تغییر دادن مجموعه

| متد | کار | اگر عضو وجود نداشته باشد؟ |
|:---|:---|:---|
| <code dir="ltr">add(x)</code> | افزودن <code dir="ltr">x</code> | — (اگر از قبل باشد، اتفاقی نمی‌افتد) |
| <code dir="ltr">remove(x)</code> | حذف <code dir="ltr">x</code> | ❌ خطای <code dir="ltr">KeyError</code> |
| <code dir="ltr">discard(x)</code> | حذف <code dir="ltr">x</code> | ✅ بی‌سروصدا هیچ کاری نمی‌کند |

</div>

````python
recorded = {"Fp1", "Fp2", "C3", "C4"}

recorded.add("O1")
recorded.add("C3")          # از قبل بود؛ چیزی عوض نمی‌شود
print(recorded)

recorded.discard("T7")      # T7 وجود ندارد، اما خطایی هم نمی‌دهد
recorded.remove("Fp2")
print(recorded)
````

## ۴. بررسی عضویت با <code dir="ltr">in</code>: سریع مثل برق ⚡

<code dir="ltr">in</code> روی مجموعه هم کار می‌کند. نکته‌ی جالب این است که روی مجموعه **خیلی سریع‌تر** از لیست است!

- برای پیدا کردن چیزی در **لیست**، پایتون باید عضوها را **یکی‌یکی** نگاه کند. در لیستی با یک میلیون عضو، ممکن است یک میلیون بار مقایسه لازم باشد.
- مجموعه از روشی به نام **درهم‌سازی** (hashing) استفاده می‌کند و تقریباً **مستقیم** به جای درست می‌رود؛ مثل اینکه بدانی کتاب در کدام قفسه‌ی کتابخانه است.

بیا خودمان آزمایش کنیم: ⏱️

</div>

````python
import time

big_list = list(range(2_000_000))
big_set = set(big_list)
target = 1_999_999          # بدترین حالت برای لیست: عضو آخر

start = time.perf_counter()
for _ in range(20):
    found = target in big_list
list_time = time.perf_counter() - start

start = time.perf_counter()
for _ in range(20):
    found = target in big_set
set_time = time.perf_counter() - start

print("زمان جست‌وجو در لیست:  ", round(list_time, 4), "ثانیه")
print("زمان جست‌وجو در مجموعه:", round(set_time, 6), "ثانیه")
print("مجموعه حدود", int(list_time / set_time), "برابر سریع‌تر بود! 🚀")
````

## ۵. عملیات مجموعه‌ها: ریاضی در پایتون

احتمالاً در ریاضی با **نمودار ون** آشنا شده‌ای. پایتون همه‌ی آن عملیات را دارد:

| عملیات | نماد | متد | معنی |
|:---|:---:|:---|:---|
| اجتماع (∪) | `A \| B` | <code dir="ltr">A.union(B)</code> | عضوهایی که **دست‌کم در یکی** هستند |
| اشتراک (∩) | <code dir="ltr">A &amp; B</code> | <code dir="ltr">A.intersection(B)</code> | عضوهایی که **در هر دو** هستند |
| تفاضل (−) | <code dir="ltr">A - B</code> | <code dir="ltr">A.difference(B)</code> | عضوهایی که **در A هستند ولی در B نه** |
| تفاضل متقارن (△) | <code dir="ltr">A ^ B</code> | <code dir="ltr">A.symmetric_difference(B)</code> | عضوهایی که **فقط در یکی** هستند |

🔎 **مثال:** اگر <code dir="ltr">A = {1, 2, 3, 4}</code> و <code dir="ltr">B = {3, 4, 5}</code> باشد:

- <code dir="ltr">A | B</code> ← <code dir="ltr">{1, 2, 3, 4, 5}</code>
- <code dir="ltr">A &amp; B</code> ← <code dir="ltr">{3, 4}</code>
- <code dir="ltr">A - B</code> ← <code dir="ltr">{1, 2}</code>
- <code dir="ltr">A ^ B</code> ← <code dir="ltr">{1, 2, 5}</code>

</div>

<div dir="rtl" style="font-family: Vazirmatn, Vazir, Tahoma, 'Segoe UI', sans-serif; line-height: 2; font-size: 15px; background: #fce4ec; color: #1b1b1b; border-right: 6px solid #c2185b; border-radius: 12px; padding: 14px 20px; margin: 12px 0;">

### 🧠 علوم اعصاب: کدام نورون‌ها در هر دو کار فعال‌اند؟

پژوهشگران فعالیت گروهی از نورون‌ها را در دو کار ثبت کرده‌اند: **کار الف** (دیدن چهره) و **کار ب** (شنیدن صدای آشنا). هر نورون یک شماره دارد. سؤال‌های جالب این‌هاست:

- کدام نورون‌ها در **هر دو** کار فعال بوده‌اند؟ (احتمالاً در **شناختن آدم‌ها** نقش دارند!)
- کدام نورون‌ها **فقط** به چهره پاسخ داده‌اند؟
- روی هم رفته، چند نورون **دست‌کم در یک کار** فعال بوده‌اند؟

🤯 **دانستنی:** در سال ۲۰۰۵، پژوهشگران در مغز چند بیمار نورون‌هایی پیدا کردند که فقط به **یک آدم مشهور خاص** پاسخ می‌دادند؛ چه عکسش را می‌دیدند و چه نامش را می‌خواندند! بعدها به این نورون‌ها **سلول‌های مفهوم** (concept cells) گفتند.

</div>

````python
face_task  = {3, 7, 12, 15, 21, 28, 33, 40}
voice_task = {7, 9, 15, 21, 25, 40, 44}

both      = face_task & voice_task
any_task  = face_task | voice_task
face_only = face_task - voice_task
one_only  = face_task ^ voice_task

print("فعال در هر دو کار:      ", sorted(both))
print("فعال دست‌کم در یک کار:  ", sorted(any_task), "← تعداد:", len(any_task))
print("فقط در کار چهره:        ", sorted(face_only))
print("فقط در یکی از دو کار:   ", sorted(one_only))
````

````python
# 🧠 کانال‌های لازم برای تحلیل در برابر کانال‌هایی که واقعاً ثبت شده‌اند
required = {"Fp1", "Fp2", "F3", "F4", "C3", "C4", "P3", "P4", "O1", "O2"}
recorded = {"Fp1", "Fp2", "F3", "C3", "C4", "P3", "P4", "O1", "O2", "EOG"}

missing = required - recorded
extra = recorded - required

print("کانال‌های جاافتاده:", missing)
print("کانال‌های اضافه:", extra)

if len(missing) == 0:
    print("✅ همه‌ی کانال‌های لازم ثبت شده‌اند.")
else:
    print("⚠️ پیش از تحلیل باید تکلیف", len(missing), "کانال جاافتاده روشن شود.")
````

## ۶. زیرمجموعه و ابرمجموعه

- <code dir="ltr">A &lt;= B</code> یا <code dir="ltr">A.issubset(B)</code>: آیا **همه‌ی** عضوهای A در B هستند؟
- <code dir="ltr">A &gt;= B</code> یا <code dir="ltr">A.issuperset(B)</code>: آیا A **همه‌ی** عضوهای B را دارد؟
- <code dir="ltr">A.isdisjoint(B)</code>: آیا A و B **هیچ** عضو مشترکی ندارند؟

</div>

````python
# 🔤 جمله‌ی «همه‌حرفی» (pangram): جمله‌ای که همه‌ی حروف الفبای انگلیسی را دارد
alphabet = set("abcdefghijklmnopqrstuvwxyz")

sentences = [
    "The quick brown fox jumps over the lazy dog",
    "Python is fun",
    "Pack my box with five dozen liquor jugs",
]

for s in sentences:
    letters = set(s.lower())
    if alphabet <= letters:
        print("✅ همه‌حرفی است:", s)
    else:
        missing = alphabet - letters
        print("❌ همه‌حرفی نیست:", s, "← تعداد حروف جاافتاده:", len(missing))
````

````python
# 🔐 بررسی کاراکترهای مجاز در نام کاربری
allowed = set("abcdefghijklmnopqrstuvwxyz0123456789_")

usernames = ["neuro_kid", "Shiva!", "brain2026", "hello world"]

for u in usernames:
    if set(u) <= allowed:
        print("✅", u)
    else:
        bad = set(u) - allowed
        print("❌", u, "← کاراکترهای غیرمجاز:", bad)
````

## ۷. حلقه روی مجموعه

می‌توانیم با <code dir="ltr">for</code> روی مجموعه حلقه بزنیم، اما ترتیب **تضمین‌شده نیست**. اگر ترتیب مهم است، اول <code dir="ltr">sorted</code> کن.

</div>

````python
friends_of_sara = {"نیما", "الهام", "کیان", "رویا"}
friends_of_reza = {"کیان", "مانی", "رویا", "بهار"}

mutual = friends_of_sara & friends_of_reza
print("دوستان مشترک سارا و رضا:")
for friend in sorted(mutual):
    print("  🤝", friend)

suggestions = friends_of_reza - friends_of_sara
print("پیشنهاد دوستی برای سارا:", sorted(suggestions))
````

## ۸. لیست یا مجموعه؟

| ویژگی | لیست <code dir="ltr">[ ]</code> | مجموعه <code dir="ltr">{ }</code> |
|:---|:---:|:---:|
| ترتیب دارد؟ | ✅ | ❌ |
| اندیس دارد؟ (<code dir="ltr">x[0]</code>) | ✅ | ❌ |
| تکرار مجاز است؟ | ✅ | ❌ |
| سرعت <code dir="ltr">in</code> | 🐢 کند (برای داده‌ی بزرگ) | ⚡ خیلی سریع |
| اجتماع، اشتراک و تفاضل | ❌ | ✅ |
| می‌تواند لیست را عضو خودش کند؟ | ✅ | ❌ |

📌 **قانون سرانگشتی:** اگر **ترتیب** یا **تکرار** مهم است، لیست. اگر **یکتا بودن** یا **عضویت سریع** مهم است، مجموعه.

### 📜 داستان تاریخ: کانتور، بی‌نهایت‌ها و پایگاه داده‌ها

**گئورگ کانتور** (Georg Cantor، ۱۸۴۵ تا ۱۹۱۸)، ریاضی‌دان آلمانی، در دهه‌های ۱۸۷۰ و ۱۸۸۰ **نظریه‌ی مجموعه‌ها** را پایه‌گذاری کرد. او با ایده‌ای ساده اما انقلابی کار کرد: دو مجموعه **هم‌اندازه‌اند** اگر بتوان عضوهایشان را **یک‌به‌یک** با هم جفت کرد.

با همین ایده، کانتور به نتیجه‌ای حیرت‌انگیز رسید: **همه‌ی بی‌نهایت‌ها هم‌اندازه نیستند!** ♾️ مثلاً بی‌نهایتِ **عددهای حقیقی** (همه‌ی عددهای اعشاری) از بی‌نهایتِ **عددهای طبیعی** (۱، ۲، ۳، ...) **بزرگ‌تر** است. بسیاری از ریاضی‌دانان آن زمان ایده‌هایش را نپذیرفتند، اما امروز نظریه‌ی مجموعه‌ها **زبان پایه‌ی تقریباً همه‌ی ریاضیات** است. ریاضی‌دان بزرگ، **داوید هیلبرت**، گفته بود: «هیچ‌کس ما را از بهشتی که کانتور برایمان ساخته بیرون نخواهد کرد.»

🗄️ حدود صد سال بعد، **ادگار کاد** (Edgar F. Codd، ۱۹۲۳ تا ۲۰۰۳)، دانشمند انگلیسی که در شرکت **IBM** کار می‌کرد، در سال ۱۹۷۰ **مدل رابطه‌ای** پایگاه داده‌ها را معرفی کرد. ایده‌ی او بر پایه‌ی **نظریه‌ی مجموعه‌ها** بود: هر جدول، مجموعه‌ای از سطرهاست و پرس‌وجوها با عملیاتی مثل اجتماع، اشتراک و تفاضل ساخته می‌شوند. زبان **SQL** که امروز تقریباً همه‌ی بانک‌ها، فروشگاه‌های اینترنتی و اپلیکیشن‌ها از آن استفاده می‌کنند، از دل همین ایده بیرون آمد. کاد در سال ۱۹۸۱ **جایزه‌ی تورینگ** را دریافت کرد.

### ✏️ تمرین ۳ (متوسط): باشگاه‌های مدرسه

سه باشگاه در مدرسه داریم. با عملیات مجموعه‌ها به این سؤال‌ها جواب بده:

1. چه کسانی در **هر سه** باشگاه عضوند؟
2. چه کسانی **فقط** در باشگاه رباتیک عضوند؟
3. در مجموع چند دانش‌آموز **دست‌کم در یک** باشگاه عضوند؟
4. 🌟 چه کسانی **دقیقاً در دو** باشگاه عضوند؟

</div>

````python
robotics  = {"آرش", "بهاره", "پرهام", "تینا", "ژیلا"}
astronomy = {"بهاره", "تینا", "جواد", "سپیده"}
biology   = {"تینا", "پرهام", "سپیده", "شایان", "بهاره"}
# کد خودت را اینجا بنویس ✍️
````

astronomy = {"بهاره", "تینا", "جواد", "سپیده"}
biology   = {"تینا", "پرهام", "سپیده", "شایان", "بهاره"}

all_three = robotics & astronomy & biology

    for club in [robotics, astronomy, biology]:
        if person in club:
            count += 1
    if count == 2:
        exactly_two.add(person)

### ✏️ تمرین ۴ (چالشی): شبکه‌ی مغز با مجموعه‌ها

🧠 در جلسه‌ی ۷، اتصال‌های مغز را با یک ماتریس نشان دادیم. راه دیگر این است که برای هر ناحیه، **مجموعه‌ی همسایه‌هایش** را نگه داریم. با داده‌ی زیر:

1. همسایه‌های **مشترک** «پیشانی» و «آهیانه» را پیدا کن.
2. برای هر **جفت** ناحیه (بدون تکرار)، تعداد همسایه‌های مشترکشان را چاپ کن.
3. 🌟 ناحیه‌هایی که با «پس‌سری» **مستقیماً** وصل نیستند، اما **همسایه‌ی مشترک** با آن دارند، کدام‌اند؟ (این همان ایده‌ی «پیشنهاد دوستی» است!)

</div>

````python
region_names = ["پیشانی", "آهیانه", "گیجگاهی", "پس‌سری", "هیپوکامپ"]
neighbors = [
    {"آهیانه", "گیجگاهی", "هیپوکامپ"},     # پیشانی
    {"پیشانی", "گیجگاهی", "پس‌سری"},       # آهیانه
    {"پیشانی", "آهیانه", "پس‌سری", "هیپوکامپ"},  # گیجگاهی
    {"آهیانه", "گیجگاهی"},                # پس‌سری
    {"پیشانی", "گیجگاهی"},                # هیپوکامپ
]
# کد خودت را اینجا بنویس ✍️
````

<div dir="rtl" style="font-family: Vazirmatn, Vazir, Tahoma, 'Segoe UI', sans-serif; line-height: 2; font-size: 15px;">

<summary><b>👀 دیدن پاسخ پیشنهادی</b></summary>

<div dir="ltr" align="left">

```python
print("مشترک پیشانی و آهیانه:", neighbors[0] & neighbors[1])   # {'گیجگاهی'}

for i in range(len(region_names)):
    for j in range(i + 1, len(region_names)):
        common = neighbors[i] & neighbors[j]
        print(region_names[i], "و", region_names[j], "→", len(common))

target = region_names.index("پس‌سری")
suggestions = set()
for i in range(len(region_names)):
    if i == target:
        continue
    name = region_names[i]
    if name in neighbors[target]:
        continue                       # از قبل مستقیماً وصل است
    if len(neighbors[i] & neighbors[target]) > 0:
        suggestions.add(name)
print("پیشنهادها:", suggestions)        # پیشانی و هیپوکامپ
```

### 📝 خلاصه‌ی جلسه

- مجموعه **بدون تکرار** و **بدون ترتیب** است: <code dir="ltr">s = {1, 2, 3}</code>. مجموعه‌ی خالی: <code dir="ltr">set()</code>.
